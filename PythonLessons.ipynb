# 🐍 Python 学习笔记｜前瞻：厨房理论与命令行生存

---

## 1. 核心概念：厨房理论 (Environment)

### 1.1 全局环境 vs. 虚拟环境
* **全局环境（大厨房）**：电脑安装 Python 后的主厨房。所有项目共用，容易把“调料”（第三方库）搞混，导致环境崩溃。
* **虚拟环境 (venv)（私密小厨位）**：
    * **比喻**：用纸板在大厨房里隔出来的专属空间。
    * **意义**：项目 A 用老抽（旧版库），项目 B 用生抽（新版库），互不干扰。
* **核心操作**：
    * **创建**：`python -m venv .venv` （盖一个小隔间）。
    * **激活**：`.venv\Scripts\activate` （走进隔间，接管灶台）。

### 1.2 库、包、pip
* **库 (Library)**：功能的集合。例如“盖城堡”或“绘图”的代码逻辑。
* **包 (Package)**：代码的物理形态。在电脑里表现为一个文件夹。
* **pip（快递员/钱包管理器）**：
    * **本质**：运行在命令行里的工具软件。
    * **任务**：你下单 `pip install`，它负责从 **PyPI**（全球中央大仓库）把你需要的“盒子（包）”搬回你的小厨位。

---

## 2. 命令行生存：地图与导航 (CLI & Path)

### 2.1 环境变量 PATH
* **比喻**：操作系统的“快速搜索清单”。
* **原理**：当你输入一个指令（如 `python` 或 `adb`）时：
    1. 电脑先看脚下（当前所在的文件夹）。
    2. 如果没找到，再翻清单（**PATH**）。
    3. 清单里有地址，它就直接去那儿执行；清单里没有，就会报错“不是内部或外部命令”。
* **建议**：把常用工具（如 adb.exe）所在的路径加入 PATH，从此告别手动“指路”。

### 2.2 路径符号（指路牌）
* **`.` (点)**：**“就在脚下”**。代表当前所在的文件夹。
    * `.\adb`：告诉电脑“别去清单找了，就在这儿跑这个文件”。
* **`..` (双点)**：**“退回门口”**。代表上一层文件夹。

### 2.3 黄金按键：Tab 键
* **功能**：自动补全。
* **心法**：打出前两个字母就按 **Tab**。
    * **补全成功**：说明路径是对的，文件确实在那儿。
    * **补全失败**：说明电脑“没看见”这个文件，别硬敲，先检查路径。

---
> **💡 预习总结**：学编程不仅仅是写代码，掌握环境隔离（虚拟环境）和系统指路（PATH）能让你少走 80% 的弯路。

# 🐍 Python 学习笔记｜第四关：交互与类型转换 (Input & Casting)

---

## 1. 核心概念
* **交互 (`input`)**：获取用户的指令。这是程序与人沟通的唯一窗口。
* **转换 (`Casting`)**：把获取的内容转换成正确的格式。
> **必经之路**：在你学习 `if` 判断和 `while` 循环之前，必须掌握如何处理用户输入的数据，否则程序会因为“性格不合”（类型错误）而崩溃。

### 实例演示
```python
user_input = input("请输入你的幸运数字：")  #input()函数返回的数据类型是字符串
print("你输入的数据类型是：", type(user_input)) 
lucky_number = int(user_input) #强制转换：不然没法做数学题
print(f"你的数字加 10 以后是：{lucky_number + 10}")
```
## 2. 知识点

### 2.1 int() 函数：类型转换大法
用于将其他类型强制转换为**整数**。

* **(1) 转换数字类型**：
    * **浮点数**：直接截断小数（不进位）。如 `int(3.9)` → `3`。
    * **布尔值**：`int(True)` → `1`，`int(False)` → `0`。
* **(2) 转换字符串**：
    * 必须是合法的整数样式。`int("12.5")` 或 `int("abc")` 都会**报错**。
* **(3) 进制转换**：
    * `int("1010", 2)` → `10`（二进制转十进制）。
* **(4) 空转换**：
    * `int()` → `0`。

### 2.2 f-string：格式化神器
在字符串前加 `f`，通过 `{}` 嵌入变量或表达式。

* **基础用法**：`f"你好，{name}"`
* **表达式运算**：`f"计算结果：{2 + 3}"` → `计算结果：5`
* **格式化控制**：
    * **保留小数**：`{pi:.2f}` （保留 2 位小数，四舍五入）。
    * **补零对齐**：`{num:0>5}` （宽度为 5，不足补 0）。
* **自查技巧**：`f"{score=}"` → 输出 `score=95`（调试神器）。
* **注意事项**：
    * **引号冲突**：内部字符串用单引号，外部用双引号。
    * **转义**：想显示大括号请用双层 `{{}}`。

# 🐍 Python 学习笔记｜第五关：逻辑判断 (If-Else)

---

## 1. 核心概念
* **逻辑判断**：让程序根据不同的条件执行不同的代码块。这是程序实现“智能化”的基础。
* **缩进 (Indentation)**：Python 的强制要求。`if`、`elif`、`else` 后面的执行代码必须整体向右缩进（通常 4 个空格）。

### 实例演示
```python
limit = 18
age_input = input("请输入你的年龄：")
age = int(age_input)  # 类型转换：将字符串转为整数

if age >= limit:
    print("验证通过") 
    print("欢迎进入成年人的编程世界")
elif age >= 12:
    print("你是青少年，建议在家长的陪同下学习")
else:
    print("小朋友，先去喝杯牛奶再来吧")

print("判断结束") # 无缩进，无论结果如何都会执行
```

---

## 2. 知识点

### 2.1 比较运算符
用于比较两个值，结果为布尔值（`True` 或 `False`）。

| 符号 | 描述 | 示例 |
| :--- | :--- | :--- |
| `>` | 大于 | `5 > 3` 为 True |
| `<` | 小于 | `5 < 3` 为 False |
| `>=` | 大于等于 | `5 >= 5` 为 True |
| `<=` | 小于等于 | `5 <= 4` 为 False |
| **`==`** | **等于** | **必考点**：一个 `=` 是赋值，两个 `==` 才是比较！ |
| `!=` | 不等于 | `5 != 3` 为 True |

### 2.2 增强赋值运算符
一种减少代码重复的缩写形式。

* **`+=`**：`a += 1` 等同于 `a = a + 1`
* **`-=`**：`a -= 1` 等同于 `a = a - 1`
* **`*=`**：`a *= 2` 等同于 `a = a * 2`
* **`/=`**：`a /= 2` 等同于 `a = a / 2`
> **注意**：Python 中几乎所有的数学运算（如 `**`、`//`）都可以这样缩写。

### 2.3 判断结构逻辑
1. **`if`**：第一个条件判断。
2. **`elif`**：上个条件不满足时，再判断这个。可以有多个 `elif`。
3. **`else`**：当上面所有条件都不行时，执行这个“最终保底”项。

---
> **⚡️ 避坑指南**：在 Python 中，冒号 `:` 是 `if` 结构的标志，漏掉它会引发语法错误。

````python
#练习
score_input  = input("请输入你的分数：")
score = int(score_input)
if score >= 90:
  print("优秀")
elif 60 <= score < 90:
  print("合格")
else:
  print("不合格,加油哦。")
````

# 🐍 Python 学习笔记｜第 6 关：循环入门 (For Loops)

---

## 1. 核心概念
* **For 循环**：就像一个“自动流水线”，它可以遍历一个序列（如一组数字或一个列表），并对其中的每一项执行相同的操作。
* **range() 函数**：循环的“控制器”。
    * `range(5)`：生成 0 到 4 的数字（**含头不含尾**，共 5 个）。
    * `range(1, 11)`：生成 1 到 10 的数字。

### 实例演示
```python
# 示例 1：基础计数
print("我们要数五个数")
for i in range(5):
    print("当前数字是:", i)
print("-" * 20)

# 示例 2：自动生成乘法表
number_input = input("请输入一个数字:")
number = int(number_input)
for i in range(1, 11):
    print(f"{number} * {i} = {number * i}")
```

---

## 2. 知识点：数据展示的三种姿势

在循环中，我们经常需要把变量和文字拼在一起显示。Python 提供了三种主流方式：

| 方式 | 代码示例 | 强制转换类型？ | 自动加空格？ | 评价 |
| :--- | :--- | :---: | :---: | :--- |
| **加号拼接** | `print("分数:" + str(score))` | **是** | 否 | 麻烦，容易忘转类型导致报错 |
| **逗号分隔** | `print("分数:", score)` | 否 | **是** | 简单省事，但控制不了空格 |
| **f-string** | `print(f"分数:{score}")` | 否 | 否 | **最推荐**！既简洁又精准 |

### 2.1 避坑指南：为什么 `range(5)` 不是到 5？
Python 的计数习惯是从 **0** 开始的。
* `range(5)` 实际上产出了：`0, 1, 2, 3, 4`。
* 这种“含头不含尾”的设计是为了方便计算长度（$5 - 0 = 5$ 个元素）。

---

## 3. 语法小结
```python
for 变量 in 序列:
    # 这里是循环体
    # 必须缩进！
```

# 🐍 Python 学习笔记｜第 7 关：条件循环 (While Loops)

---

## 1. 核心概念
* **While 循环**：只要指定的条件为 `True`，代码块就会不停地重复执行。
* **对比 For 循环**：
    * `for` 循环：适用于**已知次数**的重复。
    * `while` 循环：适用于**未知次数**，直到某个条件不再满足为止。

### 实例演示：火箭发射倒计时
```python
count = 5

while count > 0:
    print(f"当前数字是: {count}")
    count = count - 1  # 关键步骤：改变变量，趋向循环结束

print("🚀 发射！")
```

---

## 2. 知识点
### 2.1 While 循环的结构
1. **初始变量**：在循环开始前，先定义一个变量（如 `count = 5`）。
2. **条件判断**：每次循环前都会检查 `while` 后的条件。
3. **循环体**：缩进的代码部分。
4. **改变状态**：**极其重要！** 必须在循环体内修改变量（如 `count -= 1`），否则会陷入死循环。

### 2.2 避坑指南：死循环 (Infinite Loop)
如果你忘记了修改条件变量，或者条件永远为真，程序会一直运行下去，直到内存溢出或你强行停止它。
* **错误示范**：
  ```python
  while True:
      print("我停不下来了！") 
  ```
* **如何自救**：在终端（Command Line）运行程序时，按下 **`Ctrl + C`** 可以强制终止运行。

---

## 3. 常见应用场景
* **用户登录**：只要密码不对（Condition），就一直提示用户重新输入。
* **游戏主循环**：只要玩家没点“退出游戏”，画面就一直刷新。
* **传感器监控**：只要温度低于 20°C，就一直开启加热器。

---
> **💡 小技巧**：还记得上一关学的 `count -= 1` 吗？在 `while` 循环中，它能让你的代码看起来更像老手。

# 🐍 Python 学习笔记｜第 8 关：列表 (Lists)

---

## 1. 核心概念
* **列表 (Lists)**：一种可以按顺序存储多个数据的容器。就像一个多格抽屉，你可以用一个变量名装下一整堆东西。
* **访问方式**：计算机数数是从 **0** 开始的（即“索引”）。

### 实例演示：背包系统的增删改
```python
bag = []
bag.append("手机")
bag.append("电脑")
print(f"当前背包:{bag}")

bag[0]= "手电" # 通过索引修改元素
print(f"升级后:{bag}")

count = len(bag) # 获取列表里的元素数量
```

---

## 2. 知识点
### 2.1 三大括号的“权力边界”
* **`[]` 方括号**：**列表专用**。用于创建列表或找位置。例如：`bag[0]`
* **`()` 圆括号**：**函数专用**。用于下达动作指令。例如：`len(bag)`
* **`{}` 花括号**：**f-string 专用**。在字符串里插变量。例如：`f"{bag}"`

### 2.2 核心操作要点
1. **创建**：使用方括号 `[]`，元素间用逗号隔开。
2. **添加**：使用 `.append()` 往列表末尾塞入新数据。
3. **计数**：使用 `len()` 函数测量列表长度。

---

## 3. 常见应用场景
* **游戏背包**：存储玩家捡到的所有道具。
* **用户管理**：记录一连串的用户名或 ID。
* **数据收集**：只要有新数据进入，就用 `append` 存入列表。

---
> **💡 小技巧**：还记得上一关的 `while` 循环吗？你可以用 `while` 配合 `len(bag)`，让程序一个接一个地处理列表里的所有宝贝！

# 🐍 Python 学习笔记｜第 9 关：函数 (Functions)

---

## 1. 核心概念
* **函数**：把常用的代码打包成一个“技能”，给它起个名字，以后随调随用。
* **参数 (Parameters)**：函数名后面括号里的变量，像占位符，接收外部传进来的数据。
* **返回值 (Return)**：函数执行完后“吐”出来的结果，可以交给其他变量保存。

### 实例演示：自定义技能与计算
```python
# 1. 定义一个检查背包的函数
def check_bag(owner_name, bag_list): 
    # def 告诉 Python “我要定义技能了”
    print(f"---{owner_name}的背包检查中---")
    if len(bag_list) == 0:
        print("空空如也")
    else:
        print(f"里面的宝贝有: {bag_list}")

# 调用函数
my_stuff = ["手机", "电脑"]
check_bag("小明", my_stuff)

# 2. 定义一个带返回值的计算函数
def add_numbers(a, b):
    result = a + b
    return result # 把结果吐出来，给外面用

sum_result = add_numbers(5, 10)
print(f"计算结果是: {sum_result}")
```

---

## 2. 知识点
### 2.1 Return 的奥秘
* **并不是所有函数都要 return**。
* 如果函数只是为了执行动作（如 `print`），不需要 return。
* 如果需要函数算出一个结果供后续使用，必须使用 `return`。

### 2.2 参数传递
* 调用函数时，往括号里传的真实数据叫“实参”。
* 定义函数时，写的占位变量叫“形参”。

---

## 3. 常见应用场景
* **游戏系统**：创建英雄、伤害计算、经验值结算。
* **数据处理**：格式化日期、批量清洗字符串。

### 实例演示 2：处理参数与返回值 (中转站逻辑)
```python
def decorator_function(func):
    def wrapper(a, b): # wrapper 必须接受并传递参数
        print(f"传入的参数是: {a}, {b}")
        result = func(a, b) # 必须把返回值存起来
        print(f"函数执行结果是: {result}")
        return result # 如果不 return，原函数的返回值就会丢失
    return wrapper

@decorator_function
def add(a, b):
    return a + b
```

### 实例演示 3：魔法组合 *args 与 **kwargs

```python
# 示例：利用魔法组合做通用计时器
def log_execution(func):
    def wrapper(*args, **kwargs):
        import time 
        start_time = time.time() # 获取时间戳
        print(f"开始时间是:{start_time}")
        
        result = func(*args, **kwargs) # 传递所有参数
        
        end_time = time.time()
        print(f"函数执行花费的时间是:{end_time - start_time}")
        return result
    return wrapper

@log_execution
def calculate_sum(a, b):
    return a + b
```

* **垂直对齐看归属**：左边对齐的代码是平级的。

### 2.2 参数顺序
在函数定义中：`普通参数 -> *args -> **kwargs`。

# 🐍 Python 学习笔记｜第 21 关：面向对象编程 - 封装

---

## 1. 核心概念
* **封装 (Encapsulation)**：隐藏内部实现细节，只提供公共接口。好比电视机遥控器。
* **优势**：数据保护（避免意外修改）、简化接口（降低耦合度）。

---

## 2. 属性的封装
### 2.1 公有属性 (Public)
* 默认属性。可以在类内外直接访问。
* **关于 __init__ 参数**：情况 A（需要外部传参，开“窗口”收件）；情况 B（不需要传参，内部直接写死）。

### 2.2 私有属性 (Private)
* 约定：以双下划线 `__` 开头。外部无法直接访问（没有门禁卡）。
* **Getter/Setter**：在类内部定义公有方法来间接访问和修改。

```python
class MyClass:
  def __init__(self):
      self.__private_attribute = "这是一个私有属性"

  def get_private_attribute(self): # getter：获取
      return self.__private_attribute

  def set_private_attribute(self, value): # setter：修改
      self.__private_attribute = value

obj = MyClass()
# print(obj.__private_attribute)  # 会报错
print(obj.get_private_attribute()) # 通过“内部通行证”访问
```

## 一、 类方法 (@classmethod)
类方法是属于类的方法，而不是属于实例的方法。类方法可以通过类名或实例来调用，但通常推荐使用类名调用。
* **定义**：使用 `@classmethod` 装饰器，第一个参数固定为 `cls`（代表类本身）。
* **用途**：
    * **A. 操作类变量**: 当你需要修改或访问属于整个类的状态，而不是某个具体对象的状态时。
    * **B. 实现“工厂方法” (Factory Methods)**: 当 `__init__` 提供的初始化方式不够用时，用类方法提供多种创建对象的方式。

### 实例 1：Date 工厂方法
```python
class Date:
  def __init__(self, year, month, day):
      self.year = year
      self.month = month
      self.day = day

  @classmethod
  def from_string(cls, date_str):
      # 预处理字符串
      year, month, day = map(int, date_str.split('-'))
      # 返回类的新实例 (cls 等同于 Date)
      return cls(year, month, day)

# 使用工厂方法创建实例
today = Date.from_string("2026-02-07")
```



---

## 二、你的深度问答与对比
**问： 为什么要用类方法将2026-2-7改为年，月，日，而不是将日期改成年，月，日之后再输入变量呢？？？？**
**答： 为了封装性（Encapsulation）和易用性**
* **封装性**：如果你在程序的 10 个地方都要从字符串转日期，你就得把这段 split 代码写 10 遍。
* **维护困难**：万一哪天日期格式变了，你得去全局搜索并修改这 10 处代码。
* **类方法的做法**： 将这种“脏活累活”藏在类内部。

### “通过类方法访问”优于“直接访问属性”的核心理由：
1. **类方法是“活”的**：可以在返回数据前进行加工（例如 1050 变 "1k+"）。
2. **隐藏内部实现细节（解耦）**：插座接口不变，内部电线（变量名）随便改，只要插座孔位（类方法）不变，外部电器就能一直用。
3. **继承中的“多态”表现**：`cls` 参数是动态绑定的。它能感知到是谁在调用它。

## 一、核心知识点

## 二、关键魔法方法实例
```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    """
    问：为什么感觉self在两个定义中扮演的角色不同？
    答：
    在 __init__(self, x, y) 中：self 是“待装修的毛坯房”
    状态：此时对象刚刚被创建（出生），属性 x 和 y 还没挂上去。
    任务：赋值。你的任务是把传进来的参数 x 和 y 强行绑定到这个空壳 self 上。
    你可以理解为：self 是一个接收者。
    
    在 __add__(self, other) 中：self 是“成熟的参与者”
    状态：此时 v1 已经是一个完整的、有数据的对象了（它已经有了 self.x 和 self.y）。
    任务：计算。它拿着自己已有的数据，去和另一个对象 other 的数据做加法。
    你可以理解为：self 是一个操作发起者。
    """

    """
    问：为什么return Vector对象Vector(self.x + other.x, self.y + other.y)而不是return 元组(self.x + other.x, self.y + other.y)
    答：
    1. 链式调用 (Chaining)
    如果你返回的是 Vector，你可以连续进行加法运算。如果返回的是元组，程序会报错。
    v1 + v2 + v3 实际上等同于 (v1 + v2) + v3。
    2. 保留“能力”（方法和属性）
    Vector 类不仅仅有坐标，它还有你定义的其他功能，比如 len()。返回 Vector：可以计算模长。
    3. 类型的一致性 (Consistency)
    数学上，两个向量相加的结果依然是一个向量。改变类型会违反直觉。
    """

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
        # 当你直接在交互式环境中输入一个对象名时，__repr__ 会被调用。
        # 当你使用 print() 函数打印一个对象时，如果定义了__str__ 会被调用。
        # 如果你没有定义 __str__，Python 会默认使用 __repr__ 的结果

    def __len__(self):
        return int((self.x**2 + self.y**2) ** 0.5)
        # 知识点： 单星号 (*)：乘法 ； 双星号 (**)：幂运算（次方）。

v1 = Vector(3, 4)
v2 = Vector(1, 2)
v3 = v1 + v2
print(v1 + v2) 
#运算阶段：执行 v1.__add__(v2)，创建并返回了一个全新的对象：Vector(4, 6)
#转换阶段：print 函数会自动去调用你类里定义的 __str__ 方法。
print(len(v1))
print(str(v1)) 
print(repr(v1))
```

# 🐍 Python 学习笔记｜第 26 关：OOP 设计原则：封装与接口设计

---

## 一、核心知识点
* **封装（Encapsulation）**: 将数据和操作封装在类中，通过访问控制保护数据。
* **访问控制**：
    * **公有（Public）**：默认，外部可直接访问。
    * **私有（Private）**：以 `__` 开头，外部不可直接访问（伪私有）。
    * **受保护(Protected)**：以 `_` 开头，约定俗成的内部使用。
* **接口设计（Interface Design）**：类提供的一组公共方法。
* **原则**：单一职责、高内聚低耦合、依赖倒置。

---

## 二、封装与接口设计实例
```python
class BankAccount:
    def __init__(self, account_number):
        self.__balance = 0  # __私有属性
        self.__account_number = account_number

    def deposit(self, amount):  # 公有方法
        if amount > 0:
```

## 一、继承的利弊
* **利**：代码复用、逻辑复用、实现多态。
* **弊**：强耦合（子类依赖父类实现细节）、继承爆炸、菱形继承问题。

## 二、组合优于继承（Composition Over Inheritance）
* **定义**：通过将对象嵌入到另一个对象中，而非继承，实现功能复用。
* **优势**：低耦合、灵活性强（动态替换组合对象）。
* **核心区别**：继承是“我是什么”（Is-a），组合是“我有什么”（Has-a）。

---

## 三、代码对比
### 1. 继承写法（结构僵化风险）
```python
class SuperRobot(Cleaner, Printer): # 既能打扫又能打印
    pass
# 缺点：功能增加时需要定义无数个小类，且容易产生命名冲突。
```

### 2. 组合写法（动态组装）
```python
class Robot:
    def __init__(self, tools=None):
        # 核心：将功能组件作为属性传入
        self.tools = tools or []

    def perform_tasks(self):
        for tool in self.tools:
            tool.work()

# 想要什么功能就塞进什么组件
super_bot = Robot(tools=[Cleaner(), Printer()])
```

## 四、深度问答：默认参数与 or 逻辑
* **问：为什么不要使用 `tools=[]` 作为默认参数？**
* **答**：默认参数只在定义时计算一次。使用 `[]` 会变成一个**“全局共享”的常驻对象**，所有调用共用一个篮子。

* **问：为什么写 `self.tools = tools or []`？**
* **知识点：短路求值**：如果 A 为真，返回 A；如果 A 为假（None, [], 0, ""），返回 B。
* **目的**：保证 `self.tools` 始终是列表，防止 `for` 循环遍历 `None` 时抛出 `TypeError`。

## 一、核心知识点

### 1. 鸭子类型 (Duck Typing)
* **核心理念**： “如果它走起来像鸭子，叫起来也像鸭子，那它就是鸭子。”
* **逻辑**：我们不关心对象的类型是什么，只关心它能做什么（即它有哪些方法或属性）。

```python
class Duck:
    def quack(self):
        print("嘎嘎嘎！")

class Person:
    def quack(self):
        print("我在模仿鸭子叫：嘎嘎！")

def make_it_quack(animal):
    # 我们不在乎 animal 是 Duck 还是 Person
    # 只要它有 quack 方法就行
    animal.quack()

make_it_quack(Duck())    # 输出: 嘎嘎嘎！
make_it_quack(Person())  # 输出: 我在模仿鸭子叫：嘎嘎！
```

### 2. 抽象基类 (Abstract Base Classes / ABC)
* **核心理念**： “你必须先证明你是鸭子，我才让你进场。”。 
* **作用**：抽象基类（通过 abc 模块实现）为类提供了一种合同制。它定义了一套规范，子类必须实现这些规范才能被实例化。
* **使用必要性**：
    1. **强制约束**： 确保子类实现了所有必须的方法，否则在实例化时就会报错。
    2. **类型检查**： 允许使用 isinstance(obj, MyABC) 来做更有把握的判断。
    3. **接口设计**： 适合大型项目，明确定义各个组件之间的交互协议。

## 二、ABC 语法深度解析
### 1. 角色与分工
```python
from abc import ABC, abstractmethod  # 这行代码其实是去 Python 的**法律库（abc 模块）**里请了两个“专员”过来

# 角色 A：模板 (ABC)
class 模板(ABC): # ABC：它是“身份专员”。告诉 Python：“接下来的这个类是一个模板，不许直接拿来用（实例化）。”
    """
    当你写下 class 模板(ABC): 时：
    它的身份： 它是“抽象”的。
    它的限制： Python 不允许你直接用它。执行 s = Shape() 会直接报错。
    它的作用： 它像一张底稿，规定了后代必须长什么样。
    """
    
    @abstractmethod # 它是“考核专员”。装饰器贴在方法头上。告诉 Python：“这个方法是必考题，子类要是没写这道题，直接判不及格（报错）。”
    def 必须实现的方法(self):
        pass
     
# 角色 B：具体实现 (子类)
class 具体实现(模板):
    """
    当你写下 class Circle(Shape): 时：
    它的身份： 它是“具体”的。
    它的限制： 只要它把模板里的必考题（抽象方法）都做完了，它就可以被实例化。
    它的作用： 它是真正的执行者，负责把模板里的构思变成现实。
    """
    def 必须实现的方法(self):
        print("我正在干活")
```


### 2. 必须记住的“三个死理”
1. **父亲不能实例化**：你不能执行 `a = 模板()`。因为它是抽象的，Python 不允许一个“半成品”存在。
2. **儿子不准偷懒**：如果“具体实现”类没有写 `必须实现的方法`，那么儿子也会被封印，无法实例化。
3. **不继承 ABC 就没用**：如果你忘了写 `(ABC)`，那么 `@abstractmethod` 就像普通的注释一样，起不到任何强制约束作用。

> **总结**：鸭子类型是“随缘”，重在当下能不能用。抽象基类 (ABC) 是“契约”，重在事前的强制规范。

 * 通过依赖注入传递具体的解析器。

````python
#parser & validator.py
from abc import ABC, abstractmethod
class BaseParser(ABC):
    @abstractmethod
    def parse(self, file_name):
        pass
class BaseValidator(ABC):
    @abstractmethod
    def validate(self, data):
        pass
#Json组
class JsonParser(BaseParser):
    def parse(self, json_file):
        print(f"成功解析文件:{json_file}")
        with open(json_file, 'r') as f:
            data = f.read()
        return data       
class JsonValidator(BaseValidator):
    def validate(self, data):
        print(f"成功验证数据:{data}")
#Yaml组
class YamlParser(BaseParser):
    def parse(self, yaml_file):
        print(f"成功解析文件:{yaml_file}")
        with open(yaml_file, 'r') as f:
            data = f.read()
        return data# 🚨 注意：一旦执行了 return，函数就结束了！
    # print(f"成功解析文件:{yaml_file}") 这一行永远不会被执行，所以放到第一行
class YamlValidator(BaseValidator):
    def validate(self, data):
        print(f"成功验证数据:{data}")


#ConfigLoader.py
class ConfigLoader:
    def __init__(self):
        self.parser = None
        self.validator = None
    def _get_tools(self, file_name):
        """内部方法：根据后缀名装配工具"""
        if file_name.endswith(".json"):
            self.parser = JsonParser()
            self.validator = JsonValidator()
            
        elif file_name.endswith(".yaml"):
            self.parser = YamlParser()
            self.validator = YamlValidator()
    def load(self, file_name):
        """内部方法：根据后缀名装配工具"""
        self._get_tools(file_name)
        if self.parser is None:
            print(f"❌ 暂不支持格式: {file_name}")
            return
        data = self.parser.parse(file_name)
        self.validator.validate(data)

#调用
loader = ConfigLoader()
loader.load("config.json")
````

# 🐍 Python 学习笔记｜第 29 关：OOP 实战：处理复杂对象间的交互

## 一、核心知识点
### 1. 复杂对象交互的挑战
* **问题**：多个对象直接相互引用，导致代码难以维护和扩展。
* 解决方案：
#### 1. 中介者模式：通过中间对象协调交互，减少对象间直接依赖。
实例
```python
# 中介者类
class Mediator:
    def __init__(self):
        self.components = []

    def register(self, component):
        self.components.append(component)

    def notify(self, sender, event):
        print(f"【中介者】收到来自 {sender.name} 的消息: {event}")
        for component in self.components:
            # 这里的 sender 作用：跳过发送者本人
            if component != sender: # 如果当前遍历到的对象不是发送者本人
                print(f"  -> 转发给 {component.name}...")
                component.handle_event(event)
# 基础组件类
class Component:
    def __init__(self, name, mediator):
        self.name = name
        self.mediator = mediator
        mediator.register(self)
    def send_event(self, event):
        self.mediator.notify(self, event)
# 具体组件 A：闹钟
class Alarm(Component):
    def handle_event(self, event):
        if event == "起床时间到":
            print(f"[{self.name}] 疯狂响铃！叮铃铃...")
# 具体组件 B：咖啡机
class CoffeeMachine(Component):
    def handle_event(self, event):
        if event == "起床时间到":
            print(f"[{self.name}] 开始磨咖啡豆，准备热咖啡...")

#调用
hub = Mediator()
alarm = Alarm("小闹钟", hub)
coffee = CoffeeMachine("咖啡机", hub)
```

**问？：**这里的sender是什么？
**答：** 
  * 作用：防止“无限循环”（回声消除）：如果组件 A 发送了一个消息给中介者，中介者把消息转发给所有人（包括 A 自己），那么 A 收到消息后可能会再次触发 send_event，导致程序进入死循环。
  * 参数传递：
    1. 第一次跳转：alarm.send_event("起床时间到")
        **·**你看到的：只传了一个字符串 "起床时间到"。
        **·**实际发生的：Python 自动把 alarm 实例塞给了第一个参数 self。
        **·**结果：在 send_event 函数内部，self = alarm 实例。
    2. 第二次跳转（最关键）：self.mediator.notify(self, event)
        ```python
        #在 Component 类中：
        def send_event(self, event):
        # 此刻：self 是 alarm，event 是 "起床时间到"
        # 调用开始：
        self.mediator.notify(self, event) 
        #    ^^^^^^^         ^^^^  ^^^^^
        #      A              B      C
        ```
        **·**A (self.mediator)：这是 hub（中介者实例）。因为是用 hub.notify 调用的，所以 notify 函数定义里的第一个 self 自动接收了 hub。
        **·**B (self)：这是 alarm 实例。它被放在了参数列表的 第二个位置。
        **·**C (event)：这是字符串。它被放在了 第三个位置。
    3. 第三步：进入 Mediator 的定义
        ```
        # 在 Mediator 类中：
        def notify(self, sender, event):
            #      ^^^^  ^^^^^^  ^^^^^
            #       A      B       C
        ```
        **·**窗口 A (self)：接住了 A (hub)。
        **·**窗口 B (sender)：接住了 B (self / 也就是 alarm)。  <-- 这就是你要的答案！
        **·**窗口 C (event)：接住了 C (event / 字符串)。
        **总结：这就是为什么在 Python 中，实例方法定义时总要多写一个 self 的原因：它得留个位置给 Python 自动传参，剩下的位置才留给用户手动传参。**

    * 删除文件夹：

# 🐍 Python 学习笔记｜第 33 关：正则表达式（re 模块实现高级字符串匹配）。
## 一、正则表达式（Regular Expression，简称 Regex 或 RegEx）
### 定义： 是一种用特定模式（Pattern）来匹配、查找、替换文本的特殊字符串。

### 1.元字符：
`.`  匹配除换行符外的任意字符，
`\d` 匹配任意一个数字（0-9）
`\D` 匹配任意一个非数字。
`\w` 匹配任意一个字母、数字或下划线（单词字符）。
`\W` 匹配任意一个非单词字符。
`\s` 匹配任意一个空白符（空格、制表符、换行符）。
`\S` 匹配任意一个非空白符。
`[]` 用于定义字符集
`[aeiou]` 匹配任意元音字母
`[^0-9]` 匹配非数字字符, `^` 在中括号内表示“非”

### 2.量词：

| 量词 | 含义 | 举例 | 匹配结果 |
| :--- | :--- | :--- | :--- |
| `*` | 出现 0 次或多次 | `hi*` | `h`, `hi`, `hii`, `hiii` |
| `+` | 出现 1 次或多次 | `hi+` | `hi`, `hii`（但不能是 `h`） |
| `?` | 出现 0 次或 1 次（可选） | `colou?r` | `color`, `colour` |
| `{n}` | 明确出现 n 次 | `\d{4}` | 连续 4 个数字（如 2026） |
| `{n,}` | 至少出现 n 次 | `\d{2,}` | 至少 2 个数字 |
| `{n,m}` | 出现 n 到 m 次 | `\d{2,4}` | 2 位、3 位或 4 位数字 |

匹配：含义一：提取/查找（从一大堆字里，把符合样式的“揪”出来）
      含义二：验证（检查整段字符串，是不是全盘符合这个样式）

### 3.位置锚定（Anchors）： 有时候我们不关心字符是什么，只关心它在什么位置。
`^`：匹配字符串的开头（注意与中括号里的“非”区分）。
`$`：匹配字符串的结尾。
`\b`：匹配一个单词边界（即单词和空格/标点之间的那个缝隙）。

例子：
^Apple：只能匹配以 Apple 开头的句子。
Apple$：只能匹配以 Apple 结尾的句子。
\bcat\b：只能匹配孤立的单词 cat，不会匹配到 catch 或 copycat。

### 4.捕获组（Grouping）
`()` 捕获组
**作用：**
1. 整体打包（分组）：把括号内的多个字符捆绑成一个独立单元，方便后面加量词（如 +、*、?）来控制它们整体出现的次数。
2. 开辟内存（捕获）：正则引擎在匹配时，会为每个括号分配一个“储物柜”（寄存器），把括号里匹配到的具体文本单独存起来，供后续调用。

**储物柜的编号规则（从左到右）**
正则引擎是以 左括号 ( 出现的先后顺序，从左到右依次编号为 1, 2, 3... 号捕获组。
group(0)：永远代表整个正则表达式匹配到的完整文本。
group(1)：代表第 1 个左括号对应的储物柜。
group(2)：代表第 2 个左括号对应的储物柜，以此类推。

举例：
正则表达式：((\d{4})-(\d{2}))
面对文本：2026-05

group(0) = 2026-05 （整个匹配）
group(1) = 2026-05 （最外层的左括号）
group(2) = 2026    （第二层左括号，年份）
group(3) = 05      （第三层左括号，月份）

用法：
①连带修饰：(abc)+ 表示字符串 abc 重复出现一次以上。
②捕获组：括号匹配到的内容会被暂时存在内存里，后续可以“复用”（称为反向引用）。
例：
(\d{4})-\1   
\d{4} 表示 4 个任意数字。                                
加了括号变成 (\d{4})，意思就是：“匹配 4 个数字，并且把它复制下来。”
\1 当正则表达式引擎看到 () 时，它不仅会匹配里面的内容，还会默默把匹配到的文字复制下来，存进一个叫 \1 的临时变量里。（如果有多个（），那么就会自动编号\2 \3 \4）
普通的字符（比如字母、数字、横杠 -、空格等）默认就是它们自己，不需要任何特殊的匹配符号。

### 5. 分支条件（OR）

`|` 管道符 表示“或”。
cat|dog 可以匹配 cat 或者 dog。
T(o|e)st 可以匹配 Tost 或 Test。

**高级用法**
`动态正则拼接（用 | 取代 for 循环）`
当你需要从数据库的一个列表里查找匹配项时，用 for 循环一条条去对比文本，如果数据库有 1 万条数据，电脑就要扫描文本 1 万次，效率极低。而通过将列表元素用 | 拼接成一条大正则，电脑只需要扫描一遍文本，就能一次性把所有词都找出来。

例:
```python
database_list = ["apple", "banana", "orange"]

# 1. map(re.escape, database_list) -> 给列表里的每个词戴上安全防护罩
# 2. "|".join(...) -> 用竖线把它们串起来
pattern = "|".join(map(re.escape, database_list))

print(pattern)  # 得到动态正则: "apple|banana|orange"
```

### 6. 转义字符（Escape）
如果你想匹配元字符本身（比如想在文本里找一个真正的 . 或 *），需要在前面加反斜杠 \。
`\.` 匹配点号。
`\*` 匹配星号。
`\\` 匹配反斜杠本身。

### 7.re 模块方法：
import re
目标文本：text = "张三的电话是13811112222，李四的电话是13933334444。"

`re.match()`   仅从字符串起始位置匹配；如果开头第一秒没对上，它立刻放弃，直接返回 None（啥也没找到）。
用法： re.match(r"\d{11}", text)
结果： 啥也找不到（返回 None）。因为 text 的开头是“张三的...”，不是数字

`re.search() ` 从头到尾扫描整句话，直到找到第一个符合条件的就立刻收工。
用法： result = re.search(r"\d{11}", text)
结果： 找到了张三的电话 13811112222。后面李四的电话它就地忽略了，因为已经完成了任务。
怎么拿出来： 用 result.group() 就能把抓到的字符串吐出来。
适用场景： 你只在乎文本里“有没有”某个东西，或者只需要找第一个出现的关键词。

`re.findall()` 它是最常用的方法之一。它会扫描整个字符串，把所有符合条件的全部抓出来，打包成一个 List（列表） 给你。
用法： results = re.findall(r"\d{11}", text)
结果： 返回一个列表：['13811112222', '13933334444']。

`re.sub()`   sub 是 substitute（替换）的缩写。它不仅能找到符合规则的文本，还能顺手把它换成你指定的新内容。
用法： new_text = re.sub(r"\d{11}", "XXXXXXXXXXX", text)
结果： 传回一句话："张三的电话是XXXXXXXXXXX，李四的电话是XXXXXXXXXXX。"

`re.compile()` 前面 4 种方法，每次执行时，电脑都要临时去理解你写的那个 r"\d{11}" 到底是什么意思，这会消耗一点点时间。如果你有一千万行文本要处理，每次都临时理解就太慢了。re.compile() 就是先把通缉令“翻译/编译”好，存成一个固定的工具，以后直接用这个工具去搜，速度极快。
用法： 

`re.escape()`自动把字符串里所有有特殊含义的字符前面都加上反斜杠进行转义，让它们变回纯文本。
用法：safe_word = re.escape(apple.pie*)
结果：print(safe_word)  # 输出: apple\.pie\* （自动帮你在 . 和 * 前面加了反斜杠）

`r` 上述五种函数中的r 代表 Raw String（原始字符串），让字符串内部的反斜杠 \ 失去转义功能，变成一个普普通通的文本字符。
```python

phone_scanner = re.compile(r"\d{11}")
# 2. 以后直接用这个扫描枪去扫不同的文本，不用再写规则了
res1 = phone_scanner.findall(text1)
res2 = phone_scanner.findall(text2)
```

    * **文本**：`<div>苹果</div><span>香蕉</span><div>橙子</div>`
    * **正则**：`<div>.*</div>`
    * **结果**：`<div>苹果</div><span>香蕉</span><div>橙子</div>`
`非贪婪匹配 (Lazy)`
* **语法**：在量词后加问号，如 `*?`、`+?`、`??`
* **特性**：尽可能少地匹配字符，只要达成匹配条件立即停止。
* **示例**：
    * **正则**：`<div>.*?</div>`
    * **结果**：`<div>苹果</div>` 和 `<div>橙子</div>`

`正向预查 (Positive Lookahead)`
* **定义**：这是一种“零宽断言”，用于在不消耗字符的情况下，检查当前位置之后的内容。
* **语法**：`(?=pattern)`
* **特点**：
    1.  **非捕获**：检查的内容不会出现在最终结果里。
    2.  **零宽度**：匹配后光标回到原位，不占用匹配长度。

* **应用场景**
    1. 后缀过滤：匹配后面跟着特定字符的内容。
            正则：`\d+(?=元)`
            文本：`价格100元` -> 匹配结果：`100`
    2. 多重条件校验（如密码强度）：
            正则：`^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$`
            含义：必须同时包含大写字母、小写字母和数字，且长度至少8位。

## 一、多线程 vs 多进程
1. 进程（Process）：好比一家工厂。工厂之间相互独立，各自拥有独立的土地、原材料仓库（独立的内存空间）。 
2. 线程（Thread）：好比工厂里的工人。一个工厂里可以有多个工人，他们共同使用工厂里的设备和原材料（共享内存空间）。
3. 多线程（Multithreading）是计算机科学中的一个核心概念。简单来说，它是指在一个程序（进程）中，同时运行多个执行路径（线程）来共同完成任务的技术。
4. 多进程（Multiprocessing）是计算机科学中的一个核心概念。简单来说，进程（Process）是操作系统分配资源和调度的基本单位，而多进程就是指在系统中同时运行多个独立的程序（进程）。
5. 在通用的计算机世界里，多线程和多进程都能利用多核 CPU；但在 Python 的世界里，因为 GIL 的存在，只有多进程才能真正利用多核 CPU 进行并行计算。
6. 在 GIL 的存在下，Python 的多线程确实无法利用多核 CPU 来提高“计算”效率；但是在处理“输入输出（I/O）”任务时，它依然能大幅提高效率。
    * I/O 密集型任务（多线程：大显身手 🚀）
        什么是 I/O 密集型：网络请求（爬虫、下载文件）、读写硬盘文件、查询数据库等。
        为什么有用：这些任务的特点是“等”。当线程 A 发起一个网络请求时，它需要等待服务器响应，此时它不需要 CPU 进行计算。一旦线程 A 开始等待 I/O，Python 就会主动让线程 A 释放 GIL 锁，把麦克风交给线程 B。线程 B 就可以趁机开始工作。结果就是你不需要等第一张图片下载完再发起第二张的请求，而是可以同时发起 10 个下载请求。效率提升了几倍甚至几十倍。
    * 计算密集型任务 / CPU 密集型任务（多线程：确实没用 ❌）
        什么是计算密集型：视频解码、图形渲染、大规模数学计算（如矩阵运算、机器学习模型训练）。
        为什么没用：这些任务需要 CPU 疯狂运转。线程 A、B、C 都需要抢着执行代码，但因为 GIL 的限制，它们只能排队轮流用同一个 CPU 核心。更糟糕的是由于线程之间来回切换（抢麦克风）需要消耗额外的系统资源，在 Python 中用多线程跑计算密集型任务，甚至可能比单线程跑得还要慢。
7. 总结

| 特性 | 多线程 (Multithreading) | 多进程 (Multiprocessing) |
| :--- | :--- | :--- |
| **内存空间** | 共享同一内存空间（全局变量可访问） | 独立内存空间（数据通过 IPC 通信） |
| **GIL 限制** | 受 Python 全局解释器锁（GIL）限制，CPU 密集型任务效率低 | 无 GIL 限制，适合 CPU 密集型任务 |
| **启动开销** | 低 | 高 |
| **适用场景** | I/O 密集型任务（如网络请求、文件读写） | CPU 密集型任务（如数学计算、数据分析） |

## 二、threading 模块 & multiprocessing 模块

1. threading 模块：官方提供的“多线程”工具箱。
`threading.Thread`：类 - 这是整个模块里最核心的“线程调度器”。**在 Python 中，一个 Thread 对象的实例，就代表一个独立的线程。**
    作用：用来打包你的任务，并告诉操作系统：“嘿，帮我单独开辟一个新线程来跑这个任务！”
    核心参数：
        * `target`：任务目标。你得告诉这个线程要去执行哪个函数。注意，这里只传函数名字，绝对不能加括号（加了括号函数就立刻在主线程执行了，而不是交给子线程）。
        * `args`：任务参数。如果你的函数需要传参，必须用元组 (Tuple) 的形式传给 args。   
    `.start()`：类方法 - 启动键。告诉操作系统：“这个线程可以开始跑了！”调用后，新线程会立刻独立飞出去执行它的任务，而你的主程序（主线程）会继续往下走，不会在这里傻等。
    `.join()`： 类方法 - 同步键（强行等待）。告诉主程序：“你先别急着往下走，给我在这里等着，直到这个子线程把活干完了，你再继续！”
示例：
```python
# 第一步：把工具箱搬过来
import threading
import time

# 第二步：定义好线程要去干的“活”（函数）
def watch_movie(movie_name):
    print(f"【子线程】开始看电影: {movie_name}")
    time.sleep(2)  # 模拟看了2秒钟
    print(f"【子线程】电影 {movie_name} 看完了！")

# 第三步：主程序开始运行（主线程）
if __name__ == '__main__':
    print("【主线程】主程序启动了...")
    
    # 第四步：从工具箱里拿出一个 Thread 工人，把活交给他
    # target 指向函数名，args 传入元组参数 (注意结尾的逗号)
    t1 = threading.Thread(target=watch_movie, args=("《黑客帝国》",))
    
    # 第五步：按下启动键
    t1.start()  
    
    # 此时，t1 线程已经飞出去看电影了，主线程不会卡住，会立刻往下走
    print("【主线程】我已经把看电影的任务交出去了，我自己先喝杯茶...")
    
    # 第六步：按下同步键，主线程在这里死等 t1 完工
    t1.join()  
    
    print("【主线程】好了，子线程看完了，主程序彻底退出。")
```

2. multiprocessing 模块：官方提供的“多进程”工具箱。
`multiprocessing.Process`: 类 - 这是多进程里的“进程创建器”。**一个 Process 对象就代表操作系统中的一个独立进程。**
    作用：克隆一个全新的 Python 解释器环境，拥有自己独立的内存。
    核心参数：
        * `target`：任务目标。需要传入子进程要去执行的函数名字。注意：只传函数名，绝对不能加括号（加了括号函数就会立刻在当前的主进程中执行，而不会交给子进程）。
        * `args`：任务参数。如果你的目标函数需要传参，必须用元组 (Tuple) 的形式传给 args（例如单个参数要写成 args=("参数1",)，末尾逗号不能省）。
    `.start()`：类方法 - 启动键。告诉操作系统：“这个子进程可以开始跑了！”调用后，操作系统会正式创建并启动子进程去执行它的任务。此时主程序（主进程）会继续往下走，不会在这里傻等。
    `.join()`： 类方法 - 同步键（强行等待）。告诉主程序（主进程）：“你先别急着往下走，给我在这里等着，直到这个子进程把活全部干完了，你再继续往下运行！”
示例：
```python
# 第一步：导入多进程工具箱
from multiprocessing import Process
import time

# 第二步：定义高强度的计算任务
def heavy_calculation(number):
    print(f"【子进程】开始计算 {number} 的平方和...")
    result = sum(i**2 for i in range(number))
    print(f"【子进程】计算完成！结果是: {result}")

# 第三步：Windows 下多进程必须加这句安全锁！
if __name__ == '__main__':
    print("【主进程】开始...")
    
    # 第四步：实例化一个进程对象
    p1 = Process(target=heavy_calculation, args=(1000000,))
    
    # 第五步：启动进程
    p1.start()
    
    # 第六步：等待进程结束
    p1.join()
    
    print("【主进程】结束。")
```

3. `if __name__ == '__main__':` : **进程专属**的“安全锁”， 防止代码在被当作模块导入时误执行。
`__name__` 变量记录的不是“文件叫什么名字”，而是“这个文件目前扮演什么角色”,主进程中，__name__ == '__main__'

示例：
**无锁**
```python
# my_program.py
from multiprocessing import Process
import time
print("--- 这一行代码被执行了 ---")
def my_task():  # 此时没有进程执行这个函数
    print("子进程在干活...")
# 没加锁，直接暴露在外面
p = Process(target=my_task) #创建一个进程对象
p.start()#开始进程。
```
【第 1 阶段：主进程启动】
你在电脑上双击运行了 my_program.py。
操作系统创建了【主进程】。
【主进程】开始从上到下读代码：
读到第 4 行，控制台打印出：--- 这一行代码被执行了 ---。
读到第 9 行，遇到了 p.start()。
【主进程】对系统说：“帮我开一个【子进程 A】来运行 my_task 吧！”

【第 2 阶段：子进程 A 诞生】
操作系统创建了【子进程 A】。
【核心关键点】：【子进程 A】刚生出来的时候脑子是空的，它不知道 my_task 是什么。Windows 系统说：“那没办法，【子进程 A】，你把你爸的文件 my_program.py 重新打开读取一遍吧。”
于是，【子进程 A】也开始从上到下读代码：
读到第 4 行，控制台再次打印出：--- 这一行代码被执行了 ---。（这是第二次看到了）
恐怖的事情发生了：【子进程 A】读到第 9 行，它也遇到了 p.start()！
【子进程 A】误以为这也是它的任务，于是对系统说：“帮我也开一个【子进程 B】来运行 my_task 吧！”
【第 3 阶段：无限套娃，系统崩溃】

**有锁**
```python
from multiprocessing import Process
print("--- 这一行在锁外面 ---")
def my_task():
    print("子进程在干活...")
if __name__ == '__main__':  # 这是安全锁
    print("--- 这一行在锁里面 ---")
    p = Process(target=my_task)
    p.start()
```
【主进程】启动：
它是老大，它的变量 __name__ 确实等于 '__main__'。
它打印了锁外面的字。
遇到 if 判断，条件成立！成功进入锁内，打印锁里面的字，并创建了【子进程 A】。
【子进程 A】启动：
Windows 依然让它重新读取 my_program.py。
【子进程 A】打印了锁外面的字（因为锁外面的代码谁都能执行）。
转折点：当它走到 if __name__ == '__main__': 时，因为它是被偷偷导入进来的，它的 __name__ 变成了 'my_program'，不等于 '__main__'！   （解释：Windows 系统的多进程机制（spawn 模式）有一个雷打不动的硬性规定：子进程必须在后台重新 import 一次主文件，用来初始化自己的环境，请注意这个词：import（导入）。对于主进程来说： 它是你亲手启动的，所以主进程的工牌（__name__）写着："__main__"。对于子进程来说： 它不是你手动启动的，它是被多进程系统在后台通过 import my_program 这种方式偷偷加载进来的。因为子进程是通过 import 加载这个文件的，Python 解释器就会按照规则，把子进程环境里的那块工牌（__name__）改写成："my_program"。 ）
结果：if 条件不成立！【子进程 A】直接跳过了锁里面的 p.start() 代码。
循环在这里被彻底斩断。
安全结束：
【子进程 A】由于没有触发新的 p.start()，它高高兴兴地去执行它的 my_task() 函数，打印出 "子进程在干活..."，然后安全退出。


**问**
跳过 if 之后，子进程去哪了？
既然跳过了 if 块，下面又没有其他代码了，按照正常的 Python 逻辑，子进程不是应该结束运行了吗？
并没有！因为子进程在出生时，身上还带了一个“隐藏任务”。
还记得前面提到的 Windows spawn 机制的第二步吗？
主进程在调用 p.start() 的那一刻，就已经通过系统底层，把一个任务小纸条（序列化数据）死死地贴在了子进程的脑门上。
这个小纸条上写着：
“当你把代码文件全部加载完之后，不要退出！立刻在内存里找到一个叫 my_task 的函数，然后去执行它！”

### 补充:windows的spawn机制
第一步：白手起家
Windows 操作系统接到命令，在后台启动一个全新、空白的 Python 解释器进程。这个进程现在和你的主程序没有任何血缘关系，它只知道自己是个 Python 进程。
第二步：打包行李（Pickle 序列化）
主进程把需要传给子进程的参数、数据、以及任务目标（比如 target=my_task），通过 Python 的 pickle 模块打包成二进制数据，顺着系统管道丢给这个新的子进程。
第三步：重读代码（这就是为什么要加锁的原因！）
新进程拿到行李后，发现自己脑子里还是没有 my_task 的代码定义。于是它必须做一件事：在后台隐式地重新 import 你的主脚本文件。
这一步的目的，是为了把文件里所有的 def（图纸）和全局变量在子进程的内存里重新生成一遍。
因为是 import 导入，所以子进程的 __name__ 变成了文件名，而不是 __main__。 从而触发了我们前面讲的“安全锁”机制。
第四步：解包并开工
环境初始化完成后，子进程解开主进程丢过来的行李包（反序列化），找到了 my_task 函数在当前新内存里的地址，最后真正开始执行 my_task()。

## 三、Python 线程同步与互斥锁

1.临界区 (Critical Section)：在多线程程序中，多个线程共享并同时进行修改的内存区域或变量（如全局变量、共享文件、数据库连接等）。如果不对临界区进行保护，多个线程同时读写就会导致数据错乱

2.竞态条件 (Race Condition):当多个线程同时访问和操作同一块数据，且最终执行结果取决于线程运行的精确时序（谁快谁慢）时，就形成了竞态条件。

根源：许多高级语言中的单行代码（如 count += 1），在底层其实是由“读取-修改-写入”等多个 CPU 原子指令组成的。线程在这些中间步骤之间发生切换，就会导致数据覆盖。

```python （错误示例:不加锁的惨痛教训）
import threading
counter = 0
def unsafe_increment():
    global counter
    for _ in range(100000):
        # 临界区：没有任何保护，多线程交替执行这行代码会引发数据覆盖
        counter += 1
# 启动两个线程同时改一个变量
t1 = threading.Thread(target=unsafe_increment)
t2 = threading.Thread(target=unsafe_increment)

#调用
t1.start()
t2.start()
t1.join()
t2.join()
print(f"最终结果: {counter}")  # 结果往往小于 200000，且每次运行都不一样
```

3.互斥锁 (Mutex Lock): 互斥锁是一种最基础的同步机制，用于保证在任何时刻，只有一个线程能够进入临界区。
锁定 (Acquire)：线程尝试获取锁。如果锁未被占用，该线程获取成功并将其锁定，进入临界区；如果锁已被其他线程占用，该线程将被阻塞（挂起等待）。
释放 (Release)：当持有锁的线程离开临界区时，必须释放锁。此时操作系统会唤醒其他等待该锁的线程去重新竞争。

`threading.Lock 类`： 在Python 中用来实现互斥锁

`with lock: `: **隐式管理（上下文管理器 - 推荐）**
优点：安全。无论临界区代码是否发生异常、报何种错误，Python 都会在离开 with 块时自动释放锁，有效避免死锁。
`lock.acquire() & `lock.release()`: **显式管理（手动控制）**
注意：必须配合 try...finally 结构使用，确保 release() 在 finally 块中执行，否则代码报错会导致锁永远无法释放。


```
二、 互斥锁 (Lock)
知识点：互斥锁（threading.Lock）能保证同一时刻只有一个线程进入临界区。推荐使用 with lock: 上下文管理器，它能确保即使中间代码报错，锁也会被安全释放，防止死锁。

正确示例（使用 with 自动管理锁）：

Python
import threading

counter = 0
cell_lock = threading.Lock()  # 创建互斥锁


def safe_increment():
    global counter
    for _ in range(100000):
        # 使用 with 语句，进入时自动 acquire()，离开时自动 release()
        with cell_lock:
            counter += 1


t1 = threading.Thread(target=safe_increment)
t2 = threading.Thread(target=safe_increment)
t1.start()
t2.start()
t1.join()
t2.join()

print(f"安全加密后的结果: {counter}")  # 必定输出：200000
三、 可重入锁 (RLock)
知识点：普通的 Lock 无法被同一个线程连续获取多次（会把自己锁死）。而 RLock（可重入锁）允许同一个线程多次执行 acquire()，适合用于递归函数或者复杂的嵌套调用。

实例（同一线程内的嵌套加锁）：

Python
import threading

# 如果这里用 threading.Lock()，程序运行到 bar() 就会永久卡死
nested_lock = threading.RLock()


def foo():
    with nested_lock:
        print("foo 拿到了锁，准备调用 bar")
        bar()  # 在同一线程中，bar 内部再次去拿这把锁


def bar():
    with nested_lock:  # 同一个线程再次拿锁成功，不会死锁
        print("bar 也成功拿到了同一把锁")


t = threading.Thread(target=foo)
t.start()
t.join()
四、 信号量 (Semaphore)
知识点：互斥锁只允许 1 个线程访问。而信号量允许 N 个线程同时访问资源，超过 N 的线程必须排队。这常用于对稀缺资源的限流（如数据库连接池、爬虫并发控制）。

实例（限制最多 3 个线程同时工作）：

Python
import time
import threading

# 限制最多允许 3 个线程同时进入
pool_semaphore = threading.Semaphore(3)


def worker(worker_id):
    with pool_semaphore:
        print(f"工人 {worker_id} 占用了坑位。")
        time.sleep(1)  # 模拟工作耗时
        print(f"工人 {worker_id} 工作结束，释放坑位。")


# 开启 5 个线程，但因为信号量限制，前 3 个先执行，后 2 个等有人出来再执行
threads = [threading.Thread(target=worker, args=(i,)) for i in range(5)]
for t in threads:
    t.start()
for t in threads:
    t.join()
五、 事件 (Event)
知识点：用于线程间的基本通知机制。一个线程负责把红绿灯变成“绿灯”（set()），其他所有在等待的线程（wait()）就会同时被唤醒并继续执行。

实例（主线程通知子线程开始跑数）：

Python
import time
import threading

start_signal = threading.Event()


def run_task(name):
    print(f"线程 {name} 已准备就绪，等待主线程发令...")
    start_signal.wait()  # 卡在这里，直到主线程调用 set()
    print(f"【信号已到】线程 {name} 开始疯狂工作！")


t1 = threading.Thread(target=run_task, args=("A",))
t2 = threading.Thread(target=run_task, args=("B",))
t1.start()
t2.start()

time.sleep(2)  # 主线程模拟准备工作
print("主线程准备完毕，拉响警报！")
start_signal.set()  # 唤醒所有在 wait 的子线程

t1.join()
t2.join()
六、 条件变量 (Condition)
知识点：比 Event 更复杂的同步工具。它拥有一把锁以及一个等待队列。通常用于生产者-消费者模型：当队列满了，生产者等待；当队列空了，消费者等待。可以通过 notify() 唤醒特定数量的等待线程。

实例（简单的生产者与消费者协作）：

Python
import time
import threading

con = threading.Condition()
product_count = 0  # 共享的货物库存


def producer():
    global product_count
    with con:
        print("生产者：正在生产商品...")
        time.sleep(1)
        product_count += 1
        print("生产者：商品已上架，通知消费者！")
        con.notify()  # 唤醒一个正在等待的消费者


def consumer():
    global product_count
    with con:
        while product_count == 0:
            print("消费者：没货了，老老实实排队等...")
            con.wait()  # 释放锁并进入等待状态，直到被 notify 唤醒
        product_count -= 1
        print("消费者：终于买到商品了！")


# 先让消费者去买（此时没货，会进入 wait）
c = threading.Thread(target=consumer)
p = threading.Thread(target=producer)

c.start()
time.sleep(0.5)  # 确保消费者先运行并卡住
p.start()

c.join()
p.join()
```
