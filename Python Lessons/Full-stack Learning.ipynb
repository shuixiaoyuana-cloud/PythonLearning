第一阶段：前端（面子工程）
从零开始构建用户看到的每一个像素。

第 1 关：HTML 核心 —— 掌握 div, p, h1-h6, a, img。

第 2 关：HTML 表单 —— 专门攻克 input, button, select, form（这是和后端交互的入口）。

第 3 关：CSS 基础选择器 —— 学会用 class 和 id 精准给 HTML 元素“上色”。

第 4 关：CSS 盒模型 —— 彻底理解 padding, margin, border，解决“为什么我的按钮挤在一起”的问题。

第 5 关：Flexbox 布局 —— 掌握 flex-direction, justify-content，学会做横向排列和居中。

第 6 关：CSS Grid 布局 —— 学会像九宫格一样划分网页区域。

第 7 关：响应式设计 —— 学习 Media Queries，让网页在手机上也能正常看。

第 8 关：JS 基础语法 —— 在网页环境里用 JS 定义变量和函数。

第 9 关：DOM 查找与修改 —— 用 document.querySelector 找到网页元素并改掉它的文字。

第 10 关：JS 事件监听 —— 绑定点击 (click)、输入 (input) 事件。


第二阶段：后端（大脑逻辑 - FastAPI）
用 Python 搭建处理请求的工厂。

第 11 关：环境搭建 —— 安装 uvicorn 和 fastapi，启动第一个 Web 服务。

第 12 关：路径参数 —— 学习如何处理类似 user/123 里的那个 123。

第 13 关：查询参数 —— 学习如何处理类似 search?keyword=python 里的关键词。

第 14 关：请求体 (Request Body) —— 学习如何接收前端发来的复杂 JSON 数据。

第 15 关：Pydantic 模型 —— 为你的数据创建“说明书”，自动校验数据对不对。

第 16 关：响应模型 —— 规范后端吐给前端的数据格式，不泄露敏感信息。

第 17 关：状态码与异常 —— 学习什么时候回传 200（成功），什么时候回传 404 或 500。

第 18 关：依赖注入 (Depends) —— 学习如何复用代码逻辑（比如检查用户是否登录）。

第三阶段：数据库（记忆系统）
让数据真正落地生根。

第 19 关：数据库选型与安装 —— 认识 SQLite (轻量) 和 PostgreSQL (强大)。

第 20 关：SQL 基础增删改查 —— 亲手在黑窗口里写 SELECT * FROM users。

第 21 关：SQLAlchemy 模型定义 —— 用 Python 的 class 描述数据库表的结构。

第 22 关：Session 生命周期 —— 理解数据库连接的“打开-操作-提交-关闭”。

第 23 关：CRUD 实战 —— 编写一套完整的 Python 函数来操作数据库记录。

第 24 关：多表关联查询 —— 处理“帖子”和“评论”之间的 1 对多关系。

第四阶段：全栈合体（产品交付）
打通任督二脉，把所有东西串起来。

第 25 关：CORS 跨域解封 —— 解决前端无法访问后端的经典报错。

第 26 关：Fetch API 异步调用 —— 在 JS 里调用 fetch() 从 Python 获取数据并刷新网页。

第 27 关：JWT 登录原理 —— 学习生成和校验加密令牌（Token）。

第 28 关：密码哈希加密 —— 永远不要在数据库里存明文密码，学会用 passlib 加密。

第 29 关：文件上传与存储 —— 学习如何处理用户头像和图片。

第 30 关：一键部署 —— 使用 Docker 或云平台让你的项目在互联网上运行。
