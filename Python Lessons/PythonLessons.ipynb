# 🐍 Python 学习笔记｜第四关：交互与类型转换 (Input & Casting)

## 一、核心概念
1. 交互 (`input`)：获取用户的指令。这是程序与人沟通的唯一窗口。
2. 转换 (`Casting`)：把获取的内容转换成正确的格式。

## 二、核心知识点
1. int() 函数：类型转换大法
    用于将其他类型强制转换为**整数**。
    * (1) 转换数字类型：
        * **浮点数**：直接截断小数（不进位）。如 `int(3.9)` → `3`。
        * **布尔值**：`int(True)` → `1`，`int(False)` → `0`。
    * (2) 转换字符串：
        * 必须是合法的整数样式。`int("12.5")` 或 `int("abc")` 都会报错。
    * (3) 进制转换：
        * `int("1010", 2)` → `10`（二进制转十进制）。
    * (4) 空转换：
        * `int()` → `0`。

2. f-string：格式化神器
    在字符串前加 `f`，通过 `{}` 嵌入变量或表达式。

    * 基础用法：`f"你好，{name}"`
    * 表达式运算：`f"计算结果：{2 + 3}"` → `计算结果：5`
    * 格式化控制：
        * 保留小数：`{pi:.2f}` （保留 2 位小数，四舍五入）。
        * 补零对齐：`{num:0>5}` （宽度为 5，不足补 0）。
    * 自查技巧：`f"{score=}"` → 输出 `score=95`（调试神器）。
    * 注意事项：
        * 引号冲突：内部字符串用单引号，外部用双引号。
        * 转义：想显示大括号请用双层 `{{}}`。
3. 实例演示
```python
user_input = input("请输入你的幸运数字：")  #input()函数返回的数据类型是字符串
print("你输入的数据类型是：", type(user_input)) 
lucky_number = int(user_input) #强制转换：不然没法做数学题
print(f"你的数字加 10 以后是：{lucky_number + 10}")
```

# 🐍 Python 学习笔记｜第五关：逻辑判断 (If-Else)


## 一、核心概念
1. 逻辑判断：让程序根据不同的条件执行不同的代码块。这是程序实现“智能化”的基础。
2. 缩进 (Indentation)：Python 的强制要求。`if`、`elif`、`else` 后面的执行代码必须整体向右缩进（通常 4 个空格）。

3. 实例演示
```python
limit = 18
age_input = input("请输入你的年龄：")
age = int(age_input)  # 类型转换：将字符串转为整数

if age >= limit:
    print("验证通过") 
    print("欢迎进入成年人的编程世界")
elif age >= 12:
    print("你是青少年，建议在家长的陪同下学习")
else:
    print("小朋友，先去喝杯牛奶再来吧")

print("判断结束") # 无缩进，无论结果如何都会执行
```

---

## 二、核心知识点

1. 比较运算符
    用于比较两个值，结果为布尔值（`True` 或 `False`）。

    | 符号 | 描述 | 示例 |
    | :--- | :--- | :--- |
    | `>` | 大于 | `5 > 3` 为 True |
    | `<` | 小于 | `5 < 3` 为 False |
    | `>=` | 大于等于 | `5 >= 5` 为 True |
    | `<=` | 小于等于 | `5 <= 4` 为 False |
    | **`==`** | **等于** | **必考点**：一个 `=` 是赋值，两个 `==` 才是比较！ |
    | `!=` | 不等于 | `5 != 3` 为 True |

2. 增强赋值运算符
    一种减少代码重复的缩写形式。

    `+=`：`a += 1` 等同于 `a = a + 1`
    `-=`：`a -= 1` 等同于 `a = a - 1`
    `*=`：`a *= 2` 等同于 `a = a * 2`
    `/=`：`a /= 2` 等同于 `a = a / 2`
    注意：Python 中几乎所有的数学运算（如 `**`、`//`）都可以这样缩写。

3. 判断结构逻辑
    `if`：第一个条件判断。
    `elif`：上个条件不满足时，再判断这个。可以有多个 `elif`。
    `else`：当上面所有条件都不行时，执行这个“最终保底”项。

**避坑指南**：在 Python 中，冒号 `:` 是 `if` 结构的标志，漏掉它会引发语法错误。

````python
#练习
score_input  = input("请输入你的分数：")
score = int(score_input)
if score >= 90:
  print("优秀")
elif 60 <= score < 90:
  print("合格")
else:
  print("不合格,加油哦。")
````

# 🐍 Python 学习笔记｜第 6 关：循环入门 (For Loops)

---

## 一、核心概念
1. `For 循环`：就像一个“自动流水线”，它可以遍历一个序列（如一组数字或一个列表），并对其中的每一项执行相同的操作。
2. `range() 函数`：循环的“控制器”。
    * `range(5)`：生成 0 到 4 的数字（**含头不含尾**，共 5 个）。
    * `range(1, 11)`：生成 1 到 10 的数字。

3. 实例演示
```python
# 示例 1：基础计数
print("我们要数五个数")
for i in range(5):
    print("当前数字是:", i)
print("-" * 20)

# 示例 2：自动生成乘法表
number_input = input("请输入一个数字:")
number = int(number_input)
for i in range(1, 11):
    print(f"{number} * {i} = {number * i}")
```

---

## 二、知识点：数据展示的三种姿势

1. 在循环中，我们经常需要把变量和文字拼在一起显示。Python 提供了三种主流方式：

| 方式 | 代码示例 | 强制转换类型？ | 自动加空格？ | 评价 |
| :--- | :--- | :---: | :---: | :--- |
| **加号拼接** | `print("分数:" + str(score))` | **是** | 否 | 麻烦，容易忘转类型导致报错 |
| **逗号分隔** | `print("分数:", score)` | 否 | **是** | 简单省事，但控制不了空格 |
| **f-string** | `print(f"分数:{score}")` | 否 | 否 | **最推荐**！既简洁又精准 |

2. 避坑指南：为什么 `range(5)` 不是到 5？
    Python 的计数习惯是从 **0** 开始的。
    * `range(5)` 实际上产出了：`0, 1, 2, 3, 4`。
    * 这种“含头不含尾”的设计是为了方便计算长度（$5 - 0 = 5$ 个元素）。

---

## 三、语法小结

```python
for 变量 in 序列:
    # 这里是循环体
    # 必须缩进！
```
**💡 进阶思考**：如果在循环里再嵌套一个循环，是不是就能一秒生成完整的“九九乘法表”了呢？

# 🐍 Python 学习笔记｜第 7 关：条件循环 (While Loops)

---

## 一、 核心概念
1. **While 循环**：只要指定的条件为 `True`，代码块就会不停地重复执行。
2. **对比 For 循环**：
    * `for` 循环：适用于**已知次数**的重复。
    * `while` 循环：适用于**未知次数**，直到某个条件不再满足为止。

3. 实例演示：火箭发射倒计时
    ```python
    count = 5
    while count > 0:
        print(f"当前数字是: {count}")
        count -= 1  # 关键步骤：改变变量，趋向循环结束

    print("🚀 发射！")
    ```

---

## 二、核心知识点
1. While 循环的结构
    1. **初始变量**：在循环开始前，先定义一个变量（如 `count = 5`）。
    2. **条件判断**：每次循环前都会检查 `while` 后的条件。
    3. **循环体**：缩进的代码部分。
    4. **改变状态**：**极其重要！** 必须在循环体内修改变量（如 `count -= 1`），否则会陷入死循环。

2. 避坑指南：死循环 (Infinite Loop)
如果你忘记了修改loop variable（循环变量），或者条件永远为真，程序会一直运行下去，直到内存溢出或你强行停止它。
3. 错误示范
    ```python
    while True:
        print("我停不下来了！") 
    ```
    **如何自救**：在终端（Command Line）运行程序时，按下 **`Ctrl + C`** 可以强制终止运行。

---

## 三、常见应用场景
* **用户登录**：只要密码不对（Condition），就一直提示用户重新输入。
* **游戏主循环**：只要玩家没点“退出游戏”，画面就一直刷新。
* **传感器监控**：只要温度低于 20°C，就一直开启加热器。

# 🐍 Python 学习笔记｜第 8 关：列表 (Lists)

## 一、核心概念
1. 列表 (Lists)：一种可以按顺序存储多个数据的容器。就像一个多格抽屉，你可以用一个变量名装下一整堆东西。
2. 访问方式：计算机数数是从 **0** 开始的（即“索引”）。
3.核心操作要点
    1. **创建**：使用方括号 `[]`，元素间用逗号隔开。
    2. **添加**：使用 `.append()` 往列表末尾塞入新数据。
    3. **计数**：使用 `len()` 函数测量列表长度。
4. 实例演示：背包系统的增删改
    ```python
    bag = []
    bag.append("手机")
    bag.append("电脑")
    print(f"当前背包:{bag}")

    bag[0]= "手电" # 通过索引修改元素
    print(f"升级后:{bag}")

    count = len(bag) # 获取列表里的元素数量
    ```
5. 应用场景
    * 游戏背包：存储玩家捡到的所有道具。
    * 用户管理：记录一连串的用户名或 ID。
    * 数据收集：只要有新数据进入，就用 `append` 存入列表。

## 2. 补充知识点
1. 三大括号的“权力边界”
    1. `[]` ：方括号的核心关键词是“列表”和“索引”。
        * 创建列表 (List)：列表是Python中最常用的可变有序序列。
        * 访问元素（索引与切片）:无论是列表、元组、字符串还是字典，只要想通过位置或键来获取内部数据，都用方括号。
        * 列表推导式 (List Comprehension):用一行代码快速生成列表的高级语法。
            ```python
            evens = [x for x in range(10) if x % 2 == 0] # [0, 2, 4, 6, 8]
            ```
    2. `()` ：圆括号
        * 函数调用与定义：定义一个函数或者调用它时，必须使用圆括号来传递参数。
        * 创建元组 (Tuple)：元组是一种不可变的有序序列。
        * 控制运算优先级：和数学中的用法完全一样，用来强制先计算某一部分。
        * 生成器表达式 (Generator Expression)：用圆括号包裹的推导式不会立刻生成列表，而是返回一个生成器，省内存神器。
            ```python
            squares = (x**2 for x in range(5))  # 返回一个生成器对象
            ```
    3. `{}` 花括号：花括号在Python中属于键值对结构或去重结构，主要有两个身份。
        * 创建字典 (Dictionary)：字典是无序的键值对集合（Python 3.7+ 内部实现变更为保留插入顺序）。
        * 创建集合 (Set)：集合是一个无序且不重复的元素集合。
            **注意：**如果要创建一个空字典，用 empty_dict = {}；如果要创建一个空集合，必须用 empty_set = set()，因为 {} 默 认被字典占用了。
        * 格式化字符串 (f-string): 在字符串前加 f 时，花括号用来在字符串中嵌入变量或表达式。

# 🐍 Python 学习笔记｜第 9 关：函数 (Functions)

## 一、核心概念
1. 函数：把常用的代码打包成一个“技能”，给它起个名字，以后随调随用。
2. 参数 (Parameters)：函数名后面括号里的变量，像占位符，接收外部传进来的数据。
3. 返回值 (Return)：函数执行完后“吐”出来的结果，可以交给其他变量保存。

## 二、核心知识点
1. Return 的奥秘
    * **并不是所有函数都要 return**。
    * 如果函数只是为了执行动作（如 `print`），不需要 return。
    * 如果需要函数算出一个结果供后续使用，必须使用 `return`。
2. 参数传递
    * 调用函数时，往括号里传的真实数据叫“实参”。
    * 定义函数时，写的占位变量叫“形参”。

3. 实例演示：自定义技能与计算
```python
# 1. 定义一个检查背包的函数
def check_bag(owner_name, bag_list): 
    # def 告诉 Python “我要定义技能了”
    print(f"---{owner_name}的背包检查中---")
    if len(bag_list) == 0:
        print("空空如也")
    else:
        print(f"里面的宝贝有: {bag_list}")

# 调用函数
my_stuff = ["手机", "电脑"]
check_bag("小明", my_stuff)

# 2. 定义一个带返回值的计算函数
def add_numbers(a, b):
    result = a + b
    return result # 把结果吐出来，给外面用

sum_result = add_numbers(5, 10)
print(f"计算结果是: {sum_result}")
```
4. 常见应用场景
    **游戏系统**：创建英雄、伤害计算、经验值结算。
    **数据处理**：格式化日期、批量清洗字符串。

# 🐍 Python 学习笔记｜第 10 关：综合实战 (Final Project)

## 一、核心知识点
1. Continue 的妙用
2.`continue` 会让程序立刻“掉头”，跳过当前循环剩下的逻辑，直接进入下一次循环条件检查。
3. Random 库
    `import random`
    `random.randint(start, end)`：包含首尾的随机整数生成器。

## 二、练习 
项目：数字炸弹小游戏 💣
综合运用 `random` 库、`while` 循环、`if` 判断以及 `continue` 关键字。

````python
import random # 导入随机数库

def start_game():
    secret_number = random.randint(1, 100) # 生成 1-100 随机数
    guess = 0
    attempts = 0
    print("---💣 欢迎来到数字炸弹游戏 💣---")

    while guess != secret_number:
        user_input = input("请输入 1 到 100 之间的数字: ")
        
        # 空输入检查
        if user_input == "":
            print("别发呆了，快输入数字！")
            continue  # 关键点：跳过本次剩余代码，直接回到 while 开头
            
        guess = int(user_input)
        attempts += 1
        
        if guess > secret_number:
            print("太大了，再小一点")
        elif guess < secret_number:
            print("太小了，再大一点")
        else:
            print(f"🎉 恭喜你猜对了！答案是 {secret_number}，你用了 {attempts} 次！")

# 启动游戏
start_game()
````

# 🐍 Python 学习笔记｜第 11 关：字典 (Dictionaries)

## 一、核心概念
1. 字典：像现实中的字典一样，通过“键 (Key)”查找“值 (Value)”。
2. 语法：使用花括号 `{}`，键值对用冒号 `:` 连接，多组数据用逗号隔开。
3. 键与值的规则
    Key (键)：必须是**不可变**的（如字符串、数字）。它是唯一的索引。
    Value (值)：什么都可以放（列表、字典、数字、字符串）。

    从字典里通过“键（Key）”获取对应的“值（Value）
    `value = dict.get('key')`
    `value = dict['key']` 
    **💡 小技巧**：访问字典时，如果键不存在会报错。试着学习 `hero.get("魔力值", 0)`，如果找不到键，它会优雅地返回 0 而不是让程序崩溃。

3. 实例演示：英雄属性表
    ```python
    # 创建字典
    hero = {"名字": "齐天大圣", "生命值": 100, "攻击力": 50}

## 二、 文件处理
1. 基础语法
    1. `open()`:打开一个文件，并返回一个文件对象（File Object）
        `file_object = open(file_name, mode, encoding)` 
            file_name: 文件的路径（可以是相对路径或绝对路径）。
            mode: 打开文件的模式（见下表）。
                `'w'`：写入。覆盖原文件，不存在则新建。
                `'a'`：追加。在末尾增加，不删除旧内容。
                `'r'`：读取。只能看，不能改。
                `'b'`: 二进制模式	配合其他模式使用（如 'rb', 'wb'），用于图片、音频、PDF 等非文本文件.
            encoding: 文本文件的编码方式，通常在处理中文时需要指定 encoding='utf-8'。
        `file_object.close()`： 操作完文件后需要调用 f.close()
        **缺点**：操作完文件后需要调用 f.close()。但如果程序在中间报错或者忘记调用close，就会导致内存泄漏或文件损坏。

    2. `with open("filename", "open模式"，编码方式) as f`：with 语句（上下文管理器）。它会在代码块执行完毕后，自动关闭文件，即使中途报错也不例外。
    

    3. 读取与写入方法
        1. 读取方法
            `f.read(size)`: 默认读取所有内容。如果文件很大，会非常占内存。
            `f.readline()`：逐行读取.每次只读取文件的一行，适合处理超大文本。
            `f.readlines()`：读取所有行并返回列表
            `for line in f:` 文件对象本身是可迭代的。如果要逐行处理大文件，这种方法最省内存。
                `print(line.strip())` Python自带换行, print() 函数在打印完内容后，默认会自动在末尾加上一个换行。所以使用strip() 会去掉字符串首尾的所有空白字符（包括空格、制表符 \t 和换行符 \n）

        2. 写入方法
            `write(string)`：写入字符串
            `lines = ["第一行\n", "第二行\n", "第三行\n"]`：写入字符串列表
            `writelines(lines)`

    4. 文件指针定位：tell() 和 seek()
        在读写文件时，有一个隐形的“光标”在移动，这就是文件指针。
        `f.tell()`: 返回文件指针当前的位置（距离文件开头的字节数）。
        `f.seek(offset, whence)`: 移动文件指针到指定位置。
            `offset`: 必填参数，表示你要移动多少个字节。偏移量（字节数）。
            `whence`: 参考位置（0 代表文件开头，默认值；1 代表当前位置；2 代表文件末尾）。

        ```python
        with open('test.txt', 'r', encoding='utf-8') as f:
        print(f.read(5))  # 读取前 5 个字节
        print(f.tell())   # 输出 5
        
        f.seek(0)         # 把指针放回文件开头
        print(f.read(5))  # 重新读取前 5 个字节
        ```

**💡 小技巧**：`f.write()` 括号里必须是字符串。如果你想写数字，记得用 `str()` 转换一下！

2. 实例演示 2：文件操作 (File I/O)
```python

# with 是管家，会自动合上文件；encoding="utf-8" 保证中文不乱码

# 写入模式 'w'：会覆盖原文件

    f.write("小明,98,60\n")
    f.write("小红,95,85\n")
print("成绩已成功保存到 scores.txt")

# 追加模式 'a'：练习循环输入并保存

    while True:
        text_input = input(">")
        if text_input == "q":
            break
        f.write(text_input + "\n") # \n 是换行符，像回车键
```
3. 补充知识点
当运行 for A in B: 时
Python在后台其实是跟in后的对象对话：
Python 解释器： “老兄，我要开始循环了，把你的下一个数据给我。”
对象B： “没问题，我是个文本文件/普通字符串/列表，我的规矩就是一次给一整行/单个字符/单个元素。拿好。”
Python 解释器： “收到，我把这一行文字/这个字符/这个元素 装进 A 这个变量里，供接下来的代码使用。”
Python 解释器： “再给我下一个！”
对象 B ： “报告长官，数据已经彻底掏空，一个都没有了！（抛出 StopIteration 异常）”
Python 解释器： “收到，那咱们收工，循环结束！”

# 🐍 Python 学习笔记｜第 16 关：异常处理

## 一、核心概念
1. `try - except`：处理程序中的“意外情况”。当异常发生时，程序会跳转到 except 块，而不会直接崩溃。

2. 实例演示 1：基础结构
    ```python
    try:
        num = 10
        result = num / 0  # 会引发 ZeroDivisionError
    except ZeroDivisionError:
        print("不能除以零")
    else: # try 块没有异常时执行
        print(f"结果是: {result}")
    finally: # 无论是否引发异常，都会执行
        print("无论是否找到文件，都会执行这里")
    ```
3. 实例演示 2：综合练习
```python
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("除数不能为零")  
    except TypeError:
        print("参数类型错误，只能接受数字")
    else:
        return result

a = int(input("请输入一个数字>"))
b = int(input("请输入另一个数字>"))
print(divide_numbers(a, b))
```
**💡 小技巧**：异常处理能让你的程序变得“坚强”。在处理用户输入或读取外部文件时，这是必不可少的保护伞。

# 🐍 Python 学习笔记｜第 17 关：迭代器与生成器

## 一、迭代器
1. 迭代器 (Iterator)：包含指向原数据的指针和计数器，能够一个接一个地访问一组数据（比如列表、集合、字典等）中的每一个元素，而不需要关心这些数据在底层到底是写在内存的连续连续空间里，还是像链表一样散落各处。通过 `next()` 访问，每次返回一个元素并将指针后移。
        可迭代对象（Iterable）：就像一本书。它里面有内容，你可以读它，但书本身不会自己翻页。置了 __iter__() 方法的对象。比如 list, tuple, dict, set, str。


        迭代器（Iterator）：就像一个书签 + 一根翻页的手指。它记录了当前读到了哪一页，并且拥有一页一页往后翻的能力。内置了 __iter__() 和 __next__() 方法的对象。比如文件。

2. 问：当对一个迭代器使用for loop迭代时，发生了什么
   答：首先一个对象被叫做迭代器，是因为一内部可以使用系统指针执行next()函数二由于其具有可迭代性，为了兼容性和统一调用方式，给迭代器本身定义了 __iter__ 方法并返回它自己（return self）。至此，迭代器形成。for loop循环时先调用iter返回迭代器本身，然后调用自身next()进行迭代

3. 实例演示 1：迭代器基础
```python
my_list = [1, 2, 3, 4, 5] # 可迭代对象 (Iterable)
my_iterator = iter(my_list) # 创建迭代器 (Iterator)
# 迭代器是一次性的
print(next(my_iterator)) # 1
# 走到末尾后再次调用会抛出 StopIteration 异常
#for loop实际上就是哪部调用了这段代码
```

4. 实例演示 2：模拟迭代器逻辑 (Yield)
```python
def simulate_iterator(data):
    index = 0
    while index < len(data):
        yield data[index]
        index += 1

words = "who are you!"
my_iterator = simulate_iterator(words)
for char in my_iterator: # for 循环会自动处理 next 和异常
    print(char)
```

## 二、生成器 (Generator)
1. 生成器是一种特殊的迭代器。 它自动实现了迭代器协议（即不需要你手动写 __next__ 和抛出 StopIteration 异常），你只需要用一种非常简单的语法就能写出来。使用 `yield` 而不是 `return`。

2. 实例演示 3：生成器练习
```python
def my_generator(n):
    for i in range(n):
        yield i + 1 # 暂停并记录，下次继续

n = 10 
# print(next(my_generator(n))) # 每次调用 my_generator(n) 都会产生新生成器

for num in my_generator(n):
    result = num * num
    print(result)
```
3. 生成器优势：按需生成，不占用大量内存。

💡 小技巧：`yield` 有点像 `return`，但它会暂停函数。就像看视频按了“暂停”，下次点播放（调用 `next`）时会接着看。

## 三、Python 迭代器（Iterator） vs 生成器（Generator）底层机制全景对比

| 对比维度 | 自定义迭代器 (类实现) | 生成器 (yield 实现) |
| :--- | :--- | :--- |
| **表面代码量** | 🔴 臃肿。需要手动写 `class`、`__init__`、`__iter__` 和 `__next__`。 | 🟢 极简。只需一个普通的 `def` 函数配合 `yield` 关键字。 |
| **谁写了 `__next__`** | **你（开发者）自己写。** 必须显式定义 `def __next__(self)`。 | **Python 官方（C 语言）写好了。** CPython 底层的 `gen_iternext` 函数充当了 `__next__` 的驱动。 |
| **状态/变量存哪里** | **存在“类对象的属性”中。**（例如 `self.a`, `self.b`），通过对象属性来肉眼可见地记录状态。 | **存在“堆内存的栈帧（Stack Frame）”中。** 函数内部的局部变量在暂停时会被打包封印在堆里。 |
| **`next()` 触发的底层动作** | **普通的属性读取与方法调用。** 每次进入 `__next__` 都是一次全新的函数执行，靠读取 `self` 属性找回记忆。 | **栈帧的“解冻与挂载”。** 底层 C 驱动把堆里的函数栈帧重新激活，挂到执行栈顶，让 CPU 从上次 `yield` 后面继续跑。 |
| **遇到 `yield` / `return`** | 无法暂停。必须通过常规的 `return` 返回数据，函数单次执行随即结束。 | **栈帧的“冻结与缷载”。** 遇到 `yield` 时，CPU 指令指针和局部变量瞬间在堆中静止、冻结，同时把数据吐出去。 |
| **如何优雅地终止循环** | **手动引发异常。** 必须在代码里硬编码：`raise StopIteration` 来通知 `for` 循环。 | **自动引发异常。** 当整个函数体的字节码跑到尽头（或遇到 `return`），C 语言底层会自动抛出 `StopIteration`。 |
| **`for` 循环眼中的它们** | 没区别。一视同仁地调用 `iter()` 获取迭代器，然后疯狂调用 `next()` 直到撞到 `StopIteration` 墙壁。 | 没区别。同样走标准的迭代器协议，只是背后驱动数据输出的“发动机”换成了高阶的状态机。 |
| **核心哲学** | 把底层的迭代机制完全暴露出来，让开发者**手工全职护理**。 | **把简单留给用户，把复杂留给底层。** 官方用 C 语言把脏活累活全写了，封装成“语法糖”。 |

2. 迭代器（类实现）
```python
class FibonacciIterator:
    def __init__(self, limit):
        self.limit = limit
        self.count, self.a, self.b = 0, 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration  # 👈 手动死板地抛出异常
        result = self.a
        self.a, self.b = self.b, self.a + self.b  # 👈 手动维护状态
        self.count += 1
        return result

# 使用
for num in FibonacciIterator(5):
    print(num)
```

3. 生成器（yield 语法糖）
```Python
def fibonacci_generator(limit):
    count = 0
    a, b = 0, 1
    while count < limit:
        yield a  # 👈 C语言底层在此处自动“冻结”整个时空栈帧
        a, b = b, a + b
        count += 1
    # 👈 函数走到尽头，C语言底层自动引发 StopIteration

# 使用
for num in fibonacci_generator(5):
    print(num)
```

# 🐍 Python 学习笔记｜第 18 关：装饰器（Decorators）

## 一、核心概念
1. 装饰器**：接受函数 `func` 为参数，返回一个 `wrapper` 闭包函数。
   作用：不改动原代码，通过外壳添加逻辑（日志、计时、权限等）。
2. 基础语法：
    `def decorator_function(func):` :定义一个生成装饰器的函数并将待装饰函数作为参数传入        ------->这是定义一个生成装饰器的函数，调用它可以生成一个装饰器
        `def wrapper():`            : 在生成装饰器函数内部定义一个装饰器                    |
                                    : 待装饰函数执行前执行的代码                           |------>这段代码是在生成一个装饰器
            `func()`                : 待执行函数                                          |--- -->并将被装饰函数包裹起来
                                    : 待装饰函数执行后执行的代码                           |
        `return wrapper`            : 返回生成的装饰器                                      ------->返回装饰器，这样就可以将装饰器呈现出来了

    `@decorator_function`           :当解释器看到 @decorator_function 作用在 my_function 上时，它会在后台自动执行以下操作：创建原函数：首先在内存中创建原本的 my_function。
        `def my_function():`        偷梁换柱（核心）：Python 自动执行了这样一行隐藏代码：my_function = decorator_function(my_function) = wrapper
                                    
    调用函数
                        1. Python 的动作：它在脑子里登记：“有一个函数叫 decorator_function。”， 此时 wrapper 执行了吗？ 完全没有。因为现在只是定义，根本没人调用它。
                        2. 当 Python 继续往下读，看到了 @decorator_function 作用在 my_function 上时：这时候，Python 会在加载阶段立刻去调用 decorator_function(my_function)。
                                decorator_function 内部是怎么跑的？
                                    进入函数：把原 my_function 作为参数 func 传进来。
                                    定义内层（重点）：遇到 def wrapper():。Python 说：“哦，这里定义了一个叫 wrapper 的新函数，但我现在不执行它里面的 print 和 func()，我先把它的代码打包好。”
                                    执行 return：紧接着，执行 return wrapper。
                            结果：decorator_function 执行完了，它把那个还没开始跑、但已经打包好的 wrapper 函數对象给返回了出来，并交给了 my_function 这个变量名。

    `my_function()`      3. 因为 my_function 现在就是 wrapper，所以这时候 wrapper 内部的代码（“前”、func()、“后”）才真正第一次被执行。


2. 实例演示 1：基础结构与语法糖
```python
def decorator_function(func):
    def wrapper():
        print("在被装饰函数执行前执行的代码")
        func()
        print("在被装饰函数执行后执行的代码")
    return wrapper

@decorator_function
def my_function():
    print("这是我的函数")

my_function()
```

3. 实例演示 2：处理参数与返回值 (中转站逻辑)
```python
def decorator_function(func):
    def wrapper(a, b): # wrapper 必须接受并传递参数
        print(f"传入的参数是: {a}, {b}")
        result = func(a, b) # 必须把返回值存起来
        print(f"函数执行结果是: {result}")
        return result # 如果不 return，原函数的返回值就会丢失
    return wrapper

@decorator_function
def add(a, b):
    return a + b
```

### 实例演示 3：魔法组合 *args 与 **kwargs
* **`*args`**：把位置参数打包成元组。（根据参数的位置顺序来传参）
* **`**kwargs`**：把关键字参数（key=value）打包成字典。关键字参数（Keyword Arguments） 就是在调用函数时，通过“参数名 = 参数值”的形式明确指定要把值传给哪一个参数。
    ```python
    register_user(age=18, city="北京", name="张三") 
    # 依然精准输出：姓名: 张三, 年龄: 18, 城市: 北京
    ```

```python
# 示例：利用魔法组合做通用计时器
def log_execution(func):
    def wrapper(*args, **kwargs):
        import time 
        start_time = time.time() # 获取时间戳
        print(f"开始时间是:{start_time}")
        
        result = func(*args, **kwargs) # 传递所有参数
        
        end_time = time.time()
        print(f"函数执行花费的时间是:{end_time - start_time}")
        return result
    return wrapper

@log_execution
def calculate_sum(a, b):
    return a + b
```

---

## 2. 知识点
### 2.1 编程规范（缩进与冒号）
* **冒号是信号灯**：`def`, `if`, `for` 后下一行必须缩进。
* **垂直对齐看归属**：左边对齐的代码是平级的。

### 2.2 参数顺序
在函数定义中：`普通参数 -> *args -> **kwargs`。

> **💡 小技巧**：`wrapper` 就像一个快递中转站。它不仅要负责把货物（参数）送进去，还得负责把签收单（返回值）带出来！

# 🐍 Python 学习笔记｜第 19 关：面向对象编程 - 类与对象基础

## 1. 核心概念
* **面向对象编程 (OOP)**：一种编程范式，将数据（属性）和操作数据的函数（方法）封装在一起。可以想象成现实中的事物，如汽车。
* **类 (Class)**：蓝图。定义了“新物种”的长相和能力。
    * **属性 (Attributes)**：它是“什么”？（静态数据，如品牌、颜色）。
    * **行为 (Methods)**：它能“做什么”？（动态动作，如加速、刹车）。
* **对象 (Object)**：根据蓝图制造的实物，也叫 **实例 (Instance)**。每个对象的数据是独立的。

### 实例演示：设计并制造汽车
```python
# 第一阶段：设计蓝图（定义类 class）
class Car:
    def __init__(self, brand, color):

        # self.brand = brand 意为“把传进来的品牌名，贴到这辆车自己身上”。
        self.brand = brand  # 属性
        self.color = color  # 属性

    def drive(self): # 行为
        # 逻辑：通过 self 直接访问自己身上的属性
        print(f"这辆{self.color}的{self.brand}正在路上飞驰！")

neighbor_car = Car("比亚迪", "银色")

```

# 🐍 Python 学习笔记｜第 20 关：面向对象编程 - 继承与多态

---

## 1. 继承 (Inheritance)
* **逻辑**：允许子类从父类获取属性和方法，减少重复代码，方便扩展。

```python
class Animal:    
    def __init__(self, name):
        self.name = name
    def speak(self):
        print(f"{self.name} 发出声音")
    
# 子类 Dog 继承 Animal
class Dog(Animal):
    def bark(self):
        print(f"{self.name} 汪汪叫")

dog = Dog("旺财")
dog.speak() # 调用父类方法
dog.bark()  # 调用自己特有的方法
```

---

## 2. 方法重写 (Method Overriding)
* **逻辑**：当子类对父类的方法有不同实现时，可以重新定义它。

```python
class Dog(Animal):
    def speak(self): # 重写父类方法
        print(f"{self.name} 汪汪叫")
      
class Cat(Animal):
    def speak(self): # 重写父类方法
        print(f"{self.name} 喵喵叫")
```

---

## 3. 多态 (Polymorphism)
* **逻辑**：同一个方法调用，根据对象的不同类型，有不同的行为。
* **优势**：解耦（主程序不需知道子类细节）、可扩展性、灵活性。

```python
animals = [Dog("旺财"), Cat("咪咪")]
for animal in animals:
    animal.speak() # 根据“身份”给出不同反馈
```

# 🐍 Python 学习笔记｜第 21 关：面向对象编程 - 封装

---

## 1. 核心概念
* **封装 (Encapsulation)**：隐藏内部实现细节，只提供公共接口。好比电视机遥控器。
* **优势**：数据保护（避免意外修改）、简化接口（降低耦合度）。

---

## 2. 属性的封装
### 2.1 公有属性 (Public)
* 默认属性。可以在类内外直接访问。
* **关于 __init__ 参数**：情况 A（需要外部传参，开“窗口”收件）；情况 B（不需要传参，内部直接写死）。

### 2.2 私有属性 (Private)
* 约定：以双下划线 `__` 开头。外部无法直接访问（没有门禁卡）。
* **Getter/Setter**：在类内部定义公有方法来间接访问和修改。

```python
class MyClass:
  def __init__(self):
      self.__private_attribute = "这是一个私有属性"

  def get_private_attribute(self): # getter：获取
      return self.__private_attribute

  def set_private_attribute(self, value): # setter：修改
      self.__private_attribute = value

obj = MyClass()
# print(obj.__private_attribute)  # 会报错
print(obj.get_private_attribute()) # 通过“内部通行证”访问
```

### 2.3 “捉迷藏”游戏：名称修饰
* Python 并没有物理封死私有属性，而是重命名为：`_类名__属性名`。
* `print(obj._MyClass__private_attribute)` 依然可以访问（但不推荐）。

---

## 3. 拓展：内置函数 dir()
* **Directory (目录)**：列出对象拥有的所有“属性”和“方法”名字。
* **内置魔法 (Magic Methods)**：以 `__` 开头和结尾，如 `__init__`, `__str__`。
* **模块内容**：用于探索不熟悉的库。

---

## 4. 方法的封装
* 以 `__` 开头的方法。只能在类内部被其他方法调用。

```python
class MyClass:
  def __private_method(self):
      print("这是一个私有方法")

  def public_method(self):
      print("公有方法调用私有方法...")
      self.__private_method()

obj = MyClass()
obj.public_method() # 间接调用
```

### 综合练习：银行账户封装
```python
class BankAccount: 
  def __init__(self,account_number): 
    self.__balance = 0  # 私有属性：余额
    self.__account_number = account_number # 私有属性：账号

  def deposit(self,amount): # 存钱
    self.__balance += amount

  def withdraw(self,amount): # 取钱逻辑
    if self.__balance - amount < 0:
      print("余额不足，取款失败")
    else:
      self.__balance -= amount

  def get_balance(self):
    print(f"账户{self.__account_number}的余额是{self.__balance}")

my_account = BankAccount("123456789")
my_account.deposit(1000)
my_account.withdraw(300)
my_account.get_balance()
```

# 🐍 Python 学习笔记｜第 22 关：类属性与实例属性

---

## 一、类属性（Class Attributes）
直接在类中定义的变量，所有实例共享，通过类名或实例访问（推荐使用类名）类属性是属于类的，而不是属于实例的。这意味着所有实例共享同一个类属性，并且可以通过类名或实例来访问和修改它。

## 二、实例属性（Instance Attributes）
在 `__init__` 方法中定义的变量，每个实例独有，通过实例访问。实例属性是属于实例的，每个实例都有自己的实例属性，并且可以通过实例来访问和修改它。每个实例独立修改，不影响其他实例。

### 1. 作用域实例
```python
class MyClass:
  class_attr = 10 #类属性
  def __init__(self,instance_attr): #实例属性
     self.instance_attr = instance_attr

#创建实例
obj1 = MyClass(20)
obj2 = MyClass(30)
#访问类属性
print(obj1.class_attr)
print(obj2.class_attr)
#通过类名修改类属性
MyClass.class_attr = 100
print(obj1.class_attr) #所有实例共享
print(obj2.class_attr)

#通过实例修改类属性
# (在 Python 中，当你执行 obj1.class_attr = 200 时，发生的不是对类变量的赋值，而是属性遮蔽（Attribute Shadowing）。
# 类属性依然稳稳地坐在 MyClass.__dict__ 里，数值还是 100。
# 实例属性被新建在 obj1.__dict__ 里，数值是 200)
obj1.class_attr = 200 #当你通过 obj1.class_attr = 200 给它赋值时，你并没有“修改”那个全局的类属性，而是在 obj1 这个实例的内存空间里新建了一个同名的实例属性。
print(obj1.class_attr) #创建的新的属于obj1的实例属性
print(obj2.class_attr) #类属性
print(obj1.instance_attr)
print(obj2.instance_attr)
```



## 三、总结
> 类属性是大家共享的“公家财产”。通过类名修改 (MyClass.attr = x)：修改公家财产，所有人看到的都变了。通过实例赋值 (obj.attr = x)：相当于员工领了一份私人物资。从此以后，这个员工谈到 attr 时指的就是自己的私人物资，不再关心公家的了。

---

## 四、拓展：动态属性赋值与 __slots__
Python 是一门极其灵活的动态语言。即便你在类（MyClass）的定义里完全没提到某个属性，你依然可以在程序运行的过程中，随时随地给某个特定的实例塞进一个新的属性。

如果你希望严格限制属性，不准别人在外部乱加，Python 提供了一个名为 `__slots__` 的魔法变量：
```python
class RestrictedClass:
  __slots__ = ("name", "age") # 只允许有这两个属性

obj = RestrictedClass()
obj.name = "Gemini"
# obj.hobby = "编程"  # 这行会直接报错 (AttributeError)
```

# 🐍 Python 学习笔记｜第 23 关：类方法与静态方法

---

## 一、 类方法 (@classmethod)
类方法是属于类的方法，而不是属于实例的方法。类方法可以通过类名或实例来调用，但通常推荐使用类名调用。
* **定义**：使用 `@classmethod` 装饰器，第一个参数固定为 `cls`（代表类本身）。
* **用途**：
    * **A. 操作类变量**: 当你需要修改或访问属于整个类的状态，而不是某个具体对象的状态时。
    * **B. 实现“工厂方法” (Factory Methods)**: 当 `__init__` 提供的初始化方式不够用时，用类方法提供多种创建对象的方式。

### 实例 1：Date 工厂方法
```python
class Date:
  def __init__(self, year, month, day):
      self.year = year
      self.month = month
      self.day = day

  @classmethod
  def from_string(cls, date_str):
      # 预处理字符串
      year, month, day = map(int, date_str.split('-'))
      # 返回类的新实例 (cls 等同于 Date)
      return cls(year, month, day)

#使用实例方法创建实例
date1 = Date(2026, 10, 28)
print(date1.year)

# 使用工厂方法创建实例
today = Date.from_string("2026-02-07")
```



---

## 二、你的深度问答与对比
**问： 为什么要用类方法将2026-2-7改为年，月，日，而不是将日期改成年，月，日之后再输入变量呢？？？？**
**答： 为了封装性（Encapsulation）和易用性**
* **封装性**：如果你在程序的 10 个地方都要从字符串转日期，你就得把这段 split 代码写 10 遍。
* **维护困难**：万一哪天日期格式变了，你得去全局搜索并修改这 10 处代码。
* **类方法的做法**： 将这种“脏活累活”藏在类内部。

### “通过类方法访问”优于“直接访问属性”的核心理由：
1. **类方法是“活”的**：可以在返回数据前进行加工（例如 1050 变 "1k+"）。
2. **隐藏内部实现细节（解耦）**：插座接口不变，内部电线（变量名）随便改，只要插座孔位（类方法）不变，外部电器就能一直用。
3. **继承中的“多态”表现**：`cls` 参数是动态绑定的。它能感知到是谁在调用它。

---

## 三、实例 2：“星际矿工机器人”系统
```python
class MiningRobot:
  __total_count = 0 # 1. 类变量

  def __init__(self, name, iron_ore):
      self.name = name
      self.iron_ore = iron_ore 
      MiningRobot.__total_count += 1

  @classmethod
  def from_string(cls, data_str): # 2. 类方法（工厂接口）
      name, ore_str = data_str.split(",")
      return cls(name, int(ore_str))

  @classmethod
  def get_system_report(cls): # 3. 类方法（统计接口）
      return f"【星际总部报告】当前在线机器人：{cls.__total_count} 台"

  @property
  def capacity_status(self): # 4. 属性方法（状态接口）
      if self.iron_ore > 100:
          return "仓库已满"
      return "继续挖矿中..."

# 这段代码展示了三个层级的“行为”：
# 1. from_string(cls, ...)：这是“造物主”的行为。利用 cls 生产出具体的机器人。
# 2. get_system_report(cls)：这是“上帝视角”的行为。它只关心 cls（整个机器人族群）的统计数据。
# 3. capacity_status(self)：这是“个人档案”的行为。它只关心 self（我这个机器人）的状态。
```

## 四、员工管理系统：对比不使用类方法
### 1. 操作类变量的对比
* **直接访问**：`Employee.raise_amount = 1.10`（缺点：没有校验，语义不明）。
* **类方法**：`@classmethod def set_raise_amt(cls, amount)`（优点：可以加 `if amount < 1` 校验，自动记录日志）。

### 2. 解析数据（工厂模式）的对比
* **直接处理**：必须在主程序里写解析逻辑，逻辑外泄且重复劳动。
* **类方法**：`emp_2 = Employee.from_string("Bob-7000")`（优点：封装性，格式变了只需改类内部）。

### 3. 你的比喻
> **直接修改属性/外部处理逻辑**：就像是**“自助餐厅”**。你自己去拿盘子、切肉、倒酒。如果餐厅换了盘子位置，你就找不到了。
> **使用类方法**：就像是**“点餐服务”**。你只需要告诉服务员（类方法）：“我要一份套餐 A”。哪怕后厨换了厨师或菜谱，你手里的菜单依然有效。

## 五、静态方法 (@staticmethod)
执行与类相关但不需要访问类或实例属性的操作。
```python
class DateProcessor:
    @staticmethod
    def is_valid_date(day, month, year):
        # 纯逻辑判断，不依赖具体的 DateProcessor 实例
        return 1 <= day <= 31 and 1 <= month <= 12
```

## 六、静态方法、类方法&实例方法对比

```python
class Employee:
    # 类属性（所有员工共享的全局信息）
    company_name = "TechCorp"
    
    def __init__(self, name, salary):
        # 实例属性（每个员工独有的信息）
        self.name = name
        self.salary = salary

    # ================= 1. 实例方法 =================
    # 默认不加装饰器，第一个参数必须是 self
    def show_info(self):
        # 它可以访问 self.name 和 self.salary
        return f"员工: {self.name}, 薪水: {self.salary}, 属于公司: {self.company_name}"

    # ================= 2. 类方法 =================
    # 使用 @classmethod，第一个参数必须是 cls
    @classmethod
    def change_company(cls, new_name):
        # 它可以修改类属性
        cls.company_name = new_name
        
    @classmethod
    def from_string(cls, info_str):
        # 经典应用（工厂模式）：解析字符串并直接生成一个员工对象
        name, salary = info_str.split("-")
        return cls(name, int(salary))  # 等同于 Employee(name, salary)

    # ================= 3. 静态方法 =================
    # 使用 @staticmethod，参数完全自由，没有 self 和 cls
    @staticmethod
    def is_valid_salary(salary):
        # 它不关心具体的员工是谁，也不关心公司叫什么
        # 它只负责完成一个纯粹的逻辑判断：薪水是否合法
        return 0 < salary < 1000000

# 🐍 Python 学习笔记｜第 24 关：多重继承与 MRO

---

## 一、多重继承
多重继承是指一个类可以同时继承多个父类。这意味着子类可以拥有所有父类的属性和方法。
* **语法**：`class Child(Parent1, Parent2)`
* **KP pass**: 占位符。为了满足语法结构的完整性。

### 实例演示
```python
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print(f"{self.name}发出声音")
class Fly(Animal):
    def fly(self):
        print(f"{self.name}飞")
class Mammal(Animal):
    def walk(self):
        print("哺乳动物走路")
class Bat(Mammal,Fly):  
    pass

Rat = Animal("老鼠")
Rat.speak()
bat_1 = Bat("蝙蝠")
bat_1.speak()
bat_1.fly()
bat_1.walk()
```

---

## 二、 方法解析顺序（MRO）
* **核心算法**：Python 3 使用的是 C3 线性化 (C3 Linearization) 算法。
* **三大原则**: 
    1. 子类优先于父类。
    2. 从左到右。
    3. 最后找祖先（菱形继承中，基类最后才被搜索）。



---

## 三、 super()：MRO 的最佳拍档
`super()` 不是简单地调用“父类”，而是根据 MRO 列表，找到当前类在排队序列中的 **下一个角色**。
* `super()` 必须放在方法内部执行。
* 如果不写 `super()`，链条则会断掉，代码会因为找不到属性和方法报错。

### 你的深度理解
> `super()` 能够跨越到“兄弟类”（如 Mammal 到 WingedAnimal ），是因为它始终引用的是**最初触发调用的那个实例（ Bat 的实例）**的 MRO。

### 属性继承实例
```python
class Animal:
    def __init__(self):
        print("进入 Animal")
        self.alive = True
        print("离开 Animal")

class Mammal(Animal):
    def __init__(self):
        print("进入 Mammal")
        super().__init__() 
        self.has_fur = True
        print("离开 Mammal")

class WingedAnimal(Animal):
    def __init__(self):
        print("进入 WingedAnimal")
        super().__init__()
        self.can_fly = True
        print("离开 WingedAnimal")

class Bat(Mammal, WingedAnimal):
    def __init__(self):
        print("--- 蝙蝠初始化开始 ---")
        super().__init__()
        print("--- 蝙蝠初始化结束 ---")

my_bat = Bat()
```

# 🐍 Python 学习笔记｜第 25 关：魔法方法（Magic Methods）

---

## 一、核心知识点
### 1. 魔法方法基础
* **定义**：以 `__` 开头和结尾的方法，用于实现 Python 的特殊行为。
* **常见用途**：
    1. **对象初始化**: (`__init__`)
    2. **字符串表示**: (`__str__`, `__repr__`)
    3. **运算符重载**: (`__add__`, `__len__`, `__sub__`, `__eq__`)
    4. **上下文管理**: (`__enter__`, `__exit__`)
* **__eq__**: 定义当两个对象使用 `==` 运算符比较时，到底该怎么算“相等”？

---

## 二、关键魔法方法实例
```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    """
    问：为什么感觉self在两个定义中扮演的角色不同？
    答：
    在 __init__(self, x, y) 中：self 是“待装修的毛坯房”
    状态：此时对象刚刚被创建（出生），属性 x 和 y 还没挂上去。
    任务：赋值。你的任务是把传进来的参数 x 和 y 强行绑定到这个空壳 self 上。
    你可以理解为：self 是一个接收者。
    
    在 __add__(self, other) 中：self 是“成熟的参与者”
    状态：此时 v1 已经是一个完整的、有数据的对象了（它已经有了 self.x 和 self.y）。
    任务：计算。它拿着自己已有的数据，去和另一个对象 other 的数据做加法。
    你可以理解为：self 是一个操作发起者。
    """

    """
    问：为什么return Vector对象Vector(self.x + other.x, self.y + other.y)而不是return 元组(self.x + other.x, self.y + other.y)
    答：
    1. 链式调用 (Chaining)
    如果你返回的是 Vector，你可以连续进行加法运算。如果返回的是元组，程序会报错。
    v1 + v2 + v3 实际上等同于 (v1 + v2) + v3。
    2. 保留“能力”（方法和属性）
    Vector 类不仅仅有坐标，它还有你定义的其他功能，比如 len()。返回 Vector：可以计算模长。
    3. 类型的一致性 (Consistency)
    数学上，两个向量相加的结果依然是一个向量。改变类型会违反直觉。
    """

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
        # 当你直接在交互式环境中输入一个对象名时，__repr__ 会被调用。
        # 当你使用 print() 函数打印一个对象时，如果定义了__str__ 会被调用。
        # 如果你没有定义 __str__，Python 会默认使用 __repr__ 的结果

    def __len__(self):
        return int((self.x**2 + self.y**2) ** 0.5)
        # 知识点： 单星号 (*)：乘法 ； 双星号 (**)：幂运算（次方）。

v1 = Vector(3, 4)
v2 = Vector(1, 2)
v3 = v1 + v2
print(v1 + v2) 
#运算阶段：执行 v1.__add__(v2)，创建并返回了一个全新的对象：Vector(4, 6)
#转换阶段：print 函数会自动去调用你类里定义的 __str__ 方法。
print(len(v1))
print(str(v1)) 
print(repr(v1))
```

# 🐍 Python 学习笔记｜第 26 关：OOP 设计原则：封装与接口设计

---

## 一、核心知识点
* **封装（Encapsulation）**: 将数据和操作封装在类中，通过访问控制保护数据。
* **访问控制**：
    * **公有（Public）**：默认，外部可直接访问。
    * **私有（Private）**：以 `__` 开头，外部不可直接访问（伪私有）。
    * **受保护(Protected)**：以 `_` 开头，约定俗成的内部使用。
* **接口设计（Interface Design）**：类提供的一组公共方法。
* **原则**：单一职责、高内聚低耦合、依赖倒置。

---

## 二、封装与接口设计实例
```python
class BankAccount:
    def __init__(self, account_number):
        self.__balance = 0  # __私有属性
        self.__account_number = account_number

    def deposit(self, amount):  # 公有方法
        if amount > 0:
            self.__balance += amount

    def get_balance(self):  # 接口方法
        return self.__balance #类内部能够访问self.__balance。通过公有方法传递私有属性实现接口方法

acct1 =  BankAccount("123456789")
print(acct1._BankAccount__balance) # 名称改写后的访问方式
acct1.deposit(1000)
# print(acct1.__balance) # 报错：在类定义外部，Python 假装这个属性不存在。
print(acct1.get_balance())
```

# 🐍 Python 学习笔记｜第 27 关：OOP 设计原则：继承的艺术与组合优于继承

---

## 一、继承的利弊
* **利**：代码复用、逻辑复用、实现多态。
* **弊**：强耦合（子类依赖父类实现细节）、继承爆炸、菱形继承问题。

## 二、组合优于继承（Composition Over Inheritance）
* **定义**：通过将对象嵌入到另一个对象中，而非继承，实现功能复用。
* **优势**：低耦合、灵活性强（动态替换组合对象）。
* **核心区别**：继承是“我是什么”（Is-a），组合是“我有什么”（Has-a）。

---

## 三、代码对比
### 1. 继承写法（结构僵化风险）
```python
class SuperRobot(Cleaner, Printer): # 既能打扫又能打印
    pass
# 缺点：功能增加时需要定义无数个小类，且容易产生命名冲突。
```

### 2. 组合写法（动态组装）
```python
class Robot:
    def __init__(self, tools=None):
        # 核心：将功能组件作为属性传入
        self.tools = tools or []

    def perform_tasks(self):
        for tool in self.tools:
            tool.work()

# 想要什么功能就塞进什么组件
super_bot = Robot(tools=[Cleaner(), Printer()])
```

---

## 四、深度问答：默认参数与 or 逻辑
* **问：为什么不要使用 `tools=[]` 作为默认参数？**
* **答**：默认参数只在定义时计算一次。使用 `[]` 会变成一个**“全局共享”的常驻对象**，所有调用共用一个篮子。

* **问：为什么写 `self.tools = tools or []`？**
* **知识点：短路求值**：如果 A 为真，返回 A；如果 A 为假（None, [], 0, ""），返回 B。
* **目的**：保证 `self.tools` 始终是列表，防止 `for` 循环遍历 `None` 时抛出 `TypeError`。

---

## 五、练习任务：播放器优化
### 1. 继承写法（冗余且僵化）
```python
class VideoPlayer(AudioPlayer, MediaPlayer): 
    # 修正：AudioPlayer 已继承 MediaPlayer，写 (AudioPlayer) 即可。
    def adjust_resolution(self): # 必须带 self
        print ("调整分辨率")

### 2. 组合写法（解耦版）

class Audio:
    def play(self): print("播放音频")
    def adjust_volume(self): print ("调整音量")
    def adjust_resolution(self): pass

class Video:

    def adjust_resolution(self): print ("调整分辨率")
    def adjust_volume(self): print ("调整音量")

class MediaPlayer:

        self.players = players or []

```

# 🐍 Python 学习笔记｜第 28 关：OOP 实战：构建可复用的类库

---

## 一、核心知识点

### 1. 鸭子类型 (Duck Typing)
* **核心理念**： “如果它走起来像鸭子，叫起来也像鸭子，那它就是鸭子。”
* **逻辑**：我们不关心对象的类型是什么，只关心它能做什么（即它有哪些方法或属性）。

```python
class Duck:
    def quack(self):
        print("嘎嘎嘎！")

class Person:
    def quack(self):
        print("我在模仿鸭子叫：嘎嘎！")

def make_it_quack(animal):
    # 我们不在乎 animal 是 Duck 还是 Person
    # 只要它有 quack 方法就行
    animal.quack()

make_it_quack(Duck())    # 输出: 嘎嘎嘎！
make_it_quack(Person())  # 输出: 我在模仿鸭子叫：嘎嘎！
```

### 2. 抽象基类 (Abstract Base Classes / ABC)
* **核心理念**： “你必须先证明你是鸭子，我才让你进场。”。 
* **作用**：抽象基类（通过 abc 模块实现）为类提供了一种合同制。它定义了一套规范，子类必须实现这些规范才能被实例化。
* **使用必要性**：
    1. **强制约束**： 确保子类实现了所有必须的方法，否则在实例化时就会报错。
    2. **类型检查**： 允许使用 isinstance(obj, MyABC) 来做更有把握的判断。
    3. **接口设计**： 适合大型项目，明确定义各个组件之间的交互协议。

---

## 二、ABC 语法深度解析
### 1. 角色与分工
```python
from abc import ABC, abstractmethod  # 这行代码其实是去 Python 的**法律库（abc 模块）**里请了两个“专员”过来

# 角色 A：模板 (ABC)
class 模板(ABC): # ABC：它是“身份专员”。告诉 Python：“接下来的这个类是一个模板，不许直接拿来用（实例化）。”
    """
    当你写下 class 模板(ABC): 时：
    它的身份： 它是“抽象”的。
    它的限制： Python 不允许你直接用它。执行 s = Shape() 会直接报错。
    它的作用： 它像一张底稿，规定了后代必须长什么样。
    """
    
    @abstractmethod # 它是“考核专员”。装饰器贴在方法头上。告诉 Python：“这个方法是必考题，子类要是没写这道题，直接判不及格（报错）。”
    def 必须实现的方法(self):
        pass
     
# 角色 B：具体实现 (子类)
class 具体实现(模板):
    """
    当你写下 class Circle(Shape): 时：
    它的身份： 它是“具体”的。
    它的限制： 只要它把模板里的必考题（抽象方法）都做完了，它就可以被实例化。
    它的作用： 它是真正的执行者，负责把模板里的构思变成现实。
    """
    def 必须实现的方法(self):
        print("我正在干活")
```


### 2. 必须记住的“三个死理”
1. **父亲不能实例化**：你不能执行 `a = 模板()`。因为它是抽象的，Python 不允许一个“半成品”存在。
2. **儿子不准偷懒**：如果“具体实现”类没有写 `必须实现的方法`，那么儿子也会被封印，无法实例化。
3. **不继承 ABC 就没用**：如果你忘了写 `(ABC)`，那么 `@abstractmethod` 就像普通的注释一样，起不到任何强制约束作用。

> **总结**：鸭子类型是“随缘”，重在当下能不能用。抽象基类 (ABC) 是“契约”，重在事前的强制规范。

---

## 三、类库设计原则与实战
### 1. 设计原则
* **单一职责**：每个类只负责一个功能。
* **低耦合高内聚**：类之间依赖最小化，内部功能紧密关联。
* **扩展性**：通过抽象类或接口支持未来扩展。
* **提供清晰的使用说明和完善的测试用例**。

### 2. 可复用类库实例：可复用的日志系统
```python
# logger.py
class Logger: 
    """
    【指挥官类】
    它不关心日志具体怎么写（写文件还是发邮件），它只负责“下达命令”。
    这就是“低耦合”：Logger 不需要知道 FileHandler 的具体实现。
    """
    def __init__(self, handlers): 
        # 这里的 handlers 是通过外部传进来的（依赖注入）
        # 就像给机器人装上不同的工具手，装上手电筒它就能照明，装上电钻它就能打洞。
        self.handlers = handlers

    def log(self, message): 
        # Logger 类的核心方法。遍历所有的 handlers，并调用它们的 write 方法。
        for handler in self.handlers:
            handler.write(message) 
            # 【关键点】它假定每个 handler 都有 write 方法。
            # 只要你能写(write)，你就是我的处理器(handler)。这就是“鸭子类型”。

# handlers.py
class FileHandler: 
    """
    【专业工人 A：文件搬运工】
    专门负责把文字塞进硬盘里的文件。
    """
    def __init__(self, filename): # 这里的 filename 是这个工人私有的属性
        self.filename = filename

    def write(self, message): # 具体的苦力活在这里实现
        with open(self.filename, "a") as f: # 打开以传入参数filename命名的文件
            f.write(f"{message}\n")  # 写入日志信息

# 1. 准备工具（实例化具体的功能类）
file_handler = FileHandler("app.log")

# 2. 组建团队（将工具注入到指挥官类中）
# 注意这里传的是 [列表]，意味着你可以放好几个不同的 handler 进去
logger = Logger([file_handler])

# 3. 发射指令
logger.log("系统启动") 
```

## 四、单元测试 & 版本管理
### 1. pytest 基础
* **命名规则**：测试文件名必须以 `test_` 开头。
* **运行方式**：在终端运行 `pytest`，它会自动找到所有 `test_` 开头的文件。

```python
# test_calc.py
from calc import add # 从 calc.py 把 add 功能搬过来（代码搬运语法）

def test_add_function():
    # 我们“断言” 1+1 应该等于 2
    assert add(1, 1) == 2

def test_add_negative():
    # 测试一下负数
    assert add(-1, 1) == 0
```

### 2. 知识点补充
* **代码搬运语法 (Import)**：
    * `import calc`：把整个文件搬过来，使用时需带前缀 `calc.add()`。
    * `from calc import add`：把具体功能搬过来，直接用 `add()`。
* **断言 (assert)**：
    * 查相等：`assert a == b`
    * 查不相等：`assert a != b`
    * 查包含：`assert "apple" in ["apple", "banana"]`
    * 查真假：`assert True` / `assert False`


### 3. 版本管理：使用 Git 管理代码版本，遵循语义化版本号（Semantic Versioning）。/通过 __version__ 类属性标识库版本。

 * ConfigLoader：

   * 使用 JSONSValidator 验证 JSON 配置的结构。
   * 使用 YAMLValidator 验证 YAML 配置的构。

````python
#parser & validator.py
from abc import ABC, abstractmethod
class BaseParser(ABC):
    @abstractmethod
    def parse(self, file_name):
        pass
class BaseValidator(ABC):
    @abstractmethod
    def validate(self, data):
        pass
#Json组
class JsonParser(BaseParser):
    def parse(self, json_file):
        print(f"成功解析文件:{json_file}")
        with open(json_file, 'r') as f:
            data = f.read()
        return data       
class JsonValidator(BaseValidator):
    def validate(self, data):
        print(f"成功验证数据:{data}")
#Yaml组
class YamlParser(BaseParser):
    def parse(self, yaml_file):
        print(f"成功解析文件:{yaml_file}")
        with open(yaml_file, 'r') as f:
            data = f.read()
        return data# 🚨 注意：一旦执行了 return，函数就结束了！
    # print(f"成功解析文件:{yaml_file}") 这一行永远不会被执行，所以放到第一行
class YamlValidator(BaseValidator):
    def validate(self, data):
        print(f"成功验证数据:{data}")


#ConfigLoader.py
class ConfigLoader:
    def __init__(self):
        self.parser = None
        self.validator = None
    def _get_tools(self, file_name):
        """内部方法：根据后缀名装配工具"""
        if file_name.endswith(".json"):
            self.parser = JsonParser()
            self.validator = JsonValidator()
            
        elif file_name.endswith(".yaml"):
            self.parser = YamlParser()
            self.validator = YamlValidator()
    def load(self, file_name):
        """内部方法：根据后缀名装配工具"""
        self._get_tools(file_name)
        if self.parser is None:
            print(f"❌ 暂不支持格式: {file_name}")
            return
        data = self.parser.parse(file_name)
        self.validator.validate(data)

#调用
loader = ConfigLoader()
loader.load("config.json")
````

### 练习总结
* 在进行选择或者需要内部参数传递时可以使用内部方法！！！


## 知识点补充
### JSON
1. 核心语法规则
    * 数据以“键值对”形式存在。
    * 键（Key）必须是字符串，且必须用双引号 "" 包裹。
    * 数据由逗号 , 分隔。
    * 大括号 {} 保存对象。
    * 中括号 [] 保存数组。
    
    
2. 支持的数据类型
| 类型 | 描述 | 示例 |
| :--- | :--- | :--- |
| **String** | 字符串，必须使用双引号包裹 | `"Hello World"` |
| **Number** | 数值，支持整数、浮点数及科学计数法 | `42`, `3.14`, `1.0e+10` |
| **Object** | 无序键值对集合（对象） | `{"name": "Alice"}` |
| **Array** | 有序值的集合（数组） | `[1, 2, "test"]` |
| **Boolean** | 布尔值 | `true`, `false` |
| **Null** | 空值 | `null` |
3. JSON 结构示例
 ```json
{
  "project": "JSON Tutorial",
  "version": 1.0,
  "is_active": true,
  "author": {
    "name": "Gemini",
    "role": "AI Assistant"
  },
  "features": [
    "Lightweight",
    "Language Independent",
    "Standardized"
  ],
  "remarks": null
}
```
4. 注意事项
    * 不能存储函数：JSON 只能存储数据，不能包含 JavaScript 函数或变量引用。
    * JSON 文本应使用 UTF-8 编码。

5. json.dump(data, f, ensure_ascii=False, indent=4)
    * **json.dump**：一个“翻译官”：它把 Python 的字典（Dict）翻译成 JSON 格式的字符串，并顺手写进文件里。
    * **data**: 翻译的内容:在你的练习中，它是一个 Python 字典，比如 {"patient": "张三", "treatment": "..."}。
    * **f**: 目标文件：文件句柄。就是你通过 with open(...) as f 打开的那个文件对象。dump 会直接把翻译好的内容“倒进”这个文件里。
    * **ensure_ascii=False**: 中文补丁: 默认情况 (True)：JSON 最初是为英文设计的，它会把中文转义成一串你看不懂的编码（比如 \u5f20\u4e09）。设置为 False：它会告诉 Python：“嘿，直接把中文写进去，不要转码。” 这样你打开生成的 .json 文件时，看到的才是正常的汉字。
    * **indent=4**: 排版美化: 设置 indent=4：它会在每一层数据前面自动加 4 个空格的缩进。

### YAML
1. 核心语法规则： YAML 的语法极度精简，主要依赖**视觉缩进**：
    * 大小写敏感。
    * 强制缩进：使用空格表示层级关系，禁止使用 Tab 键。
    * 键值对分隔：冒号后必须有一个空格（如 key: value）。
    * 支持注释：使用 # 开头，这是与 JSON 的最大区别。
2. 支持的数据类型

3. YAML结构示例
```
# 用户基础信息
id: 101
name: 张三
isVip: true
tags:
  - 开发者
  - 摄影师
  - 骑行
address:
  city: 北京
  district: 朝阳区
scores: ~   # 代表 null
```
4. 常见操作：多行处理与引用,这是 YAML 的“独门绝技”：
    * 多行字符串：
        · `|`：保留换行符。
        · `>`：将多行合并为单行。
    * 锚点 & 引用：
        · `&`: (Anchor - 锚点)：标记。相当于“给这块内容起个变量名”。
        · `*`: (Alias - 别名/引用)：调用。相当于“在这里引用那个变量名”。
        · `<<`: (Merge Key - 合并)：展开。将引用的内容“打散”并合并到当前对象中。

        ```YAML
        default: &base # 这里我们告诉解析器：我要给 timeout: 30 这个对象做一个标记，标记的名字叫 base。
            timeout: 30

        development:
            <<: *base #[合并指令] [分隔符] [引用目标]
            # *base 找到了之前标记的内容。
            # <<: 把 base 里的内容（timeout: 30）拆开，塞进 development 下面。
            port: 8080    

         #操作结束后相当于
         default:
            timeout: 30

        development:
            timeout: 30
            port: 8080    
        ```
        · **注意：**如果你在引用了锚点后，又写了相同的键，后写的会覆盖引用的：
        ```YAML
        default: &base
            timeout: 30
            retries: 3

        production:
            <<: *base
            timeout: 5  # 这里会覆盖掉 base 里的 30，但保留 retries: 3
        ```
5. 注意事项与常见错误
    * 空格陷阱：冒号后面没加空格（如 key:value）是新手最常犯的错误。
    * Tab 键报错：在 YAML 中按下 Tab 会导致解析失败，请务必设置编辑器将 Tab 转为空格。
    * 缩进不一致：同一层级的元素必须左对齐。
    * 布尔值陷阱：在某些旧版本中，on 或 yes 也会被当成 true，如果想表达字符串，必须加引号 "on"。

# 🐍 Python 学习笔记｜第 29 关：OOP 实战：处理复杂对象间的交互

---

## 一、核心知识点
### 1. 复杂对象交互的挑战
* **问题**：多个对象直接相互引用，导致代码难以维护和扩展。
* 解决方案：
#### 1. 中介者模式：通过中间对象协调交互，减少对象间直接依赖。
实例
```python
# 中介者类
class Mediator:
    def __init__(self):
        self.components = []

    def register(self, component):
        self.components.append(component)

    def notify(self, sender, event):
        print(f"【中介者】收到来自 {sender.name} 的消息: {event}")
        for component in self.components:
            # 这里的 sender 作用：跳过发送者本人
            if component != sender: # 如果当前遍历到的对象不是发送者本人
                print(f"  -> 转发给 {component.name}...")
                component.handle_event(event)
# 基础组件类
class Component:
    def __init__(self, name, mediator):
        self.name = name
        self.mediator = mediator
        mediator.register(self)
    def send_event(self, event):
        self.mediator.notify(self, event)
# 具体组件 A：闹钟
class Alarm(Component):
    def handle_event(self, event):
        if event == "起床时间到":
            print(f"[{self.name}] 疯狂响铃！叮铃铃...")
# 具体组件 B：咖啡机
class CoffeeMachine(Component):
    def handle_event(self, event):
        if event == "起床时间到":
            print(f"[{self.name}] 开始磨咖啡豆，准备热咖啡...")

#调用
hub = Mediator()
alarm = Alarm("小闹钟", hub)
coffee = CoffeeMachine("咖啡机", hub)

print("--- 模拟：闹钟响了 ---")
alarm.send_event("起床时间到")
#执行 alarm.send_event，内部其实是调用：self.mediator.notify(self, "起床时间到")。
#这里的 sender 就是 alarm（闹钟实例），event 是字符串。
```
**问？：**这里的sender是什么？
**答：** 
  * 作用：防止“无限循环”（回声消除）：如果组件 A 发送了一个消息给中介者，中介者把消息转发给所有人（包括 A 自己），那么 A 收到消息后可能会再次触发 send_event，导致程序进入死循环。
  * 参数传递：
    1. 第一次跳转：alarm.send_event("起床时间到")
        **·**你看到的：只传了一个字符串 "起床时间到"。
        **·**实际发生的：Python 自动把 alarm 实例塞给了第一个参数 self。
        **·**结果：在 send_event 函数内部，self = alarm 实例。
    2. 第二次跳转（最关键）：self.mediator.notify(self, event)
        ```python
        #在 Component 类中：
        def send_event(self, event):
        # 此刻：self 是 alarm，event 是 "起床时间到"
        # 调用开始：
        self.mediator.notify(self, event) 
        #    ^^^^^^^         ^^^^  ^^^^^
        #      A              B      C
        ```
        **·**A (self.mediator)：这是 hub（中介者实例）。因为是用 hub.notify 调用的，所以 notify 函数定义里的第一个 self 自动接收了 hub。
        **·**B (self)：这是 alarm 实例。它被放在了参数列表的 第二个位置。
        **·**C (event)：这是字符串。它被放在了 第三个位置。
    3. 第三步：进入 Mediator 的定义
        ```
        # 在 Mediator 类中：
        def notify(self, sender, event):
            #      ^^^^  ^^^^^^  ^^^^^
            #       A      B       C
        ```
        **·**窗口 A (self)：接住了 A (hub)。
        **·**窗口 B (sender)：接住了 B (self / 也就是 alarm)。  <-- 这就是你要的答案！
        **·**窗口 C (event)：接住了 C (event / 字符串)。
        **总结：这就是为什么在 Python 中，实例方法定义时总要多写一个 self 的原因：它得留个位置给 Python 自动传参，剩下的位置才留给用户手动传参。**
#### 2. 观察者模式：实现发布 - 订阅机制，解耦通知方和接收方。
实例
```python
class Subject:
    """被观察者：博主/发布方"""
    def __init__(self):
        self._observers = [] # 维护一个观察者列表
    def attach(self, observer):
        """添加粉丝"""
        self._observers.append(observer)
    def notify(self, message):
        for observer in self._observers:# 假设每个观察者都有一个 update 方法
            observer.update(message)

class Observer:
    def __init__(self, name):
        self.name = name
    def update(self, message):
        print(f"【{self.name}】收到了推送：{message}")


blogger = Subject()
fan_a = Observer("张三")
fan_b = Observer("李四")
blogger.attach(fan_a)
blogger.attach(fan_b)
blogger.notify("今天天气不错，适合写 Python！")
```

## 一、核心知识要点
1. 系统架构设计的基本流程
    1. **需求剖析**：精准界定系统的功能以及非功能方面的需求（像性能、安全性、扩展性等）。
    2. **模块划分**：把系统拆分成若干个独立的模块，并且明确各模块的职责（例如用户管理模块、数据存储模块、业务逻辑模块等）。
    3. **模式选用**：依据需求挑选合适的设计模式（例如 MVC 模式、工厂模式、单例模式）。
    4. **接口规划**：确定模块之间的交互方式，保证接口具有清晰性和稳定性。
    5. **技术选型**：挑选适用的技术栈（例如选用 Flask 框架或者 Django 框架、使用 SQLite 数据库或者 MySQL 数据库）。
    6. **扩展性考量**：为未来的功能扩展预留接口（例如支持插件机制、采用可替换的存储后端）。

2. 分层架构示例（以学生管理系统为例）
┌───────────────┐
│   表现层      │（用户界面：控制台或者Web页面）
└───────────────┘
           ▲
           │
┌───────────────┐
│   业务逻辑层  │（处理学生的增删改查操作）
└───────────────┘
           ▲
           │
┌───────────────┐
│   数据访问层  │（与数据库进行交互：SQLAlchemy ORM）
└───────────────┘
           ▲
           │
┌───────────────┐
│   数据库      │（存储学生的相关数据）
└───────────────┘
3. 关键设计原则
   1. 单一职责原则：每个模块只承担一项独立的职责。 
   2. 开闭原则：系统应能够对扩展开放，对修改关闭。
   3. 依赖倒置原则：依赖抽象类，而不是具体的实现类。

给 App 写一个方法 run(content)，内部调用 self.writer.write_data(content)。

````python
#练习一
from abc import ABC, abstractmethod
class Gift(ABC):
    @abstractmethod
    def open_gift(self):
        pass
class Foodgift(Gift):
    def open_gift(self):
        return "真香，吃掉了"
class Toygift(Gift):
    def open_gift(self):
        return "真好玩，拆开了"

def receive(gift):
    print (f"{gift.open_gift()}")
    
gift1 = Foodgift()
gift2 = Toygift()
receive(gift1)
````

````python
#练习二
from abc import ABC, abstractmethod
class Basewriter(ABC):
    @abstractmethod
    def write_text(self, text):
        pass
class Filewriter(Basewriter):
    def __init__(self):
    #如果 __init__ 里面没有初始化任何变量，你可以直接不写这个函数，Python 会自动处理。
        pass
    def write_text(self,text):
        with open("test.txt", "w") as f:
        #"w" 模式（Write）每次运行都会把旧的 test.txt 内容清空再写新的。如果你想在后面接着写，可以用 "a" 模式（Append）。
            f.write(text)

class ConsoleWriter(Basewriter):
    def write_text(self, text):
        print(text)

class App:
    def __init__(self,writer):
        self.writer = writer
    def run(self, content):
        self.writer.write_text(content)

writer1 = Filewriter()
App1 = App(writer1)
App1.run("我是一个好人")
````

# 🐍 Python 学习笔记｜第 31 关：文件系统操作（os, shutil, pathlib）。

## 一、模块
### 模块（Module） 就是一个包含 Python 代码的 .py 文件

  * 内置模块（Standard Library）： Python 安装后自带的，如你提到的 os、shutil、sys、json 等。
  * 第三方模块： 社区开发者编写的，需要通过 pip install 安装，如 requests、pandas、numpy。
  * 自定义模块： 你自己编写的 .py 文件
### 模块调用方式
  *  整体导入：如果你直接 import datetime，你拿到了整个仓库，用的时候必须带上“前缀”。
  ```python
  import math
  print(math.pi)
  ```
  *  导入特定成员：
  ```python
  from math import pi
  from [模块/包] import [类/函数/变量]
  print(pi)
  # **别名导入（常用简化缩写）：**
  import pandas as pd
  import numpy as np
  ```
## 二、文件系统操作中常用的模块
### 常用模块对比
| 模块 | 特点 | 常用场景 |
| :--- | :--- | :--- |
| **os** | 传统模块，提供跨平台的文件系统操作（如路径处理、文件状态获取） | 简单文件操作、环境变量访问、获取系统信息 |
| **shutil** | 高级文件操作，支持复制、移动、压缩、解压等复杂功能 | 文件备份、目录递归复制、批量删除文件 |
| **pathlib** | 面向对象的路径处理，Python 3.4+ 推荐使用，语法简洁 | 现代项目中的路径管理、路径拼接与属性访问 |

### 常用模块介绍及用法

  1. **os 模块**：老牌系统交互员： os (Operating System) 是最底层的。它把路径看作是字符串。
    * os.getcwd()：获取当前代码运行的路径。cwd（current working directory）
    * os.listdir("路径")：返回一个列表，包含该目录下所有文件名。dir(directory)
    * os.makedirs("a/b/c", exist_ok=True)：递归创建多级目录。
    * os.remove("file.txt")：只能删除文件，不能删文件夹。
    * os.walk("目录")：像扫描仪一样，一层层扫出所有子文件夹和文件。
    * os.path.join("a", "b")

  2. **shutil** —— （Shell Utilities） 文件搬运工（最强力）核心逻辑： 专门处理文件的“物理移动”。如果涉及“整个文件夹”的操作，找它准没错。
    * 复制：
      shutil.copy("源", "目的地")：复制文件。
      shutil.copytree("源目录", "新目录")：复制整个文件夹。
    * 移动/重命名：
      shutil.move("旧位置", "新位置")：文件或文件夹都能移。如果“旧位置”不写路径就是在当前目录找。
    * 删除文件夹：
      shutil.rmtree("文件夹")：威力巨大，递归删除整个目录（慎用）。

  3. pathlib —— 路径处理的首选（最现代）核心逻辑： 把路径看作一个“对象”，而不是一段冰冷的文字。

    创建路径对象：p = Path("folder/subfolder")
    路径拼接：p / "file.txt" (使用 / 符号，这是它的神技)
    获取路径属性：
      .name：文件名（带后缀）
      .stem：文件名（不带后缀）
      .suffix：后缀名（如 .py, .txt）
      .parent：上一级目录
    判断与创建：
      .exists()：是否存在
      .is_file() / .is_dir()：是文件还是文件夹
      .mkdir(exist_ok=True)：创建文件夹（如果已存在也不报错）
      .mkdir(parents = true, exist_ok = true)
        * parents = true 核心逻辑： “不管目标路径有多深，只要缺了哪一级，我通通帮你补齐。”
        * parents = false核心逻辑： “我只负责建目标那个文件夹，如果它的‘爸爸’和‘爷爷’不在，我就罢工。”
      .with_suffix(".新后缀") 的逻辑是：“找到路径中最后一个点（.）之后的内容，把它替换掉。”(只是获取一个新的路径对象，而不是改名字)
      .rename("新名字")
  4. 示例一
  ```python
  import os
  import shutil
  from pathlib import Path

  # 1. 创建目录
  os.makedirs("project/data", exist_ok=True)  # 递归创建目录
  Path("project/temp").mkdir(exist_ok=True)   # 面向对象方式

  # 2. 文件操作
  shutil.copy("source.txt", "target.txt")       # 复制文件
  shutil.move("old.txt", "new/dir/renamed.txt") # 移动文件并重命名
  os.remove("temp.log")                         # 删除文件

  # 3. 路径拼接
  path1 = os.path.join("data", "file.txt")      # 使用 os.path
  path2 = Path("data") / "file.txt"             # 使用 pathlib
  ```
     示例二
  ```python
  # os.walk 遍历目录
  for root, dirs, files in os.walk("project"):
    print(f"目录: {root}")
    print(f"子目录: {dirs}")
    print(f"文件: {files}")

  # pathlib 遍历目录
  for item in Path("project").glob("*.txt"):
  #注意：item 拿到的不是字符串，而是一个 Path 对象。你可以直接对 item 接着点 .name, .stem 或者 .unlink()（删除）。这里的item既然是path 对象，就意味着包含所有path对象都包含的方法
    print(item)
  # Path("project")：先锁定要搜寻的“地盘”（project 文件夹）
  #.glob()：源于 "Global"，在计算机编程中，它代表“通配符匹配”。简单来说，它的功能就是：“在文件夹里，按照特定的‘长相’把文件揪出来。”
  #"*.txt"：搜寻的“暗号”（匹配模式）
  # * 是最强大的通配符，它代表 “任何字符，不限长度”。

  # 全局（递归）搜索：.glob("**/*.txt") 或者 .rglob("*.txt")： 如果你想把 project 文件夹里所有子文件夹中的 .txt 全找出来，就要用到这个双星号 。**代表“任意深度的文件夹”。Path("project").rglob("*.txt") 里的 r 就代表 recursive（递归）。
  ```
     示例三(错误示例)
  ```python
  # 错误：硬编码路径分隔符（Windows 上会失败，是因为 Windows 系统的“反斜杠 \” 在编程世界里有一个特殊身份：转义字符。）
  path = "C:\project\file.txt"  # 反斜杠需转义

  #解决办法
  #1. 方案 A：双反斜杠（转义转义符）
  path = "C:\\project\\file.txt"
  #原理：第一个 \ 告诉 Python：“别误会，我后面那个 \ 就是个普通的字符。”

  #2. 方案 B：原始字符串（加个 r）
  path = r"C:\project\file.txt"
  #原理：开头的 r 代表 Raw (原始)。它告诉 Python：“这一串文字里所有的斜杠都是普通字符，别乱转义。”这是处理 Windows 路径最常用的偷懒方法。

  #3. 方案 C：使用 pathlib（终极方案，墙裂推荐！）
  from pathlib import Path
  path = Path("C:/project/file.txt")
  #原理： pathlib 非常聪明。哪怕你在 Windows 上，你也可以直接用正斜杠 /（像网页地址那样）。它会自动帮你翻译成系统能听懂的路径，完全不需要担心 \ 的转义问题。
  ```

  **知识点**
  * 什么是“转义”？：
    在 Python（以及大多数编程语言）中，反斜杠 \ 后面跟着一个字母，通常代表一个特殊的控制信号，而不是一个普通的字符。
    比如：
      \n 代表 换行（New line）
      \t 代表 制表符（Tab，也就是空格缩进）
  * "C:\project\file.txt"
    Python 看到 \p：它不认识这个转义符，可能报错，也可能原样输出。
    但如果你有一个文件夹叫 test，路径写成 "C:\project\test.txt"：
    Python 看到 \t，会以为你这里想要一个 Tab 缩进！
    最终路径会变成：C:project    est.txt（中间断开了）。

   示例四
  ```python
  try:
      os.remove("file.txt")
  except FileNotFoundError:
      print("文件不存在")
  except PermissionError:
      print("无删除权限")
  ```
  **知识点** 
  * try:（尝试执行区）
    含义：你把觉得“可能会出事”的代码放在这里。
    动作：Python 会尝试运行这里的代码。如果一路顺风，它会跳过后面的 except。
  * except FileNotFoundError:（精准拦截区 A）
    含义：如果 try 里的代码抛出了“找不到文件”的错误，立刻跳到这里。
    结果：程序不会崩溃，而是执行你写的 print("文件不存在")。
  * except PermissionError:（精准拦截区 B）
    含义：如果报错原因是“权限不足”（比如文件被锁定或它是系统文件），就跳到这里。

## 三、练习
需求：编写一个文件管理工具，实现以下功能：
创建目录：根据输入路径创建多层目录（例如 mkdir -p project/data）。
复制文件：将指定文件复制到目标目录，并保留原文件名。
批量重命名：将目录中所有 .bak 文件重命名为 .old。
清理临时文件：删除目录下所有以 temp_ 开头的文件。


````python
from pathlib import Path
import shutil
import os
class FileTool():
    @staticmethod #加上@staticmethod因为这些方法不需要访问类的实例属性
    def create_dir(pathstr):
        Path(pathstr).mkdir(parents=True, exist_ok = True)# 不写parents=True会不建父目录报错
        print(f"目录已就绪: {pathstr}")#实现交互，否则操作的人不知道

    @staticmethod
    def copyfile(opath,gpath):
        shutil.copy(opath,gpath)
        print(f"已将 {opath} 复制到 {gpath}")

    @staticmethod
    def multiple_rename(project):
        for item in Path(project).glob("*.bak"):
            new_name = item.with_suffix(".old")
            item.rename(new_name)
            print(f"重命名: {item.name} -> {new_name.name}")
    @staticmethod
    def clear_tempt_document(project):
        for item in Path(project).glob("temp_*"):
            if item.is_file():#确保只删文件，不删文件夹（除非你想删）
                item.unlink()
            print(f"已清理临时文件: {item.name}")
    
        
````

# 🐍 Python 学习笔记｜第 32 关：时间与日期处理（datetime, time 模块）。
## 一、表格对比
| 模块 | 核心特点 | 常用类 / 函数 | 常用场景 |
| :--- | :--- | :--- | :--- |
| **`datetime`** | 处理日期和时间的组合，提供面向对象的接口，支持日期运算和时区。 | `datetime`, `date`, `time`, `timedelta` | 日常时间计算（如加减天数）、日期格式化输出、日志记录。 |
| **`time`** | 处理 Unix 时间戳，侧重于系统层面的时间处理和性能测量。 | `time()`, `sleep()`, `strftime()`, `perf_counter()` | 时间戳转换、程序执行计时（性能评估）、控制程序休眠。 |

### 知识点
1. *datetime模块*（有很多面向对象的类）
    .datetime : 类（实际上是datetime的子类）
        - datetime(year, month, day...) : 类的构造函数（实例化） - 构造指定日期和时间。
        - .now() : 类方法 - 获取当前的本地日期和时间。
        - strptime (p = Parse)：类方法 - “从有到无”。把一个纯字符串解析成复杂的 Python 日期对象。
        - strftime (f = Format)：实例方法 - “从无到有”。把一个 Python 日期对象包装、格式化成你想要的     
        - .weekday() ： 实例方法 - 获得datetime对象的weekday，返回的是整数

    .date : 类
        - date(year, month, day) : 类的构造函数（实例化） - 构造指定日期。
        - .today() : 类方法 - 获取今天的日期。

    .time : 类
        - time(hour, minute, second) : 类的构造函数（实例化） - 构造指定时间（时分秒）。

    .timedelta : 类
        - timedelta(days=0, seconds=0, microseconds = 0) : 类的构造函数（实例化） - 构造时间间隔（用于时间的加减）。
            - .days          属性（不带括号）
            - .seconds       属性 算出的是剔除完整天数后，剩下的零头时间对应的秒数（9000 秒）。
            - .microseconds  属性  
        - .total_seconds()   是方法（带括号），算出的是整段跨度（包括那1天）的总秒数（178200.0 秒）。


2. *time模块*（“纯函数包”，它里面没有那么多复杂的面向对象概念）

    - time() : 函数 - 获取当前时间戳（自1970年1月1日以来的总秒数）。时间戳:从格林威治时间（UTC）1970年1月1日 00:00:00 起至现在的总秒数。
    
    - sleep(secs) : 函数 - 让程序休眠/暂停指定的秒数（支持小数，如0.5秒）。
    
    - strftime(format) : 函数 - 将时间格式化为字符串（配合 %Y, %m, %H 等占位符）。
    
    - perf_counter() : 函数 - 精准性能计数器，专门用于测量代码的运行耗时。

    - strptime (p = Parse)：“从有到无”。把一个纯字符串解析成复杂的 Python 日期对象。

    - strftime (f = Format)：“从无到有”。把一个 Python 日期对象包装、格式化成你想要的

3. **辨析**

| 模块 | `strptime` 的输出结果 | `strftime` 的输入对象 |
| :--- | :--- | :--- |
| **`datetime`** | `datetime.datetime` 对象 | `datetime.datetime` 对象 |
| **`time`** | `time.struct_time` 元组 | `time.struct_time` 元组 |

```python 
from datetime import datetime, timedelta
# datetime = 坐标点（位置）
# timedelta = 向量（长度/位移）
import time
# 1. 获取当前时间
now = datetime.now()  # 带时区的 datetime 对象，内部存储了年、月、日、时、分、秒等数字信息。它可以直接进行数学运算（比如加减天数）。
print(now)  # 输出：2023-10-01 14:30:45.123456
# 2. 时间格式化与解析
formatted = now.strftime("%Y-%m-%d %H:%M:%S")  # 转为字符串
print(formatted)  # 输出：2023-10-01 14:30:45
parsed = datetime.strptime("2023-10-01", "%Y-%m-%d")  # 字符串转 datetime
print(parsed)  # 输出：2023-10-01 00:00:00
# 3. 时间差计算
delta = timedelta(days=1, hours=2)  # 1天2小时
future = now + delta
print(future)  # 输出：2023-10-02 16:30:45.123456
# 4. 时间戳转换
timestamp = time.time()  # 浮点数时间戳（秒）
print(timestamp)  # 输出：1696160445.123456
```

### 补充知识点
* strftime用法（大小写是因为同一个字母大小写不同代表的含义不同）
    1. 日期与年月日（最常用）

| 占位符 | 含义说明 | 示例输出 |
| :---: | :--- | :--- |
| **`%Y`** | 四位数的年份 | `2026` |
| **`%y`** | 两位数的年份 | `26` |
| **`%m`** | 月份（带前导零） | `05` |
| **`%d`** | 一个月中的第几天（带前导零） | `10` |

    2. 时间与时分秒

| 占位符 | 含义说明 | 示例输出 |
| :---: | :--- | :--- |
| **`%H`** | 24 小时制的小时 | `14` |
| **`%I`** | 12 小时制的小时 | `02` |
| **`%M`** | 分钟 | `30` |
| **`%S`** | 秒 | `05` |
| **`%p`** | 本地时间的 AM 或 PM | `PM` |

    3. 星期与月份别名

| 占位符 | 含义说明 | 示例输出 |
| :---: | :--- | :--- |
| **`%A`** | 完整的星期名称 | `Sunday` |
| **`%a`** | 简写的星期名称 | `Sun` |
| **`%B`** | 完整的月份名称 | `May` |
| **`%b`** | 简写的月份名称 | `May` |
| **`%w`** | 星期几（`0` 是周日，`6` 是周六） | `0` |

    4. 如果你不想一个一个点，这里有几个已经调配好的“套餐”：
    标准 ISO 格式： "%Y-%m-%dT%H:%M:%S"
        效果：2026-05-10T10:30:00
    美式日期： "%D" (等同于 %m/%d/%y)
        效果：05/10/26
    带中文的（需注意编码）： "%Y年%m月%d日"
        效果：2026年05月10日

## 三、练习
需求：编写一个时间管理工具，实现以下功能：

1. 倒计时器：输入小时、分钟、秒，开始倒计时并在结束时提示。
2. 工作日计算：计算两个日期之间的工作日数（排除周末）。
3. 时间差格式化：将时间差（如 timedelta）格式化为 "X天X小时X分钟"。
示例用法
```python
tool = TimeTool()
tool.countdown(0, 1, 30)  # 倒计时 1分30秒
print(tool.calculate_workdays("2023-10-01", "2023-10-07"))  # 输出：5（假设非闰年）
print(tool.format_timedelta(timedelta(days=2, hours=5)))  # 输出：2天5小时0分钟

````python
from datetime import datetime, timedelta
import time


class TimeTool:

    @staticmethod
    def countdown(x, y, z):
        delta = timedelta(seconds = 1)
        remaintime = timedelta(hours = x, minutes = y, seconds = z)
        print(f"计时开始: {remaintime}", end="", flush=True)
        while remaintime > timedelta(0): 
            time.sleep(1)
            #remaintime 是一个 timedelta 对象，而 0 是一个整数（int）。
            #让一个时间段对象直接和整数 0 进行 != 比较，Python虽然在某些版本不会直接报错，但由于它们类型不同，remaintime != 0 可能会永远为 True，从而导致程序无法在时间结束时退出，变成死循环。
            
           
            remaintime = remaintime - delta
            print(f"\r剩余时间: {remaintime}       ", end="", flush=True)
            # print (remaintime) 
            # 现在的程序每秒会打印新的一行，屏幕会变得很长。如果你想让它在同一行刷新（就像真正的时钟），可以修改 print 函数：
            # \r 表示回到行首，end="" 表示不换行
         
        else:
            print(f"\r剩余时间: {remaintime}   ")
            print("\r时间到")
    @staticmethod
    def calculate_workdays(date1, date2):
        parsed_date1 = datetime.strptime(date1,"%Y-%m-%d")
        parsed_date2 = datetime.strptime(date2,"%Y-%m-%d")
        #weekday1 = int(parsed_date1.strftime(w%)) #通过 strftime("%w") 得到的结果是一个字符串（"5"），而不是数字（5）。而且获得的周一到周日对应的是1234560
        weekday1 = parsed_date1.weekday() #这里使用.weekday是为了获得周一到周日对应0123456以便后续做判断。.weekday()获得的就是整数
        
        date_delta = (parsed_date2 - parsed_date1).days + 1 #datetime 对象不能直接加减数字，需要加减time_delta
        full_weeks = date_delta // 7
        remain_days = date_delta % 7
        
        remain_workday = 0
        for i in range(remain_days): # 
            # x+0 第一次循环输出开始日自身 x+1 x+2 ··· x+n 加n天相当于从x开始往后加6个时间段，得到一个时间点，也就是所有的时间点0到n = n - 0 +1
            current_weekday = (weekday1+i) % 7 #如果weekday1+i 大于7，current_weekday不是星期，我们就不能通过循环判断星期八是不是工作日
            if current_weekday < 5:
                remain_workday += 1
        total_workdays = 5*full_weeks + remain_workday
        # print(total_workdays)
        return total_workdays #如果是写在一个函数/方法里，建议用 return total_workdays 把结果返回出来，这样其他地方调用这个函数时才能拿到数据。
    @staticmethod
    def format_timedelta(timedelta1):
        daytime = timedelta1.days
        hourtime = timedelta1.seconds // 3600
        minutestime = (timedelta1.seconds % 3600) // 60
        return(f"{daytime}天{hourtime}小时{minutestime}分钟")

timetool1 = TimeTool()
timetool1.countdown(0, 0, 5)
````

# 🐍 Python 学习笔记｜第 33 关：正则表达式（re 模块实现高级字符串匹配）。
## 一、正则表达式（Regular Expression，简称 Regex 或 RegEx）
### 定义： 是一种用特定模式（Pattern）来匹配、查找、替换文本的特殊字符串。

### 1.元字符：
`.`  匹配除换行符外的任意字符，
`\d` 匹配任意一个数字（0-9）
`\D` 匹配任意一个非数字。
`\w` 匹配任意一个字母、数字或下划线（单词字符）。
`\W` 匹配任意一个非单词字符。
`\s` 匹配任意一个空白符（空格、制表符、换行符）。
`\S` 匹配任意一个非空白符。
`[]` 用于定义字符集
`[aeiou]` 匹配任意元音字母
`[^0-9]` 匹配非数字字符, `^` 在中括号内表示“非”

### 2.量词：

| 量词 | 含义 | 举例 | 匹配结果 |
| :--- | :--- | :--- | :--- |
| `*` | 出现 0 次或多次 | `hi*` | `h`, `hi`, `hii`, `hiii` |
| `+` | 出现 1 次或多次 | `hi+` | `hi`, `hii`（但不能是 `h`） |
| `?` | 出现 0 次或 1 次（可选） | `colou?r` | `color`, `colour` |
| `{n}` | 明确出现 n 次 | `\d{4}` | 连续 4 个数字（如 2026） |
| `{n,}` | 至少出现 n 次 | `\d{2,}` | 至少 2 个数字 |
| `{n,m}` | 出现 n 到 m 次 | `\d{2,4}` | 2 位、3 位或 4 位数字 |

匹配：含义一：提取/查找（从一大堆字里，把符合样式的“揪”出来）
     含义二：验证（检查整段字符串，是不是全盘符合这个样式）

### 3.位置锚定（Anchors）： 有时候我们不关心字符是什么，只关心它在什么位置。
`^`：匹配字符串的开头（注意与中括号里的“非”区分）。
`$`：匹配字符串的结尾。
`\b`：匹配一个单词边界（即单词和空格/标点之间的那个缝隙）。

例子：
^Apple：只能匹配以 Apple 开头的句子。
Apple$：只能匹配以 Apple 结尾的句子。
\bcat\b：只能匹配孤立的单词 cat，不会匹配到 catch 或 copycat。

### 4.捕获组（Grouping）
`()` 捕获组
**作用：**
1. 整体打包（分组）：把括号内的多个字符捆绑成一个独立单元，方便后面加量词（如 +、*、?）来控制它们整体出现的次数。
2. 开辟内存（捕获）：正则引擎在匹配时，会为每个括号分配一个“储物柜”（寄存器），把括号里匹配到的具体文本单独存起来，供后续调用。

**储物柜的编号规则（从左到右）**
正则引擎是以 左括号 ( 出现的先后顺序，从左到右依次编号为 1, 2, 3... 号捕获组。
group(0)：永远代表整个正则表达式匹配到的完整文本。
group(1)：代表第 1 个左括号对应的储物柜。
group(2)：代表第 2 个左括号对应的储物柜，以此类推。

举例：
正则表达式：((\d{4})-(\d{2}))
面对文本：2026-05

group(0) = 2026-05 （整个匹配）
group(1) = 2026-05 （最外层的左括号）
group(2) = 2026    （第二层左括号，年份）
group(3) = 05      （第三层左括号，月份）

用法：
①连带修饰：(abc)+ 表示字符串 abc 重复出现一次以上。
②捕获组：括号匹配到的内容会被暂时存在内存里，后续可以“复用”（称为反向引用）。
例：
(\d{4})-\1   
\d{4} 表示 4 个任意数字。                                
加了括号变成 (\d{4})，意思就是：“匹配 4 个数字，并且把它复制下来。”
\1 当正则表达式引擎看到 () 时，它不仅会匹配里面的内容，还会默默把匹配到的文字复制下来，存进一个叫 \1 的临时变量里。（如果有多个（），那么就会自动编号\2 \3 \4）
普通的字符（比如字母、数字、横杠 -、空格等）默认就是它们自己，不需要任何特殊的匹配符号。

### 5. 分支条件（OR）

`|` 管道符 表示“或”。
cat|dog 可以匹配 cat 或者 dog。
T(o|e)st 可以匹配 Tost 或 Test。

**高级用法**
`动态正则拼接（用 | 取代 for 循环）`
当你需要从数据库的一个列表里查找匹配项时，用 for 循环一条条去对比文本，如果数据库有 1 万条数据，电脑就要扫描文本 1 万次，效率极低。而通过将列表元素用 | 拼接成一条大正则，电脑只需要扫描一遍文本，就能一次性把所有词都找出来。

例:
```python
database_list = ["apple", "banana", "orange"]

# 1. map(re.escape, database_list) -> 给列表里的每个词戴上安全防护罩
# 2. "|".join(...) -> 用竖线把它们串起来
pattern = "|".join(map(re.escape, database_list))

print(pattern)  # 得到动态正则: "apple|banana|orange"
```

### 6. 转义字符（Escape）
如果你想匹配元字符本身（比如想在文本里找一个真正的 . 或 *），需要在前面加反斜杠 \。
`\.` 匹配点号。
`\*` 匹配星号。
`\\` 匹配反斜杠本身。

### 7.re 模块方法：
import re
目标文本：text = "张三的电话是13811112222，李四的电话是13933334444。"

`re.match()`   仅从字符串起始位置匹配；如果开头第一秒没对上，它立刻放弃，直接返回 None（啥也没找到）。
用法： re.match(r"\d{11}", text)
结果： 啥也找不到（返回 None）。因为 text 的开头是“张三的...”，不是数字

`re.search() ` 从头到尾扫描整句话，直到找到第一个符合条件的就立刻收工。
用法： result = re.search(r"\d{11}", text)
结果： 找到了张三的电话 13811112222。后面李四的电话它就地忽略了，因为已经完成了任务。
怎么拿出来： 用 result.group() 就能把抓到的字符串吐出来。
适用场景： 你只在乎文本里“有没有”某个东西，或者只需要找第一个出现的关键词。

`re.findall()` 它是最常用的方法之一。它会扫描整个字符串，把所有符合条件的全部抓出来，打包成一个 List（列表） 给你。
用法： results = re.findall(r"\d{11}", text)
结果： 返回一个列表：['13811112222', '13933334444']。

`re.sub()`   sub 是 substitute（替换）的缩写。它不仅能找到符合规则的文本，还能顺手把它换成你指定的新内容。
用法： new_text = re.sub(r"\d{11}", "XXXXXXXXXXX", text)
结果： 传回一句话："张三的电话是XXXXXXXXXXX，李四的电话是XXXXXXXXXXX。"

`re.compile()` 前面 4 种方法，每次执行时，电脑都要临时去理解你写的那个 r"\d{11}" 到底是什么意思，这会消耗一点点时间。如果你有一千万行文本要处理，每次都临时理解就太慢了。re.compile() 就是先把通缉令“翻译/编译”好，存成一个固定的工具，以后直接用这个工具去搜，速度极快。
用法： 

`re.escape()`自动把字符串里所有有特殊含义的字符前面都加上反斜杠进行转义，让它们变回纯文本。
用法：safe_word = re.escape(apple.pie*)
结果：print(safe_word)  # 输出: apple\.pie\* （自动帮你在 . 和 * 前面加了反斜杠）

`r` 上述五种函数中的r 代表 Raw String（原始字符串），让字符串内部的反斜杠 \ 失去转义功能，变成一个普普通通的文本字符。
```python
# 1. 先编译好规则，做成一个“专用扫描枪”
phone_scanner = re.compile(r"\d{11}")
# 2. 以后直接用这个扫描枪去扫不同的文本，不用再写规则了
res1 = phone_scanner.findall(text1)
res2 = phone_scanner.findall(text2)
```

## 知识点补充一
`map()`: 
**定义**: 将一个函数依次作用到一个或多个可迭代对象（如列表、元组等）的每一个元素上，并返回一个迭代器。
**语法**: map(function, iterable, ...)

1. **基础用法**

```python 
numbers = [1, 2, 3, 4]
# 使用 lambda 匿名函数
result = map(lambda x: x**2, numbers)
# 注意：map 返回的是一个 map 对象（迭代器），需要转成 list 才能看到结果
print(list(result))  # 输出: [1, 4, 9, 16]
```

2. **处理多个可迭代对象:** 如果传入多个列表，函数会同时接收来自各个列表的元素（按最短的列表对齐）。

```python
list1 = [1, 2, 3]
list2 = [10, 20, 30]
# 两个列表对应元素相加
result = map(lambda x, y: x + y, list1, list2)
print(list(result))  # 输出: [11, 22, 33]
```

3. **避坑指南（惰性求值）**
map() 返回的是一个迭代器（Iterator），这意味着它不会立即在内存中生成所有数据，只有在你遍历它（如用 for 循环或 list() 转换）时才会计算。此外，迭代器只能被消耗一次，转成列表后，再次对该 map 对象转列表就会得到空列表。

`join()`
**定义**: 将一个可迭代对象（如列表、元组）中的所有元素（**必须是字符串**）拼接成一个新的字符串。
**语法**: "分隔符".join(iterable)

1. **基础用法**

```python
words = ["Python", "is", "awesome"]
# 用空格拼接
sentence = " ".join(words)
print(sentence)  # 输出: "Python is awesome"
# 用连字符拼接
hyphenated = "-".join(words)
print(hyphenated)  # 输出: "Python-is-awesome"
```

## 知识点补充二
`贪婪匹配 (Greedy)`
* **语法**：`*`、`+`、`{n,}`（默认模式）
* **特性**：尽可能多地匹配字符，直到最后一次出现匹配项的地方。
* **示例**：
    * **文本**：`<div>苹果</div><span>香蕉</span><div>橙子</div>`
    * **正则**：`<div>.*</div>`
    * **结果**：`<div>苹果</div><span>香蕉</span><div>橙子</div>`
`非贪婪匹配 (Lazy)`
* **语法**：在量词后加问号，如 `*?`、`+?`、`??`
* **特性**：尽可能少地匹配字符，只要达成匹配条件立即停止。
* **示例**：
    * **正则**：`<div>.*?</div>`
    * **结果**：`<div>苹果</div>` 和 `<div>橙子</div>`

`正向预查 (Positive Lookahead)`
* **定义**：这是一种“零宽断言”，用于在不消耗字符的情况下，检查当前位置之后的内容。
* **语法**：`(?=pattern)`
* **特点**：
    1.  **非捕获**：检查的内容不会出现在最终结果里。
    2.  **零宽度**：匹配后光标回到原位，不占用匹配长度。

* **应用场景**
    1. 后缀过滤：匹配后面跟着特定字符的内容。
            正则：`\d+(?=元)`
            文本：`价格100元` -> 匹配结果：`100`
    2. 多重条件校验（如密码强度）：
            正则：`^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$`
            含义：必须同时包含大写字母、小写字母和数字，且长度至少8位。

# 🐍 Python 学习笔记｜第 34 关：并行编程：多线程与多进程的基本概念与应用。

## 一、多线程 vs 多进程
1. 进程（Process）：好比一家工厂。工厂之间相互独立，各自拥有独立的土地、原材料仓库（独立的内存空间）。 
2. 线程（Thread）：好比工厂里的工人。一个工厂里可以有多个工人，他们共同使用工厂里的设备和原材料（共享内存空间）。
3. 多线程（Multithreading）是计算机科学中的一个核心概念。简单来说，它是指在一个程序（进程）中，同时运行多个执行路径（线程）来共同完成任务的技术。
4. 多进程（Multiprocessing）是计算机科学中的一个核心概念。简单来说，进程（Process）是操作系统分配资源和调度的基本单位，而多进程就是指在系统中同时运行多个独立的程序（进程）。
5. 在通用的计算机世界里，多线程和多进程都能利用多核 CPU；但在 Python 的世界里，因为 GIL 的存在，只有多进程才能真正利用多核 CPU 进行并行计算。
6. 在 GIL 的存在下，Python 的多线程确实无法利用多核 CPU 来提高“计算”效率；但是在处理“输入输出（I/O）”任务时，它依然能大幅提高效率。
    * I/O 密集型任务（多线程：大显身手 🚀）
        什么是 I/O 密集型：网络请求（爬虫、下载文件）、读写硬盘文件、查询数据库等。
        为什么有用：这些任务的特点是“等”。当线程 A 发起一个网络请求时，它需要等待服务器响应，此时它不需要 CPU 进行计算。一旦线程 A 开始等待 I/O，Python 就会主动让线程 A 释放 GIL 锁，把麦克风交给线程 B。线程 B 就可以趁机开始工作。结果就是你不需要等第一张图片下载完再发起第二张的请求，而是可以同时发起 10 个下载请求。效率提升了几倍甚至几十倍。
    * 计算密集型任务 / CPU 密集型任务（多线程：确实没用 ❌）
        什么是计算密集型：视频解码、图形渲染、大规模数学计算（如矩阵运算、机器学习模型训练）。
        为什么没用：这些任务需要 CPU 疯狂运转。线程 A、B、C 都需要抢着执行代码，但因为 GIL 的限制，它们只能排队轮流用同一个 CPU 核心。更糟糕的是由于线程之间来回切换（抢麦克风）需要消耗额外的系统资源，在 Python 中用多线程跑计算密集型任务，甚至可能比单线程跑得还要慢。
7. 总结

| 特性 | 多线程 (Multithreading) | 多进程 (Multiprocessing) |
| :--- | :--- | :--- |
| **内存空间** | 共享同一内存空间（全局变量可访问） | 独立内存空间（数据通过 IPC 通信） |
| **GIL 限制** | 受 Python 全局解释器锁（GIL）限制，CPU 密集型任务效率低 | 无 GIL 限制，适合 CPU 密集型任务 |
| **启动开销** | 低 | 高 |
| **适用场景** | I/O 密集型任务（如网络请求、文件读写） | CPU 密集型任务（如数学计算、数据分析） |

## 二、threading 模块 & multiprocessing 模块

1. threading 模块：官方提供的“多线程”工具箱。
`threading.Thread`：类 - 这是整个模块里最核心的“线程调度器”。**在 Python 中，一个 Thread 对象的实例，就代表一个独立的线程。**
    作用：用来打包你的任务，并告诉操作系统：“嘿，帮我单独开辟一个新线程来跑这个任务！”
    核心参数：
        * `target`：任务目标。你得告诉这个线程要去执行哪个函数。注意，这里只传函数名字，绝对不能加括号（加了括号函数就立刻在主线程执行了，而不是交给子线程）。
        * `args`：任务参数。如果你的函数需要传参，必须用元组 (Tuple) 的形式传给 args。   
    `.start()`：类方法 - 启动键。告诉操作系统：“这个线程可以开始跑了！”调用后，新线程会立刻独立飞出去执行它的任务，而你的主程序（主线程）会继续往下走，不会在这里傻等。
    `.join()`： 类方法 - 同步键（强行等待）。告诉主程序：“你先别急着往下走，给我在这里等着，直到这个子线程把活干完了，你再继续！”
示例：
```python
# 第一步：把工具箱搬过来
import threading
import time

# 第二步：定义好线程要去干的“活”（函数）
def watch_movie(movie_name):
    print(f"【子线程】开始看电影: {movie_name}")
    time.sleep(2)  # 模拟看了2秒钟
    print(f"【子线程】电影 {movie_name} 看完了！")

# 第三步：主程序开始运行（主线程）
if __name__ == '__main__':
    print("【主线程】主程序启动了...")
    
    # 第四步：从工具箱里拿出一个 Thread 工人，把活交给他
    # target 指向函数名，args 传入元组参数 (注意结尾的逗号)
    t1 = threading.Thread(target=watch_movie, args=("《黑客帝国》",))
    
    # 第五步：按下启动键
    t1.start()  
    
    # 此时，t1 线程已经飞出去看电影了，主线程不会卡住，会立刻往下走
    print("【主线程】我已经把看电影的任务交出去了，我自己先喝杯茶...")
    
    # 第六步：按下同步键，主线程在这里死等 t1 完工
    t1.join()  
    
    print("【主线程】好了，子线程看完了，主程序彻底退出。")
```

2. multiprocessing 模块：官方提供的“多进程”工具箱。
`multiprocessing.Process`: 类 - 这是多进程里的“进程创建器”。**一个 Process 对象就代表操作系统中的一个独立进程。**
    作用：克隆一个全新的 Python 解释器环境，拥有自己独立的内存。
    核心参数：
        * `target`：任务目标。需要传入子进程要去执行的函数名字。注意：只传函数名，绝对不能加括号（加了括号函数就会立刻在当前的主进程中执行，而不会交给子进程）。
        * `args`：任务参数。如果你的目标函数需要传参，必须用元组 (Tuple) 的形式传给 args（例如单个参数要写成 args=("参数1",)，末尾逗号不能省）。
    `.start()`：类方法 - 启动键。告诉操作系统：“这个子进程可以开始跑了！”调用后，操作系统会正式创建并启动子进程去执行它的任务。此时主程序（主进程）会继续往下走，不会在这里傻等。
    `.join()`： 类方法 - 同步键（强行等待）。告诉主程序（主进程）：“你先别急着往下走，给我在这里等着，直到这个子进程把活全部干完了，你再继续往下运行！”
示例：
```python
# 第一步：导入多进程工具箱
from multiprocessing import Process
import time

# 第二步：定义高强度的计算任务
def heavy_calculation(number):
    print(f"【子进程】开始计算 {number} 的平方和...")
    result = sum(i**2 for i in range(number))
    print(f"【子进程】计算完成！结果是: {result}")

# 第三步：Windows 下多进程必须加这句安全锁！
if __name__ == '__main__':
    print("【主进程】开始...")
    
    # 第四步：实例化一个进程对象
    p1 = Process(target=heavy_calculation, args=(1000000,))
    
    # 第五步：启动进程
    p1.start()
    
    # 第六步：等待进程结束
    p1.join()
    
    print("【主进程】结束。")
```

3. `if __name__ == '__main__':` : **进程专属**的“安全锁”， 防止代码在被当作模块导入时误执行。
`__name__` 变量记录的不是“文件叫什么名字”，而是“这个文件目前扮演什么角色”,主进程中，__name__ == '__main__'

示例：
**无锁**
```python
# my_program.py
from multiprocessing import Process
import time
print("--- 这一行代码被执行了 ---")
def my_task():  # 此时没有进程执行这个函数
    print("子进程在干活...")
# 没加锁，直接暴露在外面
p = Process(target=my_task) #创建一个进程对象
p.start()#开始进程。
```
【第 1 阶段：主进程启动】
你在电脑上双击运行了 my_program.py。
操作系统创建了【主进程】。
【主进程】开始从上到下读代码：
读到第 4 行，控制台打印出：--- 这一行代码被执行了 ---。
读到第 9 行，遇到了 p.start()。
【主进程】对系统说：“帮我开一个【子进程 A】来运行 my_task 吧！”

【第 2 阶段：子进程 A 诞生】
操作系统创建了【子进程 A】。
【核心关键点】：【子进程 A】刚生出来的时候脑子是空的，它不知道 my_task 是什么。Windows 系统说：“那没办法，【子进程 A】，你把你爸的文件 my_program.py 重新打开读取一遍吧。”
于是，【子进程 A】也开始从上到下读代码：
读到第 4 行，控制台再次打印出：--- 这一行代码被执行了 ---。（这是第二次看到了）
恐怖的事情发生了：【子进程 A】读到第 9 行，它也遇到了 p.start()！
【子进程 A】误以为这也是它的任务，于是对系统说：“帮我也开一个【子进程 B】来运行 my_task 吧！”
【第 3 阶段：无限套娃，系统崩溃】

**有锁**
```python
from multiprocessing import Process
print("--- 这一行在锁外面 ---")
def my_task():
    print("子进程在干活...")
if __name__ == '__main__':  # 这是安全锁
    print("--- 这一行在锁里面 ---")
    p = Process(target=my_task)
    p.start()
```
【主进程】启动：
它是老大，它的变量 __name__ 确实等于 '__main__'。
它打印了锁外面的字。
遇到 if 判断，条件成立！成功进入锁内，打印锁里面的字，并创建了【子进程 A】。
【子进程 A】启动：
Windows 依然让它重新读取 my_program.py。
【子进程 A】打印了锁外面的字（因为锁外面的代码谁都能执行）。
转折点：当它走到 if __name__ == '__main__': 时，因为它是被偷偷导入进来的，它的 __name__ 变成了 'my_program'，不等于 '__main__'！   （解释：Windows 系统的多进程机制（spawn 模式）有一个雷打不动的硬性规定：子进程必须在后台重新 import 一次主文件，用来初始化自己的环境，请注意这个词：import（导入）。对于主进程来说： 它是你亲手启动的，所以主进程的工牌（__name__）写着："__main__"。对于子进程来说： 它不是你手动启动的，它是被多进程系统在后台通过 import my_program 这种方式偷偷加载进来的。因为子进程是通过 import 加载这个文件的，Python 解释器就会按照规则，把子进程环境里的那块工牌（__name__）改写成："my_program"。 ）
结果：if 条件不成立！【子进程 A】直接跳过了锁里面的 p.start() 代码。
循环在这里被彻底斩断。
安全结束：
【子进程 A】由于没有触发新的 p.start()，它高高兴兴地去执行它的 my_task() 函数，打印出 "子进程在干活..."，然后安全退出。


**问**
跳过 if 之后，子进程去哪了？
既然跳过了 if 块，下面又没有其他代码了，按照正常的 Python 逻辑，子进程不是应该结束运行了吗？
并没有！因为子进程在出生时，身上还带了一个“隐藏任务”。
还记得前面提到的 Windows spawn 机制的第二步吗？
主进程在调用 p.start() 的那一刻，就已经通过系统底层，把一个任务小纸条（序列化数据）死死地贴在了子进程的脑门上。
这个小纸条上写着：
“当你把代码文件全部加载完之后，不要退出！立刻在内存里找到一个叫 my_task 的函数，然后去执行它！”

### 补充:windows的spawn机制
第一步：白手起家
Windows 操作系统接到命令，在后台启动一个全新、空白的 Python 解释器进程。这个进程现在和你的主程序没有任何血缘关系，它只知道自己是个 Python 进程。
第二步：打包行李（Pickle 序列化）
主进程把需要传给子进程的参数、数据、以及任务目标（比如 target=my_task），通过 Python 的 pickle 模块打包成二进制数据，顺着系统管道丢给这个新的子进程。
第三步：重读代码（这就是为什么要加锁的原因！）
新进程拿到行李后，发现自己脑子里还是没有 my_task 的代码定义。于是它必须做一件事：在后台隐式地重新 import 你的主脚本文件。
这一步的目的，是为了把文件里所有的 def（图纸）和全局变量在子进程的内存里重新生成一遍。
因为是 import 导入，所以子进程的 __name__ 变成了文件名，而不是 __main__。 从而触发了我们前面讲的“安全锁”机制。
第四步：解包并开工
环境初始化完成后，子进程解开主进程丢过来的行李包（反序列化），找到了 my_task 函数在当前新内存里的地址，最后真正开始执行 my_task()。

4. 待补充：多进程的“数据孤岛”与通信 - **IPC（进程间通信）**
“多进程拥有独立的内存空间”。这是一个非常重要的结论。但在实际开发中，如果子进程算出了结果，主进程怎么拿到？ 既然内存不共享，它们该怎么说话？
`multiprocessing.Queue（进程安全队列）`
`multiprocessing.Value / Array（共享内存）`

## 三、Python 线程同步

1. 临界区 (Critical Section)：在多线程程序中，多个线程共享并同时进行修改的内存区域或变量（如全局变量、共享文件、数据库连接等）。如果不对临界区进行保护，多个线程同时读写就会导致数据错乱

2. 竞态条件 (Race Condition):当多个线程同时访问和操作同一块数据，且最终执行结果取决于线程运行的精确时序（谁快谁慢）时，就形成了竞态条件。

根源：许多高级语言中的单行代码（如 count += 1），在底层其实是由“读取-修改-写入”等多个 CPU 原子指令组成的。线程在这些中间步骤之间发生切换，就会导致数据覆盖。

```python （错误示例:不加锁的惨痛教训）
import threading
counter = 0
def unsafe_increment():
    global counter
    for _ in range(100000):
        # 临界区：没有任何保护，多线程交替执行这行代码会引发数据覆盖
        counter += 1
# 启动两个线程同时改一个变量
t1 = threading.Thread(target=unsafe_increment)
t2 = threading.Thread(target=unsafe_increment)

#调用
t1.start()
t2.start()
t1.join()
t2.join()
print(f"最终结果: {counter}")  # 结果往往小于 200000，且每次运行都不一样
```

## 四、 互斥锁和可重入锁

1. 互斥锁 (Mutex Lock): 能保证同一时刻只有一个线程进入临界区。推荐使用 with lock: 上下文管理器，它能确保即使中间代码报错，锁也会被安全释放，防止死锁。
锁定 (Acquire)：线程尝试获取锁。如果锁未被占用，该线程获取成功并将其锁定，进入临界区；如果锁已被其他线程占用，该线程将被阻塞（挂起等待）。
释放 (Release)：当持有锁的线程离开临界区时，必须释放锁。此时操作系统会唤醒其他等待该锁的线程去重新竞争。

`threading.Lock 类`： 在Python 中用来实现互斥锁
    `with lock: `: **隐式管理（上下文管理器 - 推荐）**
    优点：安全。无论临界区代码是否发生异常、报何种错误，Python 都会在离开 with 块时自动释放锁，有效避免死锁。
    `lock.acquire() & lock.release()`: **显式管理（手动控制）**
    注意：必须配合 try...finally 结构使用，确保 release() 在 finally 块中执行，否则代码报错会导致锁永远无法释放。


```Python （使用 with 自动管理锁）
import threading
counter = 0
cell_lock = threading.Lock()  # 创建互斥锁
def safe_increment():
    global counter
    for _ in range(100000):
        # 使用 with 语句，进入时自动 acquire()，离开时自动 release()
        with cell_lock:
            counter += 1

t1 = threading.Thread(target=safe_increment)
t2 = threading.Thread(target=safe_increment)
t1.start()
t2.start()
t1.join()
t2.join()
print(f"安全加密后的结果: {counter}")  # 必定输出：200000
```
内部过程：有了 with cell_lock:，这三步操作被紧紧包裹在了一起，变成了一个不可分割的整体。线程 t1 抢到了锁，开始执行“读取 -> 计算 -> 写入”。即使操作系统在中途把 CPU 切换给了线程 t2，t2 走到 with cell_lock: 时，发现锁被占了。t2 无法执行任何操作，只能挂起等待。CPU 只能切回 t1，让 t1 顺利把 101 写回内存并释放锁。t2 终于拿到了锁，读取到最新值 101，再安全地加到 102。

2. 终极避坑：死锁（Deadlock）的经典制造与防御
既然学了锁（Lock），就必须知道怎么防死锁。最经典的死锁叫 “哲学家就餐问题” 或 “互相等待”。

```python 错误示范 完美死锁
#线程 A 拿着锁 1，想去拿锁 2；线程 B 拿着锁 2，想去拿锁 1。两个人都咬死不放，程序瞬间永久卡死。

import threading
import time
lock1 = threading.Lock()
lock2 = threading.Lock()
def thread_A():
    with lock1: #这个锁的作用是保证在A线程
        time.sleep(0.1) # 故意等待，让线程B锁住lock2
        with lock2:
            print("A 完成了任务")
def thread_B():
    with lock2:
        time.sleep(0.1) # 故意等待，让线程A锁住lock1
        with lock1:
            print("B 完成了任务")

t1 = threading.Thread(target=thread_A)
t2 = threading.Thread(target=thread_B)
t1.start()
t2.start()
```

真实场景：银行转账（避无可避的死锁高发区）
假设你现在是支付宝的后端工程师，用户 A 和用户 B 的账户是两个独立的数据库对象（就像两把锁）。
现在发生了两件同时进行的事：
事件一：用户 A 同时 往 用户 B 转账 100 元。（线程 A：需要同时锁住 A 和 B 的账户，防止转账中途数据被别人改了）
事件二：用户 B 同时 往 用户 A 转账 50 元。（线程 B：需要同时锁住 B 和 A 的账户）
这时候，你就没法让他们“分开跑”了：
线程 A 率先锁住了 账户 A。
线程 B 几乎同一微秒锁住了 账户 B。
接下来，线程 A 去申请锁 账户 B（卡住，等 B 放手）。
线程 B 去申请锁 账户 A（卡住，等 A 放手）。
轰隆！真实的死锁在你的业务里爆发了。 用户的手机界面会一直转圈圈，转账永远无法完成。


**防御策略：**
固定加锁顺序（推荐）：所有线程必须按照相同的顺序获取锁。比如线程 A 和线程 B 都必须先拿 lock1 再拿 lock2，死锁直接破解。

使用 timeout 放弃抵抗：不要用 with lock:，而是用 lock.acquire(timeout=2)。如果 2 秒钟拿不到锁，就主动报错或执行别的逻辑，强行打破死锁僵局。

2. 可重入锁 (RLock)
知识点：普通的 Lock（互斥锁） 无法被同一个线程连续获取多次（会把自己锁死）。而 RLock（可重入锁）允许同一个线程多次执行 acquire()，适合用于递归函数或者复杂的嵌套调用。

`threading.RLock()`： 在Python 中用来实现互斥锁
    `with lock: `: **隐式管理（上下文管理器 - 推荐）**
    优点：安全。无论临界区代码是否发生异常、报何种错误，Python 都会在离开 with 块时自动释放锁，有效避免死锁。
    `lock.acquire() & lock.release()`: **显式管理（手动控制）**
    注意：必须配合 try...finally 结构使用，确保 release() 在 finally 块中执行，否则代码报错会导致锁永远无法释放。

```python 普通互斥锁 
#普通互斥锁的规则非常死板：不管是谁，只要锁现在是关着的，想再进就必须等它打开。 哪怕现在占用这把锁的人就是你自己，也不行！
import threading
lock = threading.Lock()  # 普通互斥锁
def master_task():
    with lock:  # 第一次加锁（成功）
        print("完成了主任务...")
        sub_task()  # 调用子任务
def sub_task():
    with lock:  # 第二次加锁（卡死！）
        print("完成了子任务...")

master_task()
```

```Python 可重入锁
#可重入锁内部多了一个“记账本”，它不仅记录锁是不是关着的，还记录了是谁锁的以及锁了几次。
import threading
# 如果这里用 threading.Lock()，程序运行到 bar() 就会永久卡死
nested_lock = threading.RLock()
def foo():
    with nested_lock:
        print("foo 拿到了锁，准备调用 bar")
        bar()  # 在同一线程中，bar 内部再次去拿这把锁
def bar():
    with nested_lock:  # 同一个线程再次拿锁成功，不会死锁
        print("bar 也成功拿到了同一把锁")

t = threading.Thread(target=foo)
t.start()
t.join()
```
**RLock 的工作原理**：当一个线程第一次获取 RLock 时，锁的计数器变成 1。同一个线程再次获取这把锁时，锁不会阻塞它，而是让它直接进去，并将计数器加 1（变成 2）。每退出一层 with 代码块（执行一次 release），计数器就减 1。只有当计数器归零（变为 0）时，这把锁才算真正解开，其他线程才有机会抢到它。

3. Lock & Rlock对比

| 锁类型 | 核心特点 | 适用场景 |
| :--- | :--- | :--- |
| **`Lock`**<br>(普通互斥锁) | 性能稍高，规则简单。<br>**不准**同一个线程连续 `acquire()`，否则会造成死锁。 | 绝大多数普通的、**没有函数嵌套调用**的多线程修改数据场景。 |
| **`RLock`**<br>(可重入锁) | 性能略低一点（内部需要维护持有者线程和计数器）。<br>**允许**同一个线程重复 `acquire()`。加锁多少次，就必须释放（release）多少次，否则锁依然不会解开。 | **递归函数**，或者多个**相互调用且都加了锁**的复杂类方法。 |

## 五、信号量 (Semaphore) ,事件 (Event) & 条件变量 (Condition)

1. 信号量：一种用于控制同时访问特定资源的线程或进程数量的同步原语。互斥锁只允许 1 个线程访问。而信号量允许 N 个线程同时访问资源，超过 N 的线程必须排队。这常用于对稀缺资源的限流（如数据库连接池、爬虫并发控制）。
acquire()（获取/占用）： 计数器 减 1。如果计数器已经为 0，说明资源满了，当前线程就会阻塞（排队），直到其他线程释放资源。
release()（释放/离开）： 计数器 加 1。同时会唤醒一个正在排队等待的线程。

`threading.Semaphore()类`: 
    `with Semaphore: `: **隐式管理（上下文管理器 - 推荐）**
    优点：安全。无论临界区代码是否发生异常、报何种错误，Python 都会在离开 with 块时自动释放锁，有效避免死锁。
    `Semaphore.acquire() & Semaphore.release()`: **显式管理（手动控制）**
    注意：必须配合 try...finally 结构使用，确保 release() 在 finally 块中执行，否则代码报错会导致锁永远无法释放。

实例（限制最多 3 个线程同时工作）：

```Python
import time
import threading

pool_semaphore = threading.Semaphore(3) #限制最多允许 3 个线程同时进入

def worker(worker_id):
    with pool_semaphore:
        print(f"工人 {worker_id} 占用了坑位。")
        time.sleep(1)  # 模拟工作耗时
        print(f"工人 {worker_id} 工作结束，释放坑位。")

threads = [threading.Thread(target=worker, args=(i,)) for i in range(5)] # 开启 5 个线程，但因为信号量限制，前 3 个先执行，后 2 个等有人出来再执行

for t in threads:
    t.start()
for t in threads:  
    t.join()  #让主线程（你的主程序）在原地死等，直到被调用的线程完全执行结束。
print("📊 今日工作统计：所有工人已安全收工！")
```
### 问一：为什么加t.join()
如果加了 join()： 这行字会雷打不动地在最后打印。
如果不加 join()： 因为主程序跑得飞快，这行字会最先跳出来（甚至在“工人 0 占用坑位”之前），整个程序的逻辑和日志顺序就全乱套了。

### 问二:什么是主线程什么是子线程和进程的关系是怎样的
它既是一个进程，也是一个线程。它们是“包含关系”，同时存在！你运行的这段程序，在操作系统眼里是一个“进程”；而在程序内部执行代码时，它是靠主线程和 5 个子线程共同完成的。

主线程 ＝ 你的整段代码
子线程 ＝ 你安排出去干活的“打工人”
1. 从【进程（Process）】的角度看：它只有 1 个
当你点击运行（Run）的那一瞬间：操作系统在内存里划出了一块独立空间，给它分配了一个进程 ID（叫做 PID）。这个独立空间里包含了一条属于它的 GIL 锁、一个 Python 解释器、以及你的全部代码。在任务管理器里，它只占用一个进程的名额。
2. 从【线程（Thread）】的角度看：它一共开辟了 6 个
在这个独立的进程内部，代码开始运转，数人头的话一共产生了 6 个线程：
第 1 个：主线程（总经理）
由操作系统默认启动。负责从第一行代码一直往下读，创建信号量、创建列表。
第 2 到第 6 个：5 个子线程（打工人工人 0 ~ 4）
当主程序执行到 t.start() 循环时，主线程在进程内部一口气分裂出了 5 个子线程。

### 问三:那为什么不都用线程，要开辟新进程
因为线程有两个致命的“先天缺陷”，逼得我们不得不使用进程：“一人生病，全家吃席” 和 “一核有难，九核围观”。
1. 致命缺陷一：线程不安全，一挂全挂（稳定性问题）
多线程最大的特点，就是共享同一个内存大院。这个特点是把双刃剑：
好处： 大家传数据极快，说话都能听见。
坏处： 如果其中一个子线程犯了严重错误（比如底层 C 语言层面的段错误 Segment Fault、或者内存溢出），整个进程（也就是主程序和所有其他线程）会瞬间一起崩溃。
2. 致命缺陷二：Python 的 GIL 锁限制（性能问题）
还记得我们前面聊到的 GIL（全局解释器锁） 吗？
因为 GIL 的存在，在 Python 里，无论你的电脑 CPU 是 8 核、16 核还是 64 核，多线程在任意微观瞬间都只能榨干其中 1 个核心的算力。
如果你做的是 I/O 密集型任务（比如爬虫、等网络响应、读写文件）： 多线程很划算。因为大部分时间大家都在 sleep 或者等网络，不需要拼命计算，1 个核足够应付。
如果你做的是 CPU 密集型任务（比如视频转码、图像处理、AI 矩阵运算、大规模数学计算）：
多线程就会变成“一核有难，九核围观”。明明有 8 个核，只有 1 个在疯狂打满，另外 7 个闲得冒烟。

### 问四:那如果不加t.join主线程结束子线程还没结束就会卡住吗
不会卡住，反而会“瞬间各奔东西”。这里有两种完全不同的结局，取决于你有没有把子线程设置为“守护线程（Daemon）”。在默认情况下，你的代码不会卡住，子线程会继续把活干完，但主程序的逻辑会提前跑完。
**守护线程**：存疑

3. 多线程/多进程同步中的事件（threading.Event / multiprocessing.Event）
知识点：用于线程间的基本通知机制。它就像一个红绿灯：一个线程负责控制红绿灯（设置或清除信号），其他线程则在红灯时等待，绿灯时通行。


`threading.Event` 核心机制 - 内部维护着一个内置标志（Flag），初始状态为 False（红灯）。
    `wait(timeout=None)`：如果标志为 False，线程会阻塞（卡住）在这里；如果为 True，则直接通过。
    `set()`：将标志设置为 True（变绿灯），并唤醒所有正在等待的线程。
    `clear()`：将标志重置为 False（变回红灯）。
    `is_set()`：判断当前标志是否为 True

```Python  实例（主线程通知子线程开始跑数）：
import time
import threading

start_signal = threading.Event()

def run_task(name):
    print(f"线程 {name} 已准备就绪，等待主线程发令...")
    start_signal.wait()  # 卡在这里，直到主线程调用 set()
    print(f"【信号已到】线程 {name} 开始疯狂工作！")

t1 = threading.Thread(target=run_task, args=("A",))
t2 = threading.Thread(target=run_task, args=("B",))
t1.start()
t2.start()

time.sleep(2)  # 主线程模拟准备工作
print("主线程准备完毕，拉响警报！")
start_signal.set()  # 唤醒所有在 wait 的子线程

t1.join()
t2.join()
```
**注意：**multiprocessing.Event（多进程事件）的用法与此完全一致，专门用于多进程之间的同步。

6. 条件变量 (Condition Variable)
知识点：如果说互斥锁（Lock）解决了“同时访问”的冲突，那么条件变量解决的就是“等待某个特定条件发生”的协调问题。它让线程能够在没有满足特定条件时挂起（不消耗 CPU），并在条件满足时被其他线程唤醒。

`threading.Condition` - 核心机制 - threading.Condition 内部其实包含了一把锁（默认是 RLock 递归锁）。使用条件变量时，有一个铁律：任何线程在改变共享状态或等待条件之前，都必须先获取这把锁。
    `with cond:` **隐式管理（上下文管理器 - 推荐）**
    `cond.acquire()` & `cond.release()` **显式管理（手动控制加锁和释放锁）**
    `cond.wait(timeout=None)`: 调用 wait() 时，当前线程会释放锁，并进入阻塞等待状态。
    `cond.wait_for(predicate, timeout=None)`: 这是 wait() 的高级封装（Python 3.2+ 引入）。predicate 是一个返回布尔值的可调用对象（如 lambda 函数）。它相当于：while not predicate(): cond.wait()。**存疑**
    `cond.notify(n=1)`: 唤醒正在等待该条件的一个线程（如果有多个线程在等，随机选一个）。注意： 唤醒并不意味着被唤醒的线程立刻执行，它必须等到当前线程释放锁后，抢到锁才能继续。
    `cond.notify_all()`: 唤醒所有正在等待该条件的线程。

```Python 实例（简单的生产者与消费者协作）：
import time
import threading

con = threading.Condition() # 创建条件变量

product_count = 0  # 共享的货物库存
def producer():
    global product_count #指代全局变量
    with con: #使用with语句，进入自动上锁，离开时自动释放锁。
        print("生产者：正在生产商品...") #自动上锁，其他线程不能进入
        time.sleep(1) #模仿生产
        product_count += 1 #产出商品一件
        print("生产者：商品已上架，通知消费者！") #通知消费者
        con.notify()  # 唤醒一个正在等待的消费者
    print("已经没有锁了") #程序执行流离开了 with 缩进的代码块，锁就会立即释放。

def consumer():
    global product_count
    with con: 
        while product_count == 0:
            print("消费者：没货了，老老实实排队等...")
            con.wait()  # 释放锁并进入等待状态，直到被 notify 唤醒
        product_count -= 1
        print("消费者：终于买到商品了！")


c = threading.Thread(target=consumer) #实例化一个消费者线程
p = threading.Thread(target=producer) #实例化一个生产者线程

c.start() #线程进入自己的with cond:上锁，发现商品为0，进入等待使用con.wait()释放锁进入等待状态直到被notify 唤醒
time.sleep(0.5)  # 确保消费者先运行并卡住
p.start() #此时锁交到了生产者线程手上，生产商品，产出一件后notify消费者唤醒消费者线程。执行完毕释放锁。

c.join()
p.join()
#主线程在原地等这两个子线程彻底死掉（执行完）
```
如果此时有两个或多个消费者时这里必须使用while不能用if
解释：如果有两个消费者，使用if判断条件，消费者A和B醒来，A夺走了锁消费成功后将库存减少为0，释放锁，消费者B获得锁以后继续执行库存-1导致库存变成负数，逻辑漏洞。
底层原因： while 和 if 这两个基础关键字在所有编程语言里固有的运行机制不同

**if 语句**：if 的核心逻辑是：只在进门的时候检查一次条件。
```Python
if product_count == 0:     # 1. 进门检查：等于 0 吗？是的，进门。
    print("没货排队...")
    con.wait()             # 2. 线程在这里卡住。被唤醒后，直接从这里往下走。
                           # <--- 醒来后，直接离开 if 块，根本不回头！
product_count -= 1         # 3. 闭着眼睛直接减 1（危险！）
```

**while 语句**：while 的本质是循环。它的核心逻辑是：只要你还在我的地盘里，每次执行完或者卡住醒来，都必须回到顶端重新检查条件。

```Python
while product_count == 0:  # 1. 进门检查：等于 0 吗？是的，进门。
    print("没货排队...")
    con.wait()             # 2. 线程在这里卡住。被唤醒后，从这里苏醒。
                           # 3. 苏醒后，代码流会“被迫”回到第一步的 while 顶端！
                           # 4. 再次检查 product_count == 0 成立吗？
product_count -= 1         # 5. 只有当上面检查不成立（有货了），跳出循环才能走到这里。
```

## 六、 线程池 / 进程池：使用 concurrent.futures 简化资源管理。
1. **线程池**：使用线程池的核心目的，是为了减少频繁创建和销毁线程带来的系统开销，并能有效控制并发线程的数量，防止内存溢出。
资源复用：预先创建好固定数量的线程，循环利用。
流量控制：任务超载时，新任务会在队列中排队等待，绝不盲目创建新线程。

    ```python
    import threading
    # 每次来任务都新建一个线程
    t = threading.Thread(target=worker_func, args=(data,))
    t.start()
    ```
    这种“即用即建”的方式有两个致命缺点：
    开销大：创建和销毁线程都需要向操作系统申请资源，频繁操作很耗时。
    不可控：如果瞬间来 1000 个任务，就会创建 1000 个线程，系统可能会因为内存耗尽而直接崩溃。

2. 线程池核心用法
`concurrent.futures模块`
    `ThreadPoolExecutor` - 类
`with ThreadPoolExecutor(max_workers=线程数) as 变量名:` - 生成线程池
        `.map()` - 类方法
        `.summit()` - 类方法

    2.1 `ThreadPoolExecutor.map(func, *iterables)` —— 批量简单、按序处理

    类似于内置的 map，它把可迭代对象里的每一个元素丢进函数里异步执行。
    特点：返回一个生成器；结果的顺序与你输入的参数顺序完全一致；无法为不同的任务传递不同的额外参数。

    ```python
    from concurrent.futures import ThreadPoolExecutor
    import time

    # 模拟一个下载任务
    def download_image(img_name):
        print(f"开始下载: {img_name}")
        if img_name == "图片B.jpg":
            time.sleep(3)  # 故意让图片B下载得很慢（需要3秒）
        else:
            time.sleep(1)  # 图片A和C很快（只需要1秒）
        return f"{img_name} 下载成功！"

    # 准备好我们的任务数据
    img_list = ["图片A.jpg", "图片B.jpg", "图片C.jpg"]

    print("--- map() 开始执行 ---")
    with ThreadPoolExecutor(max_workers=3) as executor:
        # map 会直接返回一个结果生成器
        results = executor.map(download_image, img_list)
        
        # 直接用 for 循环拿结果，不需要写 .result()
        for res in results:
            print(f"收到通知: {res}")

    print("--- map() 执行结束 ---")
    ```

    2.2 `ThreadPoolExecutor.submit(fn, *args, kwargs)` ——  精确控制、极其灵活
    每次只提交一个任务，并立即返回一个 Future 对象（可以理解为一张“小票”，凭这张票可以在未来某个时刻换取结果）。你必须主动调用 小票.result()，才能拿到函数真正的返回值。
    通过submit获取result有两种方式

    方式A：死板排队获取
        ```python
        future1 = executor.submit(worker, 2)
        future2 = executor.submit(worker, 1)
        # 用法 A：直接通过 .result() 阻塞式获取（谁先 submit 就先等谁，即使后提交的先做完）
        print(future1.result()) 
        ```

    方式B：结合 as_completed()，谁先做完就先处理谁（推荐！）
        ```python
        futures = [executor.submit(worker, i) for i in [3, 1, 2]]
        for fut in as_completed(futures):
            print(f"某个任务做完了，结果是: {fut.result()}")
        ```

    ```python
    from concurrent.futures import ThreadPoolExecutor, as_completed

    def calculate(num):
        return num * 10

    # 创建线程池
    with ThreadPoolExecutor(max_workers=2) as executor:
        # 提交任务，拿到小票（Future对象）
        f1 = executor.submit(calculate, 1)
        f2 = executor.submit(calculate, 2)
        f3 = executor.submit(calculate, 3)
        
        # ---- 方式一：直接通过 .result() 获取 ----
        print(f1.result())
        print(f2.result())
        print(f3.result())
        
        # ---- 方式二：结合 as_completed() 获取 ----
        futures_list = [f1, f2, f3]
        for fut in as_completed(futures_list):
            print(fut.result())
    ```
    问：为什么要用这两个方法
    答：线程池本身只是一个“厨房”， 假设你开了一家高档餐厅，后厨里有 3 个厨师（这就是 ThreadPoolExecutor(max_workers=3)，也就是你的线程池）。厨师（线程）只负责站在灶台前：来一个菜，炒一个菜；没菜的时候，就原地待命。
    但是，身为前台经理的你，要怎么把客人的订单传给后厨呢？你传菜单的方式，就决定了你是用 map 还是 submit。

    **方式一：使用 map（批量流水线模式）**
    中午旅行团突然来了 30 个游客，他们每个人都要点一碗“兰州拉面”。这时候，你不需要一个一个去跟厨师交涉。你直接把这 30 个人的名单一拍：“后厨听令（map），按照这个名单（可迭代对象），给每个人都做一碗拉面（同一个函数）！
    ```python
    # 批量做拉面，名单是 1 到 30 号顾客
    executor.map(make_noodle, range(1, 31))
    ```
    **方式二：使用 submit（单点私人定制模式）**
    晚上店里来了几位散客，情况变得复杂了：
    A 桌客人说：“我要一份炒饭。” （任务 1）
    B 桌客人说：“我要一碗牛肉面，不加葱。” （任务 2）
    这时候你绝对不能用 map 批量处理了。你必须拿着点菜单，单独、一件一件地把任务递进厨房：
    “小张，接单（submit），做一份炒饭！” -> 拿到 A 桌的小票（Future 1）
    “小李，接单（submit），做一碗牛肉面！” -> 拿到 B 桌的小票（Future 2）
    ```python
    future1 = executor.submit(make_rice, "A桌")   # 做炒饭
    future2 = executor.submit(make_noodle, "B桌") # 做面条
    ```
3. 线程池的致命克星：Python GIL 锁
这是写 Python 多线程必须牢记的底层逻辑：Python（CPython实现）有全局解释器锁（GIL）。在同一时刻，只允许一个线程执行 Python 字节码。
这意味着：
计算密集型任务（如：大量循环计算、图像处理、加密解密）：线程池无法带来性能提升，甚至因为线程切换变得更慢。这类任务应该使用进程池（ProcessPoolExecutor）。进程池和线程池用法一摸一样（唯一不同：使用`if __name__ == '__main__':`避免进程无限循环）
I/O 密集型任务（如：网络爬虫、读写数据库、文件读写、调用外部 API）：当某个线程在等待网络返回时，它会主动释放 GIL 锁，让其他线程工作。这种场景下，线程池能带来成倍的效率提升。

4. 实战避坑与最佳实践
① 线程池大小（max_workers）设多少合适？
对于 I/O 密集型任务，通常的经验公式是：
**max_workers=CPU 核心数×2 到 CPU 核心数×5**
（Python 3.8+ 默认自动设为 min(32, os.cpu_count() + 4)，通常默认就够用了）。
不要盲目设得太大，成千上万的线程光是上下文切换（Context Switch）就能把 CPU 拖垮。

② 异常处理：千万别让错误“死得无声无息”
如果在线程池的任务里发生了报错（比如 1/0），程序不会直接在控制台崩溃，而是把异常吞掉，存放在 Future 对象里。如果你不主动去取结果，这个报错你永远不会发现！
正确做法：
```Python
future = executor.submit(div_zero_func)
try:
    res = future.result() # 此时如果任务报错，会在这里抛出
except ZeroDivisionError:
    print("捕获到了线程内部的除零错误！")
```
③ 避免在线程池内部滥用全局变量
多线程共享进程内存，如果多个线程同时去修改同一个全局变量，会导致数据错乱（线程不安全）。
如果必须要修改共享数据，必须引入锁（threading.Lock）。
最佳实践是：让线程成为纯函数。通过参数传进去，通过 return 返回出来，互不干扰。

## 知识点补充一
**全局变量**： 在函数外面定义的变量，属于整个代码文件。
**局部变量**： 在函数里面定义的变量，函数外面是看不到的，函数执行完就销毁了。

```python 修改全局变量

money = 100  # 全局变量：钱包里有 100 块钱
def spend_money():
    money = 50  # ⚠️ Python 认为：你想在函数内部新建一个叫 money 的局部变量
    print(f"函数内部的 money: {money}")  # 输出 50
spend_money()
print(f"函数外部的 money: {money}")  # 输出 100！外面的钱根本没变！

#如果你真心想在函数内部修改外面的全局变量，你就必须在函数第一行大喊一声：global 变量名！
    global money  # 📢 报告 Python！我这里说的就是外面的那个 money！
    money = 50  # 成功修改外面的全局变量
spend_money()
print(f"函数外部的 money: {money}")  # 输出 50！外面的钱真的被改了！
```
**总结**
在纯函数里，你不写 global，Python 就不知道这个 counter 是哪来的，只能当成局部变量。
在类里，你写了 self.counter 或 ClassName.counter，路径清清楚楚，Python 闭着眼睛都能找到，自然就不需要 global 关键字来引路了！

## 知识点补充二
**Python的列表推导式（List Comprehension）**: 
`[我们要塞进列表的东西 for 变量 in 可迭代对象]` : python 看到这个结构时，会自动在幕后帮你创建一个空列表，然后每循环一次，就把最前面的“东西”生成出来，自动 append（追加）到列表里，最后把这个充满元素的列表赋值给最左边的变量 threads。


```python
threads = [threading.Thread(target=worker, args=(i,)) for i in range(5)]
相当于
threads = []  # 1. 先创建一个空列表
for i in range(5):  # 2. 开始循环，i 依次为 0, 1, 2, 3, 4
    t = threading.Thread(target=worker, args=(i,))  # 3. 创建线程对象
    threads.append(t)  # 4. 把线程对象塞进列表里
```

## 知识点补充(线程池)

`list(results)`
其实，即使不写 list(results)，程序也一定会等待所有下载完成才打印总时间。因为 with 语句结束时，线程池会自动调用 shutdown(wait=True)，强行把主线程卡住，直到所有工人干完活。那么，为什么在 Python 多线程/多进程编程中，大家往往还是习惯写一句 list(results) 或者去遍历它呢？ 这背后涉及到 executor.map 的核心机制：生成器（Generator）的“惰性求值”与“异常触发”。
原理 1：激活“惰性”的生成器
executor.map() 的返回值 results 不是一个真正的列表，而是一个迭代器/生成器（Iterator/Generator）。
什么是惰性？ 当你执行 results = executor.map(...) 时，Python 只是创建了一个“备忘录”，线程池在后台开始默默异步执行任务。但此时 results 里面是空的，它没有存下任何结果，直到你向它“索要”。
list(results) 的作用：list() 就像一个催债的，它会强迫这个生成器立刻把所有的结果吐出来，并打包成一个真正的 Python 列表。为了把结果吐出来，代码就必须等待所有线程的函数执行完毕。

原理 2：为了“引爆”或捕获异常（最核心原因）
这是子线程编程中最容易踩的坑：如果子线程里报错了，主线程默认是不知道的！

原理 3：按“原本的顺序”收集结果
多线程下载时，速度有快有慢（比如第5个网址比第1个网址先下载完）。但是 executor.map 非常强迫症，它返回的 results 必须严格对应你传入的 url_list 顺序。
线程 5 先下载完了，它的结果会先收录在后台。
当你调用 list(results) 时，它会耐心等待线程 1、2、3、4。
然后按照 [结果1, 结果2, 结果3, 结果4, 结果5] 的标准顺序组装好交给你。

关卡二：多进程与安全锁实战 —— “多核火山爆发计数器”
📋 任务描述
我们来玩一个让 CPU 核心火花擦键盘的游戏。我们需要让电脑连续计算从 1 到 20,000,000（两千万）这全套两千万个数字的乘方（比如 i**2）。
🛡️ 闯关限制与要求
打破 GIL 枷锁：这是纯数学计算（CPU密集型），禁止使用多线程！必须使用 concurrent.futures.ProcessPoolExecutor（进程池）。
加上安全锁：还记得 Windows 的 spawn 机制吗？为了防止子进程无限“套娃”导致电脑死机，必须把启动进程池的代码塞进进程专属的安全锁里面！
效率大比拼：
测试 A：用单进程（普通 for 循环或列表推导式）跑完这两千万次计算，记录耗时。
测试 B：用进程池（多进程）跑完这两千万次计算，记录耗时。
打印对比：最后打印出：“单进程耗时 X 秒，多进程耗时 Y 秒，多进程提升了 Z 倍效率！”
💡 核心思路提示（不给代码）
进程池的使用方法和线程池几乎一模一样，只是把 ThreadPoolExecutor 换成 ProcessPoolExecutor。
进程池会自动根据你电脑的 CPU 核心数（比如 4 核或 8 核）去分工。你会亲眼看到，多进程跑出来的耗时，会比单进程快得惊人！

````python
#练习一（待优化）
from concurrent.futures import ThreadPoolExecutor
import time
import requests

def web_download(url):
    requests.get(url)

url_list = ["url1", "url2", "url3", "url4", "url5", "url6", "url7", "url8", "url9", "url10"]
timestamp1 = time.time()
with ThreadPoolExecutor(max_workers = 3) as executor:
    results = executor.map(web_download, url_list)
    list(results) 
    #在普通的 Python 代码中，确实是先运行完代码块里的代码，才会接着往下走。但 executor.map() 的神奇（同时也是容易让人误解）之处在于：它在代码块里其实已经“运行完了”，只是它运行的成果不是“下载好的图片”，而是一堆“承诺（Promise）”。
timestamp2 = time.time()
timespace = timestamp2 - timestamp1
print(f"总用时{timespace}秒")
````

````python
#练习二（待优化）
from concurrent.futures import ProcessPoolExecutor
import time

def calculate_power(num):
    result = num*num
    return result


if __name__ == "__main__":

# -----------------测试A-------------------------------

    num_range = range(1, 20000001)
    single_results = []
    start_time = time.time()

    for num in num_range:
        single_result = calculate_power(num)
        single_results.append(single_result)
    end_time = time.time()
    print(f"测试A总耗时{end_time - start_time}")


# -----------------测试B-------------------------------
    start_time = time.time()
    with ProcessPoolExecutor(max_workers=3) as executor:
        results = executor.map(calculate_power,range(1, 20000001), chunksize = 10000)
        sum(results)
    end_time = time.time()
    print(f"测试B总耗时{end_time - start_time}")
````

# 🐍 Python 学习笔记｜第 35 关：算机网络通信基础
## 一、网络世界的“三维坐标”（核心概念）

### IP(Internet Protocol - 互联网协议) - IP协议工作在 OSI 七层模型的网络层（Network Layer）
1. 核心职责：
    寻址（Addressing）： 为网络上的每一个节点分配一个逻辑地址，确保大家都知道“谁是谁”。
    路由（Routing）： 决定数据包从源头出发，经过哪些路由器和网络节点，最终到达目的地。这是最基础的“带路”功能。

2. IP 地址（Internet Protocol Address）：
        角色： 设备的大门牌号。
        通俗理解： 决定了数据包裹要送到哪台“设备”（比如你的电脑、手机或服务器）。
        特殊地址： 127.0.0.1（别名 localhost），代表“本地回环地址”，即你自己当前的电脑。常用于程序员自己跟自己调试代码。

    `IPv4（Internet Protocol version 4）`：我们最熟悉的地址
        格式： 
            IPv4 地址由 32 位二进制数组成。为了让人类容易阅读，我们通常使用点分十进制表示法，将 32 位拆成 4 个 8 位组（Octet），每组转换成 0~255 的十进制数，用点隔开。
        构成：
            网络号（Network ID）： 标识设备属于哪一个网络（相当于某条街道）。
            主机号（Host ID）： 标识该网络下的具体某台设备（相当于街道上的某张门牌号）。

        子网掩码：为了告诉电脑哪里是网络号、哪里是主机号，必须配合子网掩码（Subnet Mask）使用。
            子网掩码的硬性规定： 前面必须全是 1，后面必须全是 0。
            1 对应的 IP 地址部分，代表网络号（固定不动，相当于街道名）。
            0 对应的 IP 地址部分，代表主机号（可以变动，相当于门牌号）。
                例子：IP 是 192.168.1.5，子网掩码是 255.255.255.0
                
                IP地址:    192  .  168  .    1  .   5
                           ↓       ↓        ↓      ↓
                子网掩码:   255  .  255  .  255  .   0

                因为每次写 255.255.255.0 太麻烦了，网络工程师们发明了一种偷懒的数数方法。
                我们把 255.255.255.0 转换成二进制看看：
                11111111.11111111.11111111.00000000
                8           8           8       =       24个1
                所以，192.168.1.5/24 后面的这个 /24，意思就是：“这把子网掩码尺子，前面有 24 位都是 1（也就是说前 3 组数字是网络号）”。
                同理：
                如果是 /16，就说明前 16 位是 1（即 255.255.0.0），前两组数字是网络号。
                如果是 /8，就说明前 8 位是 1（即 255.0.0.0），只有第一组数字是网络号。
            
    `IPv6`：面向未来的终极解决方案
        格式：
            IPv6 采用了 128 位 地址长度。使用了冒号分隔的十六进制。
            128位二进制可以转化成32位十六进制：四位二进制 0000->1111 16种状态， 一位十六进制也是 16种状态
            所以每四位二进制可以转化成一位十六进制，128位二进制就可以转化成32位十六进制。
            将32位十六进制每4位分成一个区块可以得到一个IPv6地址
            例：2001:0db8:0000:0000:0000:0000:1428:57ab

            因为 IPv6 实在是太长了，书写时有两条可以偷懒的简写规则：
                省略前导零： 每组中的前导 0 可以省略。例如 :0370: 可以写成 :370:，:0000: 可以写成 :0:。
                双冒号压缩： 连续满是零的组，可以用双冒号 :: 代替，但一个地址中只能出现一次 ::（否则电脑不知道中间到底漏掉了几组零）。 
                上述例子缩写后： 2001:db8::1428:57ab
        优点：
            1. 地址多
            2. 更快的路由转发（报头精简）：IPv6 简化了数据包的报头结构，去掉了 IPv4 中一些不常用的字段，并且把报头长度固定下来。这让路由器处理数据时像走高速公路一样，不需要反复计算，转发效率更高。
            3. 人人都有“公网IP”（再也不需要 NAT 了）：在 IPv4 时代，因为地址不够，大家只能挤在局域网里，通过 NAT（网络地址转换）技术共用一个公网 IP，这导致点对点（P2P）连接、远程访问自己的电脑非常麻烦。IPv6 让每个设备都能拥有独立的公网 IP，物联网设备之间可以直接“面对面”直连。
            4. 原生支持更高安全性（IPsec）：IPv6 在设计时就原生把 IPsec（网络层加密安全协议）纳入了标准，虽然现在 IPv4 也能用 IPsec，但 IPv6 的底层支持让网络传输加密变得更加普遍和规范。
            5.即插即用（无状态自动配置 - SLAAC）：设备接入 IPv6 网络时，不需要 DHCP 服务器大费周章地分配地址，设备可以根据路由器的广播信息，自己“算”出一个全球唯一的 IPv6 地址，实现真正的即插即用。

3. 公网 IP 与 私网 IP
    **核心概念**
        `公网（外网）IP`：真正的互联网（Internet）。全球只有一张公网，公网 IP 是全球唯一的。只要设备拥有公网 IP，全世界任何联网设备都能直接访问它。
        `私网（内网/局域网）IP`：家庭、学校或公司内部搭建的独立小网络。私网 IP 只在同一个局域网内部有效。私网 IP 可以全球重复使用（例如：你家和邻居家的手机 IP 都可以是 192.168.1.5），因为它们处于不同的物理空间（平行宇宙），互不干扰。
        诞生背景：在 IPv4 诞生初期本想人人用公网，但因为总数只有 43 亿不够用，于是专门划出几段地址（如 192.168.x.x）作为私网保留地址，以此缓解地址枯竭危机。

    **IP 地址的层级分配体制**
        公网 IP 分配（自上而下批发）
        最高机构：IANA（全球互联网号码分配局）掌握所有 IP 根源。
        洲际机构：IANA 将 IP 划分给全球五大洲的管理机构（如亚太地区的 APNIC）。
        国家运营商/大厂（LIR）：中国的电信、联通、移动或腾讯、阿里等，每年向 APNIC 申请/购买成批的 IP 地址段。
        终端用户：用户交网费后，运营商机房从地址池中动态分配一个公网 IP 给用户的光猫/路由器（WAN口）。

        私网 IP 分配（自建局域网）
        用户家里的路由器通过 DHCP 协议，在内部网段中自动将私网 IP 动态分配给手机、电脑等终端。
    
    **数据传输的核心场景与规则**

        场景 A：私网设备 ⇄ 公网服务器（单向跨界：只能由“私网内部的设备”主动发起连接去访问“公网服务器”，反过来则不行。）
        限制：私网 IP 无法直接在公网上路由（公网路由器不认私网 IP）。
        破局者：路由器的 WAN 口（公网口） 运行 NAT（网络地址转换） 技术。
        流程：
            私网设备发起请求 → 路由器拦截。
            路由器抹去数据包中的私网 IP，换上运营商分配给它的全球唯一公网 IP 投递出去，并在后台账本上记下对账记录。
            公网回包到达路由器 → 路由器查账本 → 精准转发回内部对应的私网设备。


        场景 B：同一路由器下的私网设备直连（内部自治）
        场景：同一个路由器下的设备 A（192.168.1.2）向设备 B（192.168.1.3）传输数据。
        (注：修正了原笔记中 192.168.1.1 的小问题，因为 .1 通常被路由器自己占用了。)
        流程：
            设备 A 比对发现彼此网络号相同（都是 192.168.1），确认是“自己人”，决定不走外网。
            路由器的 LAN 口/Wi-Fi天线（私网口） 在内部充当交换机（Switch）的角色。
            路由器在底层直接将两台设备连通，数据只在客厅的局域网内流转，不消耗公网宽带，也不需要 NAT 转换。
            
    4.  OSI模型                                        TCP/IP 模型（4层实战）      
        第 1 层：物理层（网线、光纤）
        第 2 层：数据链路层（交换机、MAC 地址）-------->  1,2在实际中被称为网络接口层
        第 3 层：网络层(Network Layer)       -------->  网络层
        第 4 层: 传输层(TCP、UDP 协议)       -------->  传输层
        第 5 层：会话层（建立/管理连接）                        
        第 6 层：表示层（数据加密、解密）      -------->  5,6,7在实际中被统称应用层
        第 7 层：应用层（HTTP、微信、浏览器）

### 二、 端口（Port）
    角色： 软件的房间号。
    通俗理解： 一台电脑上同时运行着微信、浏览器、游戏等多个软件。IP 地址只帮你把数据送到电脑，而端口号（一个 0~65535 的数字）则决定了这个数据到底交给哪个软件。
    例：Web 网页服务默认占用 80 或 443 端口；我们代码中测试用了 8888 端口。
1. 在 TCP/IP 协议中，端口号是用一个 16 位的二进制数来表示的。这意味着端口号的范围是从 0 到 65535（总共 65536 个）。

    公认端口（Well-Known Ports）：0 - 1023 - 这些端口被严格分配给一些最核心、最常用的网络服务。它们就像是消防 119、报警 110 一样，是全网通用的“特权端口”。
        80 端口：HTTP 协议（普通的网页浏览）
        443 端口：HTTPS 协议（加密的网页浏览，网银、网购等）
        21 端口：FTP 协议（文件传输）
        22 端口：SSH 端口（远程安全连接服务器）
        25 端口：SMTP 协议（发送邮件）
        53 端口：DNS 协议（域名解析）
    注册端口（Registered Ports）：1024 - 49151 - 这些端口通常分配给一些公司开发的特定软件或服务。虽然没有公认端口那么绝对，但大家一般都会遵守约定。
        3306 端口：MySQL 数据库
        6379 端口：Redis 缓存数据库
        8080 端口：Tomcat 服务器 / 常常作为 HTTP 的备用端口
        27017 端口：MongoDB 数据库
    动态/私有端口（Dynamic/Private Ports）：49152 - 65535 - 这些端口是“临时工”。当你打开浏览器访问百度时，你的电脑会在这个范围内随机挑选一个端口作为你浏览器的临时端口。一旦你关闭网页，这个端口就会被释放，留给其他程序使用。

2. 端口与传输层协议（TCP vs UDP）
端口本身是依附于传输层协议存在的。这就导致了一个很有趣的现象：同一个端口号，在 TCP 和 UDP 协议下是完全独立的。TCP 端口 80 和 UDP 端口 80 是互不干扰的。
网络区分一个连接，靠的是“四元组”： **源IP+源端口+目的IP+目的端口** 只要这四个要素组合起来是唯一的，连接就不会冲突。

### 三、Socket（套接字）
    角色： 软件和网络之间的专属插座 / 电话机。
    通俗理解： 软件不能直接去碰网线，它必须向操作系统申请一个 Socket。通过这个“插座”，发送端把数据灌进去，接收端从对应的“插座”里拔出来。
    
    基础语法：
    `import socket`: 导入模块
        `.socket(参数一, 参数二)`: 创建 Socket 对象
            **第一个参数** —— 决定“地址类型”（怎么找到对方）
            `socket.AF_INET`：使用 IPv4 地址族
            `socket.AF_INET6`：使用IPv6 地址族
            **第二个参数** —— 决定“传输方式”（怎么传输数据）
            `socket.SOCK_STREAM`：TCP 流式传输协议 - TCP 是面向字节流的，因此只能发送和接收 bytes（字节）类型，发送字符串前必须编码（encode），收到字节后必须解码（decode）。
            `socket.SOCK_DGRAM`：UDP

    分支：根据传输方式的不同socket的用法不同

    1. 使用socket.SOCK_STREAM - TCP传输数据
    特点： 面向连接、可靠、基于字节流。
    核心机制： * 三次握手： 说话前先打通电话，确认双方听力和说话能力都正常。
    超时重传： 路上包裹丢了，快递员（TCP）会自动重新补发一份，直到你签收。
    四次挥手： 挂电话时，双方各自确认没有话要说了，再优雅地挂断。
    适用场景： 网页浏览（HTTP）、网购支付、文件下载、发邮件。

    **客户端与服务器通用**
        `s.send(bytes)`: 发送数据
        `s.recv(bufsize)`: 接收数据 - bufsize（缓冲区大小 - Buffer Size）告诉计算机：“我这次最多只想装这么多数据，别给我塞溢出了。”s.recv() 是一个流式的操作。如果对方发了 10,000 字节的数据，而你的 bufsize 设成了 1024，程序不会报错，它只是这次先捞走 1024 字节，剩下的数据还在队列里排队。为了拿到完整的数据，通常需要配合 while 循环来反复接收：
        `s.close()`: 关闭连接  - 在 TCP 协议中有一个硬性规定：当客户端发送完所有数据，并主动调用 close() 关闭连接时，服务端的 recv() 就会捞出一个“空字节”（即 b''）。（实际是发送了一个INF包表明后面没有数据了，由于TCP是字节传输数据所以python定义为返回了一个空字节即 b''）

    **服务器端（Server）特有语法**
        `s.bind(address)`: 绑定地址与端口
        `s.listen(backlog)`: 开启监听 - 这一步让套接字进入“监听”状态，操作系统会在底层为*它开辟两个队列（俗称半连接队列和全连接队列）*，用来存放正在走敲门流程的客户端请求。
            当一个客户端想要连接服务器时，它会发起 TCP 的“三次握手”。请注意：这个握手过程完全是由操作系统的网络协议栈（内核）自动完成的，此时 accept() 甚至都还没有参与。
            （1）客户端发来 SYN 包（敲门）。
            （2）服务器内核回应 SYN+ACK（开门并欢迎）。
            （3）客户端回应 ACK（确立关系）。
            当三次握手完美结束后，这个成功的连接就会被*放入内核的 “已完成连接队列（Fully Completed Connection Queue）”* 中排队。
        `s.accept()`: 接受客户端连接
            *如果队列里已经有连接在排队*：accept() 会立刻从队列的头部取出一个已经建立好的连接。
            如果队列是空的：在默认的阻塞模式下，accept() 就会让代码停在这一行（也就是“阻塞”住），像个老僧入定一样静静等待。直到网络上有新的客户端完成了三次握手、进入队列后，accept() 才会苏醒并把这个连接领走。
            accept() 成功领到连接后，会返回两个东西：conn - 已连接套接字（Connected Socket） 和 addr - Address（地址）
            
    **客户端（Client）特有语法**
        `s.connect(address)`: 

    ```python
    import socket
    import threading
    import time

    # ==========================================
    # 1. 服务器端逻辑 (Server)
    # ==========================================
    def run_server():
        # 【通用】创建 Socket 对象 (IPv4, TCP)
        server_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        
        # 允许端口复用（防止连续运行报错地址被占用）
        server_socket.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
        
        # 【服务器特有】s.bind() 绑定本地地址和端口 9999
        server_socket.bind(('127.0.0.1', 9999))
        
        # 【服务器特有】s.listen() 开启监听，允许排队数设为 5
        server_socket.listen(5)
        print("[Server]: 迎宾Socket已就绪，正在大门口听有没有人来敲门...")
        
        # 【服务器特有】s.accept() 阻塞等待客户端连接
        # conn 是专职服务员Socket，addr 是客户端的地址
        conn, addr = server_socket.accept()
        print(f"[Server]: 收到来自 {addr} 的连接请求，已指派专职Socket进行服务。")
        
        # 模拟接收大数据，展示你在笔记中提到的 bufsize 循环接收逻辑
        full_data = "" #创建一个空水桶以便后续连续存储信息
        while True:
            # 【通用】s.recv(bufsize) 每次最多只捞取 10 字节（故意设小，演示你的循环逻辑）
            chunk_bytes = conn.recv(10) 
            
            if not chunk_bytes: # 如果收到空字节，说明客户端拔插头(close)了
                break
                
            # 【通用】将接收到的 bytes 解码(decode)为字符串
            chunk_str = chunk_bytes.decode('utf-8')
            print(f"[Server]: 从缓冲区捞走了 10 字节 -> '{chunk_str}'")
            full_data += chunk_str
            
        print(f"[Server]: 最终拼接出的完整消息是 -> '{full_data}'")
        
        # 别忘了礼貌挂断
        conn.close()
        server_socket.close()
        print("[Server]: 服务器通信结束，插座已关闭。")

    # ==========================================
    # 2. 客户端逻辑 (Client)
    # ==========================================
    def run_client():
        # 为了确保服务器先起来，客户端稍微等个 0.5 秒再敲门
        time.sleep(0.5)
        
        # 【通用】在客户端家里创建 Socket 对象
        client_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        
        # 【客户端特有】s.connect() 主动发起连接，去找服务器的 9999 端口
        print("[Client]: 正在去敲 127.0.0.1:9999 的门...")
        client_socket.connect(('127.0.0.1', 9999))
        print("[Client]: 连接成功！")
        
        # 准备一段超过 10 字节的字符串，测试服务器的循环接收
        secret_msg = "Socket编程真有趣！" 
        
        # 【通用】s.send() 发送数据。发送前必须进行编码(encode)转为 bytes
        print(f"[Client]: 准备发送数据: '{secret_msg}'")
        client_socket.send(secret_msg.encode('utf-8'))
    
        # 【通用】s.close() 客户端聊完了，主动断开连接
        client_socket.close()
        print("[Client]: 客户端已拔出插头。")

    # ==========================================
    # 3. 启动运行（用多线程让它们在同一个电脑里打架）
    # ==========================================
    if __name__ == "__main__":
        # 启动服务器线程
        server_thread = threading.Thread(target=run_server)
        # 启动客户端线程
        client_thread = threading.Thread(target=run_client)
        
        server_thread.start()
        client_thread.start()
        
        server_thread.join()
        client_thread.join()
    ```
    **知识点补充一(存疑)**
    `数据的“跨国翻译”：encode() 与 decode()`
    `完美的“分手信号”：if not chunk_bytes`
    `专职服务员与迎宾小姐：accept() 的双重返回值`
    `暴力拆迁保护：SO_REUSEADDR 端口复用`
    `客户端的本地地址去哪了？`
    代码体现：
    服务器写了：server_socket.bind(('127.0.0.1', 9999))
    客户端只写了：client_socket.connect(('127.0.0.1', 9999))
    大白话解释：服务器必须明确死守 9999 端口，因为客户端要来找它。但客户端自己占哪个端口无所谓。所以客户端不需要显式调用 bind()，当它调用 connect() 的那一刻，操作系统会自动给客户端随机分配一个隐形端口。
    

    2. 使用socket.SOCK_DGRAM - UDP传输数据
    特点： 无连接、不可靠、速度极快。
    核心机制： 没有握手，没有签收确认，不保证顺序。抓起数据包直接往目的地扔，丢了也不管。
    适用场景： 视频通话、多人在线竞技游戏（如王者荣耀）、直播、DNS 域名解析。

    **客户端与服务器通用**
    `s.sendto(data, address)`：（核心变化）发送数据，因为没有连接，每次都必须死抠目的地地址。 - 这里的address是一个元组（'IP', 端口）
    `data, addr = s.recvfrom(bufsize)`：（核心变化）接收数据，返回一个元组，拿到数据的同时还能知道寄件人是谁。
    `s.close()`：关闭插座。

    **服务器端（Server）特有语法**
    `s.bind(address)`: 绑定地址与端口 - 只需要绑定端口蹲守。（注意：UDP 服务器不需要 listen() 和 accept()！）
    **客户端（Client）特有语法**
    什么都不需要！ 直接用 sendto 往目标 IP 和端口扔数据即可。（注意：UDP 客户端不需要 connect()！）
 
    ```python
    import socket
    import threading
    import time

    # 服务器端口
    SERVER_ADDR = ('127.0.0.1', 9999)

    # ==========================================
    # 1. UDP 服务器端逻辑
    # ==========================================
    def run_udp_server():
        # 【核心变化】：第二个参数改成了 SOCK_DGRAM
        server_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        
        # UDP 服务器同样需要固定自己的大门地址
        server_socket.bind(SERVER_ADDR)
        print("[Server]: UDP服务器已在 9999 端口蹲守，不需要 listen，随时准备捡快递...")
        
        # 【核心变化】：使用 recvfrom 接收数据
        # 它会同时返回：数据(data) 和 寄件人的地址(client_addr)
        data_bytes, client_addr = server_socket.recvfrom(1024)
        
        msg = data_bytes.decode('utf-8')
        print(f"[Server]: 捡到一个来自 {client_addr} 的包裹，内容是: '{msg}'")
        
        # 服务器给客户端回一封信
        reply_msg = "服务器收到你的UDP投递啦！"
        # 【核心变化】：使用 sendto，必须带上刚才拿到的 client_addr 地址
        server_socket.sendto(reply_msg.encode('utf-8'), client_addr)
        
        server_socket.close()
        print("[Server]: UDP服务器关闭。")


    # ==========================================
    # 2. UDP 客户端逻辑
    # ==========================================
    def run_udp_client():
        time.sleep(0.5) # 等服务器先蹲守好
        
        # 【核心变化】：客户端也创建 SOCK_DGRAM
        client_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        
        # 【注意】：UDP 客户端不需要调用 connect()！！！不需要建立连接！
        secret_msg = "Hello UDP, 我直接扔过去啦！"
        print(f"[Client]: 准备盲发数据到 {SERVER_ADDR}...")
        
        # 【核心变化】：直接使用 sendto，把数据和目标地址一起扔出去
        client_socket.sendto(secret_msg.encode('utf-8'), SERVER_ADDR)
        
        # 顺便接收服务器的回信
        reply_bytes, server_addr = client_socket.recvfrom(1024)
        print(f"[Client]: 收到来自 {server_addr} 的回信: '{reply_bytes.decode('utf-8')}'")
    
        client_socket.close()
        print("[Client]: UDP客户端关闭。")


    # ==========================================
    # 3. 启动运行
    # ==========================================
    if __name__ == "__main__":
        t1 = threading.Thread(target=run_udp_server)
        t2 = threading.Thread(target=run_udp_client)
        
        t1.start()
        t2.start()
        
        t1.join()
        t2.join()
    ```

### 问答
两台电脑通过服务器传输数据，在网络世界里，本质上就是两个客户端（Client）同时在和一个中央服务器（Server）进行交流。
电脑 A（客户端 1）：主动把数据发送给服务器。
电脑 B（客户端 2）：在另一端等待接收服务器转发过来的数据。
云端电脑（中央服务器 Server）：充当快递中转站、路由器或接线员。

问一：为什么两台电脑不能直接“手拉手”交流？
答： 
1. 著名的“内网 IP 隔离”问题（找不到对方）
    现实情况：你家里的电脑和朋友家里的电脑，通常都被关在各自宽带路由器的“局域网”里（比如你俩的局域网 IP 可能都是 192.168.1.5）。
    结果：电脑 A 在互联网上是无法直接感知和定位到电脑 B 的具体位置的。
    解决办法：但服务器不一样，服务器拥有一个在全世界都公开、固定的公网 IP（就像大英博物馆的地址一样明确）。因此，A 和 B 都能轻易找到服务器。
2. “不在线”问题（离线留存）
    如果电脑 A 直接发给电脑 B，万一电脑 B 此时刚好断网、关机或者手机没电了，这数据就彻底丢了。有了服务器，电脑 A 可以先把数据托付给服务器（比如存进数据库），等电脑 B 上线了，服务器再推给 B。
3. 打破内网隔离的“现代魔法” - P2P（点对点）通信与 NAT 打洞（Hole Punching）
    既然服务器带宽那么贵，视频通话、联机游戏如果全走服务器中转，腾讯/大厂可能会亏死。于是诞生了 NAT 打洞技术：
    牵线搭桥：电脑 A 和电脑 B 虽然都在内网，但它们都主动连接了拥有公网 IP 的中央服务器。
    暴露地址：中央服务器通过它们发来的请求，一眼就看穿了它们经过路由器 NAT 转换后的公网 IP 和外网端口号。
    交换名片：服务器把 A 的外网地址告诉 B，把 B 的外网地址告诉 A。
    互相盲发（打洞）：A 冲着 B 的外网地址发一个 UDP 包（虽然会被 B 的路由器拦截），但 A 的路由器因此记下了“我要和 B 通信”；紧接着 B 也冲着 A 的外网地址发一个 UDP 包。此时，由于双方的路由器都开启了向对方发送的绿灯，通道瞬间建立！
    直连成功：从此，服务器功成身退，A 和 B 之间的数据直接在网线上“手拉手”传输，不经过服务器中转。这就是 P2P 下载（如 BT、迅雷）和现代音视频通话的核心原理。


## 二、练习
1. 编写一对简单的 TCP 聊天程序
创建 tcp_chat_server.py 和 tcp_chat_client.py。
服务器监听在 127.0.0.1: 8080，接受连接后，先接收客户端的一条消息并打印"消息已收到"，然后向客户端回复。
客户端连接服务器后，发送一条 "你好，这是TCP测试"，然后接收服务器回复并打印。
2. 编写 UDP 回声程序
创建 udp_echo.py，让同一个程序既能发送也能接收。
发送端向 127.0.0.1:12345 发送任意文本，接收端绑定同样的端口，收到数据后立即把原数据发回给发送方。
在接收端打印收到的内容和发送方地址。
提示：练习代码都可以使用 with 语句吗？套接字对象本身支持上下文管理器（with socket.socket(...) as s:），但在实际编程中，通常将 bind、accept 等操作放在 with 块内。你可以尝试改写一下。
完成这两道练习，你就能亲手体验 TCP 的可靠连接和 UDP 的轻量无连接通信了。有任何疑问随时问我。


````python
#tcp_chat_server.py
import socket
import time
import threading

def run_TCPServer():
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    server_socket.bind(('127.0.0.1',8080))
    server_socket.listen(5)
    new_socket, client_address = server_socket.accept()
    
    received_bytes = b''
    while True:
        chunk_bytes = new_socket.recv(10)
        
        if not chunk_bytes:
            break
    received_bytes += chunk_bytes
    received_data = received_bytes.decode('utf-8')
    print("消息已收到")
    print(received_data)

    new_socket.close()
    server_socket.close()

def run_TCPClient():
    time.sleep(0.5)
    client_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    client_socket.connect(('127.0.0.1', 8080))

    msg = "你好，这是TCP测试"
    client_socket.send(msg.encode('utf-8'))

    client_socket.close()
    
if __name__ == '__main__':
    server_thread = threading.Thread(target=run_TCPServer)
    client_thread = threading.Thread(target=run_TCPClient)

    server_thread.start()
    client_thread.start()

    server_thread.join()
    client_thread.join()

#----------------------------------------------------------------------

 #udp_chat_server.py
import socket
import time
import threading

def run_UDPServer():
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    server_socket.bind(('127.0.0.1', 9999))
    data = server_socket.recvfrom(1024)[0]
    print(f"已收到消息{data.decode('utf-8')}")
    server_socket.close()

def run_UDPClient():
    time.sleep(0.5)
    client_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    msg = "测试UDP数据传输"
    client_socket.sendto(msg.encode('utf-8'), ('127.0.0.1', 9999))
    client_socket.close()

if __name__ == "__main__":
    t1 = threading.Thread(target=run_UDPServer)
    t2 = threading.Thread(target=run_UDPClient)

    t1.start()
    t2.start()

    t1.join()
    t2.join()   
````

````python
#udp_echo.py
import socket
import time
import threading

addr = ('127.0.0.1', 12345)
#receiver端
def run_receiver():
    with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as s:
        s.bind(addr)
        received_data,sender_addr = s.recvfrom(1024)
        received_msg = received_data.decode('utf-8')
        s.sendto(received_data,sender_addr)
        print(received_msg)
        print(sender_addr)
#sender端
def run_sender():
    time.sleep(1)
    with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as s:
        msg = "你好"
        s.sendto(msg.encode('utf-8'),addr)
        reply_data = s.recvfrom(1024)[0]
        reply_msg = reply_data.decode('utf-8')
        print(reply_msg)

if __name__ == "__main__":
    t1 = threading.Thread(target = run_receiver)
    t2 = threading.Thread(target = run_sender)

    t1.start()
    t2.start()

    t1.join()
    t2.join()
````

# 🐍 Python 学习笔记｜第 36 关：应用层协议: HTTP协议
`HTTP（HyperText Transfer Protocol，超文本传输协议）`: 是互联网上应用最为广泛的一种网络协议。它位于 OSI 模型或 TCP/IP 模型的应用层，是万维网（World Wide Web）的数据通信基石。
## 一、HTTP 的核心特点
1. 基于“请求-响应”模型（Request-Response）： HTTP 是一种客户端/服务器（C/S）架构的协议。永远是由客户端（如浏览器）主动发起请求，服务端（如网站服务器）接收并处理后返回响应。服务器不能主动向客户端推送数据（注：WebSocket 和 HTTP/2 Server Push 属于例外延伸）。
2. 无状态（Stateless）： HTTP 协议自身对事务处理没有记忆能力。也就是说，服务器处理完这次请求并关闭连接后，下一次请求过来时，服务器并不知道你还是刚才的那个人。
    💡 现实中如何解决无状态？
    为了实现“登录状态保持”或“购物车”功能，人们引入了 Cookie、Session 以及 Token（如 JWT） 技术，在 HTTP 之上模拟出“状态”。
3. 无连接（Connectionless）： 早期的 HTTP/1.0 每次请求都要建立一次 TCP 连接，处理完就断开。虽然现在的 HTTP/1.1 默认开启了长连接（Keep-Alive），但从协议设计的层面上看，每个请求依旧是独立的。
4. 灵活与媒体独立： 只要客户端和服务器知道如何处理，HTTP 可以传输任意类型的数据（文本、图片、视频、JSON 等）。这主要依赖于请求头中的 Content-Type 来标识。
## 二、 HTTP 请求与响应的结构(HTTP 的报文是纯文本格式的（在 HTTP/2 之前），这意味着它非常直观、可读性强。)
1. HTTP 请求报文（Request）
    * 请求行（Request Line）： 包含 请求方法（Method）、请求路径 和 协议版本。
        请求方法：最常见的 GET（获取资源）和 POST（提交数据），还有 PUT、DELETE、PATCH 等。
        请求路径：服务器上资源的地址，如 /api/users

    * 请求头部（Request Headers）： 键值对形式，用于说明客户端的环境、偏好等。例如 User-Agent（浏览器信息）、Accept（接收格式）。
    * 空行（Blank Line）： 必须有，用来分隔请求头和请求体。
    * 请求体（Request Body）： 并非所有请求都有。通常在 POST 或 PUT 请求中，用来存放要提交给服务器的数据（如表单数据、JSON）。

    ```text
    GET /index.html HTTP/1.1             ← 请求行：方法 + 路径 + 协议版本
    Host: www.example.com                ← 请求头（Headers）
    User-Agent: python-requests/2.28   
    Accept:text/html                   

    （空行）                              ← 空行（Blank Line）
    name=Alice&age=20                    ← 可选的请求体（Body）
    ```

2. HTTP 响应报文（Response）
    * 状态行（Status Line）： 包含 协议版本、状态码（Status Code）和 状态描述。例如：HTTP/1.1 200 OK
    * 响应头部（Response Headers）： 键值对形式，告知客户端服务器的信息、返回的数据类型等。例如 Content-Type: text/html、Server: Apache。
    * 空行（Blank Line）： 用来分隔响应头和响应体。
    * 响应体（Response Body）： 服务器返回给客户端的实际数据，比如 HTML 代码、图片二进制数据或 JSON 字符串。

    ```
    zHTTP/1.1 200 OK                   ← 状态行：协议版本 + 状态码 + 状态描述
    Content-Type: application/json     ← 响应头
    Content-Length: 85                 ←

    （空行）                            ←
    {"id": 1, "name": "Alice"}         ← 响应体
    ```

3. 常见请求头/响应头
    * Host： 请求的主机名，一台物理服务器上可能同时运行着好几个网站（比如 a.com 和 b.com）。当请求到达服务器时，服务器必须看 Host: a.com 才知道应该把这个请求交给哪个网站处理。
    * User-Agent：客户端身份标识，告诉服务器请求来自哪里。
    * Content-Type：指示请求体/响应体的格式，如 application/json、application/x-www-form-urlencoded。
    * Authorization：用于身份验证，如 Bearer token。
    * Cookie：客户端保存的会话信息，服务器用来识别用户状态。
    * Connection: keep-alive，它告诉服务器：“我传完这个数据先别挂断电话，我呆会儿还有别的东西要发。”

    在 requests 中，你可以自定义请求头：
    ```python
    headers = {
        'User-Agent': 'MyApp/1.0',
        'Authorization': 'Bearer your_token'
    }
    response = requests.get('https://api.example.com/data', headers=headers)
    ```
4. URL（Uniform Resource Locator）
    https://www.example.com:8080/path/to/index.html?name=jack&age=18#section1

    * 协议（Protocol）—— `https://`
        作用：告诉浏览器，这次去找服务器要数据用什么“交通工具”或“语言”。
        常见类型： http:// 或更安全的 https://（我们之前聊的状态码、请求头都是基于这两个协议的）。去下载文件时可能还会看到 ftp://。
    * 域名/主机名（Domain/Host）—— `www.example.com`
        作用： 互联网上有很多服务器，域名用来精准锁定是哪一家公司/哪台服务器。
        幕后： 域名其实是给人看的，计算机底层会把它翻译成一串数字组成的 IP 地址（比如 192.168.1.1）。这就像“故宫博物院”是域名，“东城区景山前街4号”是 IP 地址。
    * 端口（Port）—— :`8080`
        作用： 找到服务器后，用来进入哪扇特定的门（哪项服务）。
        注意： 平时我们上网很少看到端口，因为如果是 http 协议，浏览器会默认帮你填上 :80；如果是 https，默认是 :443。只有开发测试时，才会出现类似 :8080、:3000 这样的显式端口。
    * 路径（Path）—— `/path/to/index.html`
        作用： 进了大门后，你要找的具体文件在哪个房间（文件夹）。
        含义： 这代表你要访问服务器里 path 文件夹下 to 文件夹里的 index.html 网页文件。
    * 参数（Query String）—— `?name=jack&age=18`
        作用： 附带给服务器的额外条件（点菜时的特殊要求）。
        特征： 必须以问号 ? 开头，后面是以 & 符号连接的键值对。比如这里就是告诉服务器：“我要找一个名字叫 jack、年龄 18 岁的人的数据”。
    * 锚点/哈希（Fragment/Anchor）—— `#section1`
        作用： 网页加载完后，直接滚动到页面的特定位置。
        特征： 以 # 开头。比如很多长文章的“目录点击跳转”，就是靠锚点实现的。它不会发送给服务器，纯粹由浏览器自己在本地定位。
5. 处理URL参数
**requests.post(url,[data=dict_or_str]∣[json=dict_or_list])**

POST请求时参数不同决定了请求头里的Content-Type会变成什么
* data参数（默认表现为表单提交）
    底层自动加的请求头：Content-Type: application/x-www-form-urlencoded
    服务器看到的数据格式：name=jack&age=18（传统的网页表单格式）。
```python
payload = {'name': 'jack', 'age': 18}
response = requests.post('https://api.com', data=payload)
```
* json参数
    底层自动加的请求头：Content-Type: application/json
    服务器看到的数据格式：{"name": "jack", "age": 18}（纯正的 JSON 字符串）。
```python
payload = {'name': 'jack', 'age': 18}
response = requests.post('https://api.com', json=payload)
```

GET请求经常需要在URL后带查询参数，requests 提供了 params 参数，自动帮你拼接：

```python
params = {
    'q': 'python',
    'page': 1
}
response = requests.get('https://api.example.com/search', params=params)
print(response.url)  # 查看完整URL
```

## 三、 核心概念：请求方法、状态码、HTTP历代版本 & HTTP 与 HTTPS 的区别
1. requests请求方法
`.GET(url)`：获取数据（查）
    应用场景：你在浏览器里输入网址看新闻、刷小红书、或者用 Python 爬虫抓取网页。
    特点：参数在 URL 中；安全且幂等（Idempotent）
`.POST(url,data)`：提交数据（增）
    应用场景：账号注册/登录、发布朋友圈、上传文件、付款。
    特点：参数在请求体（Body）中；非幂等
`.PUT(url,data)`：更新数据（改）
    应用场景：修改个人资料、修改密码。
    特点：通常是完全替换；幂等
`.DELETE(url)`：删除数据（删）
    应用场景：注销账号、删除某条朋友圈、清空购物车。
    特点：幂等
`.PATCH`：局部更新。和 PUT 类似，但 PUT 是整体替换，PATCH 是局部修改
`.HEAD`：只要头部信息。和 GET 一样，但服务器不返回具体内容（Body），只返回状态码和文件大小等。
`.OPTIONS`：探测询问。客户端（通常是浏览器）问服务器：“你支持哪些请求方法？允许我跨域访问吗？”

```python
import requests
# 1. 发送 GET 请求（获取一篇文章）
response = requests.get('https://api.example.com/posts/1')
print(response.json())
# 2. 发送 POST 请求（创建一篇文章）
new_post = {'title': '学会了HTTP', 'content': '今天真开心'}
response = requests.post('https://api.example.com/posts', json=new_post)
# 3. 发送 PUT 请求（修改这篇文章）
updated_post = {'title': '学会了HTTP方法', 'content': '精通了GET和POST'}
response = requests.put('https://api.example.com/posts/1', json=updated_post)
# 4. 发送 DELETE 请求（删除这篇文章）
response = requests.delete('https://api.example.com/posts/1')
```


2. 解析响应
response： 这个response变量就像是一个“集装箱”，它把服务器回传的所有 HTTP 响应数据（状态行、响应头、空行、响应体）全部打包塞进了自己的属性里。
* 映射“状态行”：看结果
    `response.status_code`  # 获取状态码 (整型，如 200)
    `response.reason`       # 状态描述（如 OK、Not Found）。
* 映射“响应头部”：看规则
    `response.headers`      # 获取响应头 (字典格式) - 检查服务器返回的数据格式或冷知识。例如通过 response.headers.get('Content-Type') 确认接下来该怎么解析数据。
* 映射“响应体”：拿数据（最核心）
    `response.text`         # 获取文本格式的网页/数据 (自动解码)
    `response.content`      # 获取二进制数据 (下载图片、文件时用)
    `response.json()`       # 直接将 JSON 响应解析为 Python 字典/列表

2. 状态码

| 状态码范围 | 含义 | 常见例子 |
| :--- | :--- | :--- |
| **1xx** | 信息提示 | *无常用（多为协议切换等过渡状态）* |
| **2xx** | 成功 | `200 OK`（请求成功）<br>`201 Created`（已创建） |
| **3xx** | 重定向 | `301 Moved Permanently`（永久重定向）<br>`302 Found`（临时重定向） |
| **4xx** | 客户端错误 | `400 Bad Request`（请求有语法错误）<br>`404 Not Found`（资源不存在） |
| **5xx** | 服务器错误 | `500 Internal Server Error`（服务器内部错误）<br>`502 Bad Gateway`（网关错误） |
3. HTTP推进过程

| 版本 | 核心升级点 | 经典打比方 | 依然存在的遗憾 |
| :--- | :--- | :--- | :--- |
| **HTTP/1.0** | 开启了万维网时代 | **单件快递**：送完一件货，就得把车砸了重新买一辆。 | 每次请求都要重新建连接，太慢。 |
| **HTTP/1.1** | 默认支持长连接 | **一辆大货车**：车子可以一直开，但里面的货必须一件一件按顺序拿。 | “队头阻塞”：前面卡住，后面全等。 |
| **HTTP/2.0** | 多路复用、头部压缩 | **一条传送带**：货物拆成小零件混在一起运，到了终点再拼。 | “底层卡死”：如果丢了一个零件，整条传送带立刻停工。 |
| **HTTP/3.0** | 抛弃 TCP，拥抱 QUIC (UDP) | **多架无人机**：各自飞各自的，其中一架坠毁了，其他的完全不耽误送货。 | 普及需要时间，部分老旧路由器甚至会直接拦截 UDP 流量。 |

4. HTTP 与 HTTPS 的区别
    * 安全性：明文传输 vs 加密传输
        HTTP（不安全）：所有的数据（包括你的登录账号、密码、银行卡号）都是明文（纯文本）传输的。如果网络中间有黑客（比如你连了一个公共场所的恶意 Wi-Fi），黑客可以用简单的抓包工具直接看光你的隐私。
        HTTPS（安全）：在 HTTP 的下面加了一层加密协议（历史上叫 SSL，现在升级为 TLS）。数据在离开你的电脑前会被打乱成一堆谁也看不懂的乱码，即使中途被黑客截获，他也无法解密。
            非对称加密（RSA/ECC）：用来在最开始“交换密钥”。服务器用公钥，浏览器用私钥。这个过程很慢，但很安全，目的是为了让双方安全地对上一个“只有两方知道的暗号（对称密钥）”。
            对称加密（AES）：一旦暗号对上了，后面传输大量的网页、图片数据时，全部改用这个暗号进行“对称加密”。因为对称加密的速度极快，不吃 CPU。
    * 身份验证：防止“李鬼”冒充“李逵”
        HTTP：只要建立连接，就傻傻地开始传输，无法确认对方到底是不是真正的官方网站。
        HTTPS：引入了 CA 证书（数字证书） 机制。网站上线 HTTPS 前，必须向国际权威机构（CA）申请一张“电子身份证”。当你访问这个网站时，浏览器会强行检查这张证书。如果证书是真的，浏览器地址栏就会显示一个小绿锁 🔒；如果是假的（比如黑客伪造的钓鱼网站），浏览器会直接弹出红色的“您的连接不是私密连接”警告。
    * URL 门牌号与默认端口的不同
        HTTP: 网址开头：http://
              默认端口：:80（浏览器会自动隐藏）
        HTTPS: 网址开头：https://
               默认端口：:443（浏览器会自动隐藏）
    * 性能与成本
        HTTP:  完全免费，而且因为不需要加密解密，对服务器的 CPU 消耗极低，速度理论上很快。
        HTTPS: 成本：购买高级的商用 CA 证书需要花钱（虽然现在也有很多免费证书）。
               性能：因为多了加密解密的计算过程，而且在连接之初要多进行几次“安全握手”，理论上会比 HTTP 慢一点点。但现代计算机硬件非常强大，配合 HTTP/2 和 HTTP/3 的优化，这种延迟人类已经完全感知不到了。
## 四、练习

1. 编写一个 API 调用程序
    创建 get_user_info.py，使用 requests 向 https://jsonplaceholder.typicode.com/users/1 发送 GET 请求。
    * 打印返回的状态码和响应头中的 Content-Type。
    * 将响应体解析为 JSON，打印用户的 name、email 字段。
    * 如果请求失败（状态码不是2xx），打印错误信息。
2. 模拟登录表单提交
    创建 login_post.py，向 https://httpbin.org/post 发送 POST 请求，提交一个包含 username 和 password 的字典（用 data 参数而非 json，模拟表单提交）。
    * 打印服务器返回的 JSON 中 form 字段的内容，确认数据已正确提交。
    * 自定义一个请求头 X-Custom-Header: PythonLearning，在请求中发送，并打印响应 JSON 中的 headers 字段，查看自定义头是否被服务器回显。
    
    提示：httpbin.org 是一个专门用于测试 HTTP 请求的站点，它会将你发送的请求信息原样返回，非常适合练习。练习时注意网络访问是否正常。


````python
#get_user_info.py
import requests
import re
def request_get(url):
    response = requests.get(url)
    status_code = response.status_code
    print(status_code)
    if re.match(r'^2\d*$',str(status_code)):
        #方法 A：数字范围判断：if 200 <= status_code < 300:
        #方法 B：利用 response.ok 属性。这是 requests 贴心提供的布尔值，当状态码在 200-299 之间时它自动为 True。

        headers_dict = response.headers
        content_type = headers_dict['Content-Type']
        content_dict = response.json()
        user_name = content_dict['name']
        user_email = content_dict['email']
        print()
        print(content_type)
        print(user_email)
        print(user_name)
    else:
        print(response.reason)

url = "https://jsonplaceholder.typicode.com/users/1"
request_get(url) 
````

````python
#login_post.py
import requests

def login_post(url):
    user_account = {'user_name':'shuixiaoyuana','password':'123123123'}
    response = requests.post(url, data = user_account)
    response_dict = response.json()
    print(response_dict.get('form'))

    header = {'X-Custom-Header': 'PythonLearning'}
    response = requests.post(url, data = user_account, headers=header)
    print(response.json().get('headers').get('X-Custom-Header'))

url = "https://httpbin.org/post"
login_post(url)
````

# 🐍 Python 学习笔记｜第 37 关：数据库基础：关系型数据库与 SQL 语句初探
## 一、 数据库
1. 基础概念
数据库：用来有组织地存储和管理数据的系统。关系型数据库是其中最经典的一种，它把数据组织成表（类似 Excel 表格），表由行（记录）和列（字段）组成。我们可以通过SQL（Structured Query Language 结构化查询语言）来对表进行增删改查操作。
SQL（Structured Query Language 结构化查询语言）： 如果把数据库想象成一个电子化的大型文件柜，那么 SQL 就是你用来存取、修改和管理这些文件的“普通话”。
    查询（Read）：从数据库里找出你想要的数据。例如：“帮我找出所有住在北京且年龄大于25岁的用户。”
    插入（Create）：把新数据存入数据库。例如：“有一个新用户注册了，把他的信息录入进去。”
    更新（Update）：修改已经存在的数据。例如：“用户修改了密码，更新一下他的密码记录。”
    删除（Delete）：把不需要的数据删掉。例如：“用户注销了账号，删除他的个人资料。”

2. 基础 SQL 语句

2. 1.  *准备工作：建立一张表（CREATE TABLE）:* 在往数据库里塞数据之前，我们得先规划好数据长什么样。这就像在 Excel 里，你得先建一个工作表，并写好表头（列名）。
    语法：
    `CREATE TABLE IF NOT EXISTS 表名 (列名 数据类型 约束条件1 约束条件2…, 列名 数据类型 约束条件1 约束条件2…,);`
    列名（Column Name）： 你给这一列起的名字（如 id, name, age）。
    数据类型（Data Type）： 规定这一列能存什么样的数据（如 INTEGER 整数, TEXT 文本）。
        `INTEGER`：带符号的整数。
        `REAL`：浮点数（小数）。
        `TEXT`：字符串文本。
        `BLOB`：二进制大对象（比如用来存图片、音频文件，虽然一般不推荐这么做）。
        `NULL`：空值。
    约束条件（Constraints）： 对这一列的数据做限制或赋予特殊功能（如 NOT NULL 不能为空）。
    逗号（,）： 每定义完一列，用英文逗号隔开，最后一列后面不要加逗号。
        `PRIMARY KEY`：主键。一张表只能有一个，里面的值必须唯一且不能为 NULL。
        `NOT NULL`：强制该列不能输入空值。
        `UNIQUE`：确保这一列的所有值都是唯一的（比如身份证号、邮箱，不能重复）。
        `DEFAULT [值]`：默认值。如果插数据时没填这一列，系统会自动填入默认值（例如：status TEXT DEFAULT 'active'）。
        `AUTOINCREMENT`: 自动递增, 必须和 INTEGER PRIMARY KEY 一起使用 - 每次你向表里插入一条新记录，SQLite 会自动为这一列生成一个比之前更大的整数。在 SQLite 中，如果你只写 INTEGER PRIMARY KEY（不加 AUTOINCREMENT），它默认也会自增。加上 AUTOINCREMENT可以防止 ID 被复用。即使把前面的数据全删光，ID 也会一直往前走，绝对不回头。

        ```SQL
        CREATE TABLE IF NOT EXISTS students (      
            id INTEGER PRIMARY KEY AUTOINCREMENT, -- 建立一个列名为id，类型是INTEGER约束条件是PRIMARY KEY AUTOINCREMENT
            name TEXT NOT NULL,
            age INTEGER,
            score REAL
        );
        ```

2. 2. *增：插入数据（INSERT）*: 表建好了，现在我们要往里面填入第一行数据。
    语法：
    `INSERT INTO 表名 (列名1, 列名2) VALUES (值1, 值2);`: 字符串类型（TEXT）的值，必须用单引号`' '` 包起来；数字则不用。因为 id 是自增的主键，所以这里我们根本不需要写 id，它会自动变成 1。
   
        ```SQL
        INSERT INTO students (name, age, score) VALUES ('张三', 18, 92.5);
        ```

2. 3. *查：获取数据（SELECT）*：这是数据库最强大的功能。你想看什么，就用 SELECT 把它挑出来。
    语法：
    `SELECT 列名1, 列名2 FROM 表名 WHERE 条件;`
        各种查法示例: 

        ```SQL
        -- 查所有人的所有信息（最常用，* 代表所有列）：
        SELECT * FROM students; 

        -- 只看所有人的名字和分数（不看年龄和 ID）：
        SELECT name, score FROM students;

        -- 条件查询（带上 WHERE）： 只看分数大于 90 分的优秀学生。
        SELECT * FROM students WHERE score > 90;

        -- 精准找人： 找名字叫“张三”的学生。
        SELECT * FROM students WHERE name = '张三';
        ```

2. 4. *改：更新修改数据（UPDATE）*: 如果张三期末考试成绩复核，分数从 92.5 变成了 95，我们就需要修改它。
    语法：
    `UPDATE 表名 SET 列名 = 新值 WHERE 条件;`: ⚠️ 极度重要： 写 UPDATE 的时候，千万千万不能漏掉 WHERE 条件！ * 如果你写成 UPDATE students SET score = 95.0;（没加 WHERE），数据库会认为你要把全校所有学生的分数全部改成 95 分！这就变成教学事故了。

        ```SQL
        UPDATE students SET score = 95.0 WHERE name = '张三';
        ```

2. 5. *删：删除数据（DELETE）*: 如果某个学生退学了，我们需要把他的记录删掉。
    语法：
    `DELETE FROM 表名 WHERE 条件;`: ⚠️ 同样极度重要： 写 DELETE 也绝对不能漏掉 WHERE 条件！ * 如果直接写 DELETE FROM students;，意味着清空整张表，所有学生的数据瞬间人间蒸发。通常在企业里，我们会根据唯一的 id 来删除，这样最精准，不会误伤别人。

    ```SQL
    DELETE FROM students WHERE id = 1;
    ```

## 二、 sqlite3 模块

1. 基本概念：Python 内置了 sqlite3 模块，支持轻量级数据库 SQLite，无需额外安装服务器，非常适合学习。数据库文件就是一个 .db 文件，方便携带。sqlite3 模块本质上是在扮演一个“传话筒”的角色：你的 Python 代码是甲方，SQLite 数据库是乙方，而 sqlite3 模块负责把你的想法翻译成数据库能懂的语言送过去。

    [ 你的 Python 程序 ] 
        │
    1. 连接 (Connection) ── 铺设一条去数据库的路：它是 Python 程序和数据库文件之间建立的通道。
        │
    2. 游标 (Cursor)     ── 雇佣一个跑腿的服务员：它是具体执行命令、搬运数据的工具
        │
    3. SQL 语句          ── 菜单和指令

2. 相关语法：
    `import sqlite3` - 导入sqlite3模块
    `connection = sqlite3.connect('name.db')` - 创建连接，连接数据库，如果文件不存在，会自动创建。
    `cursor = connection.cursor()` - 创建游标（Cursor）：它是执行 SQL 语句并获取结果的“指针”
    `cursor.execute(SQL语句)` - 执行 SQL
    `cursor.fetchone() / .fetchall()`：意思是“把执行完的结果端过来”。
    `connexction.commit()` - 确认保存修改。
    `cursor.close() & conn.close()` - end进程释放内存

3. 1. 核心五步法（标准模板）: 在 Python 中操作 SQLite3，核心流程可以总结为 5 个步骤：建立连接 -> 创建游标 -> 执行 SQL -> 提交事务 -> 关闭连接。

    ```python
    import sqlite3
    # 1. 建立连接：如果文件不存在，会自动在当前目录下创建
    conn = sqlite3.connect('students.db')
    # 2. 创建游标（Cursor）：它是执行 SQL 语句并获取结果的“指针”
    cursor = conn.cursor()
    # 3. 执行 SQL：创建一张学生表
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS students (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            age INTEGER,
            score REAL
        )
    ''')
    # 4. 提交事务：对数据库做了增、删、改的操作后，必须 commit 才能真正保存
    conn.commit()
    # 5. 关闭资源：好的习惯是及时关闭游标和连接
    cursor.close()
    conn.close()
    ```

3. 2. 核心操作：增、删、改、查
    1. 插入数据（增）与 防止 SQL 注入

        ```python
        # 单条插入：传入一个元组 (Tuple)
        sql_single = "INSERT INTO students (name, age, score) VALUES (?, ?, ?)"
        cursor.execute(sql_single, ('Alice', 20, 95.5))

        # 批量插入：使用 executemany 传入列表，效率极高
        sql_batch = "INSERT INTO students (name, age, score) VALUES (?, ?, ?)"
        data = [
            ('Bob', 22, 88.0),
            ('Charlie', 21, 74.5),
            ('David', 23, 91.0)
        ]
        cursor.executemany(sql_batch, data)
        conn.commit()
        ```

        1. 补充：占位符 `?`的用法
        在 Python 的 sqlite3 中，我们使用 ? 作为占位符，然后把数据放进一个元组（Tuple）或列表（List）中传给 execute() 函数。
        注意1：不管数据是文本还是数字，? 外面绝对不能加任何单引号或双引号！
        注意2：把数据打包成一个元组 (元组里只有一个元素时，末尾必须加逗号)

        2. 问答
            问： 
            为什么不直接使用 sql_single = "INSERT INTO students (name, age, score) VALUES ('Alice', 20, 95.5)"
            答：
            1. 致命的安全问题：SQL 注入攻击（SQL Injection

                    如果你习惯用字符串拼接的方式来传参，比如这样写：

                    ```Python
                    # 假设 name 是用户在网页输入框里填写的
                    user_input = "Alice"
                    sql = f"INSERT INTO students (name) VALUES ('{user_input}')"
                    ```
                
                    如果用户是个黑客

                    ```Python
                    # 黑客输入
                    user_input = "Alice'); DROP TABLE students; --"
                    sql = f"INSERT INTO students (name) VALUES ('{user_input}')"

                    #拼接后sql= "INSERT INTO students (name) VALUES ('Alice'); DROP TABLE students; --')"
                    #sql语句变成 INSERT INTO students (name) VALUES ('Alice');  - 数据库增加name信息
                    #           DROP TABLE students;                           - 删除table
                    #           --')                                           - 将')变成注释
                    ```
                
                    结果： 数据库不仅插入了数据，还顺便把你的 students 表给彻底删除了（DROP TABLE）！
                    解决办法：使用占位符 ? . 数据库会把 ? 接收到的值严格当成普通的字符串或数字处理。即使黑客输入了 DROP TABLE，数据库也只会把这一整串安全地存进 name 字段里，而不会把它当成代码去执行。当你使用 ? 时，SQLite 数据库内部会执行两个步骤：先把 INSERT INTO students (name) VALUES (?) 编译好，数据库这时候已经确定了：“这句命令唯一要干的事就是插入一个名字，绝对没有第二句命令！”随后，它把黑客输入的 Alice'); DROP TABLE students; -- 当成一个纯文本字符串，死死地塞进了 ? 那个坑位里。

                    ```python
                    user_input = "Alice'); DROP TABLE students; --"
                    user_input_tuple = (user_input,)
                    sql = "INSERT INTO students (name) VALUES (?)"
                    #即使拼接完了也是INSERT INTO students (name) VALUES ("Alice'); DROP TABLE students; --")
                    cursor.execute(sql, data_tuple)
                    ```

                2. 避免字符转义的“语法恶梦”
                如果直接拼字符串，当数据本身带有特殊符号时，SQL 语句就会直接崩溃。
                    举个例子：
                    你想插入一个英文名字叫 O'Neal（奥尼尔）的学生。
                    直接拼接： VALUES ('O'Neal', 20, 95.5)
                    结果： SQL 看到 O' 就认为字符串结束了，后面的 Neal' 会引发严重的语法错误（Syntax Error）。
                    在 SQL 的世界里，单引号 ' 的唯一功能就是“字符串的边界围栏”。它就像一盏红绿灯：看到第一个单引号，代表字符串“开始”；看到第二个单引号，代表字符串“结束”。

                3. 批量插入时无法复用代码
                如果你不用 ? 占位符，你就得写一个 for 循环，在循环内部不停地用 f-string 拼接出好几条不同的 SQL 语句，然后一条一条去 execute。
                使用 ? 的优势： sql_batch 这句模板只需要定义一次。数据库会对这句结构固定的 SQL 进行预编译（Prepare），然后像流水线机器一样，直接把 data 列表里的数据一个个套进 ? 里发射出去。这样不仅代码优雅，执行效率也比一条条拼字符串执行快得多。
    2. 查询数据（查）的几种方式
        查询不需要 conn.commit()，但需要通过游标来获取（fetch）结果。
    
        ```python
        cursor.execute("SELECT * FROM students WHERE age > ?", (21,))

        # 方式 A：获取一条数据 (返回一个元组)
        row = cursor.fetchone()
        print("单条数据:", row) # 输出: (2, 'Bob', 22, 88.0)

        # 方式 B：获取剩下的所有数据 (返回元组列表)
        rows = cursor.fetchall()
        for r in rows:
            print("循环遍历:", r)

        # 方式 C：直接遍历游标（推荐，省内存）
        cursor.execute("SELECT * FROM students")
        for r in cursor:
            print("高效遍历:", r)
            3. 更新与删除（改/删）
        ```
        注意： Cursor指针进行一次execute时只会从上往下执行，不会检查完一条后回到开头执行，所以方式B是获取剩下的数据而不是全部
        补充： 直接 for r in cursor，其实是 Python 在后台自动、隐式地帮你一条一条地去执行 fetchone()。读一条，打印一条，销毁一条，再读下一条。
    3. 更新与删除（改/删）
        记得调用 conn.commit() 提交修改。

        ```python
        # 更新 Bob 的分数为 90
        cursor.execute("UPDATE students SET score = ? WHERE name = ?", (90.0, 'Bob'))

        # 删除 id 为 3 的学生
        cursor.execute("DELETE FROM students WHERE id = ?", (3,))

        conn.commit()
        print(f"受到影响的行数: {cursor.rowcount}")
        ```

3. 3. 进阶进阶：Pythonic 的高级技巧

    1. 技巧 1：使用 with 语句自动管理连接（推荐）: 每次都写 close() 很麻烦，用 with 可以确保发生异常时也能正确关闭或回滚。
        ```Python
        import sqlite3
        # with conn 会在执行完毕后自动 commit，如果中途报错则会自动 rollback
        with sqlite3.connect('students.db') as conn:
            cursor = conn.cursor()
            cursor.execute("UPDATE students SET score = 100 WHERE name = 'Alice'")
            # 无需手动 conn.commit()，with 帮你想着呢
            
        # 注意：连接依然需要手动关闭，或者配合 contextlib，但通常写在函数中很安全
        ```

    2. 技巧 2：让查询结果变成“字典”形式: 默认情况下，fetchall() 返回的是元组 (1, 'Alice', 20)，你只能通过下标 row[1] 访问名字。通过修改 row_factory，可以像字典一样通过列名访问.
    
        ```Python
        conn = sqlite3.connect('students.db')
        # 关键这一行：将行工厂设置为 sqlite3.Row
        conn.row_factory = sqlite3.Row

        cursor = conn.cursor()
        cursor.execute("SELECT * FROM students")

        row = cursor.fetchone()
        print(row['name'])  # 输出: Alice
        print(row['score']) # 输出: 95.5
        print(row.keys())   # 输出所有的列名: ['id', 'name', 'age', 'score']
        ```

    3. 技巧 3：在内存中创建临时数据库：如果你只是想写个测试，不想在硬盘上生成一个 .db 文件，可以使用 :memory:：
        ```Python
        # 这是一个完全存在于内存中的数据库，程序结束运行后自动消失
        conn = sqlite3.connect(':memory:')
        ```

    * 插入至少三本书的信息（如《三体》、《活着》、《1984》）。

提示：使用 input() 获取用户输入，并将输入作为参数传递给 SQL 的 WHERE 子句（使用 ? 占位符）。记得处理可能的异常（如数据库文件不存在）。

````python
#library.py

import sqlite3
conn  = sqlite3.connect(':memory:')
cursor = conn.cursor()

def setup_books():
    cursor.execute('''CREATE TABLE IF NOT EXISTS books(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT NOT NULL,
        author TEXT,
        year INTEGER
    )''')
    conn.commit()
    
def insert_data(data):
    sql = "INSERT INTO books(title, author, year) VALUES (?,?,?)"
    cursor.executemany(sql,data)
    conn.commit()

def search_title_and_author():
    sql = "SELECT title, author FROM books"
    cursor.execute(sql)
    rows = cursor.fetchall()
    for r in rows:
        print(r)


books_data = [('《三体》', '刘慈欣', 2000),('《活着》', '余华', 2001),('《西游记》', '吴承恩', 1700)]
setup_books()
insert_data(books_data)
search_title_and_author()


#这里不能写在里面     在编写数据库代码时，有一个至高无上的原则：“谁负责开门，谁就负责关门。”
# 如果 conn = sqlite3.connect(...) 是写在函数外面的（全局变量），那关闭的代码 conn.close() 就必须写在函数外面（通常放在整个文件的最后一行）。
# 如果 conn = sqlite3.connect(...) 是写在函数里面的（局部变量），那关闭的代码就必须写在函数里面的最后。


cursor.close()
conn.close()
````

````python
#query_book.py
import sqlite3
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
books_data = [('《三体》', '刘慈欣', 2000),('《活着》', '余华', 2001),('《西游记》', '吴承恩', 1700)]

def setup_books():
    cursor.execute('''CREATE TABLE IF NOT EXISTS books(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT NOT NULL,
        author TEXT,
        year INTEGER
    )''')
    conn.commit()
    
def insert_data(data):
    sql = "INSERT INTO books(title, author, year) VALUES (?,?,?)"
    cursor.executemany(sql,data)
    conn.commit()

def search_book():
    data_input = input("请输入作者姓名查询图书:")

    sql = "SELECT title, year FROM books WHERE author = ? "
    cursor.execute(sql, (data_input,))
    rows = cursor.fetchall()
    if rows:
        print(rows)
        
    else:
        print("未找到该作者")
    #如果数据很多的话用下面的方法，否则可能撑爆内存
    #has_data = False                                --------  默认“有数据”为假事件
    #for row in cursor:
    #   print(f"书名：{row[0]} - {row[1]}")           
    #   has_data = True                              --------  “有数据”变成真事件
    #if not has_data:                                --------   这里只有逻辑变成True才会print，所以需要给has_data前加上not逻辑取反
    #   print("未找到该作者")
    # has_data 本身：确实是 False（代表“有数据”这件事是假的）。
    # not has_data 这一整串：结果是 True（代表“没有数据”这件事是真的）。
    #虽然“钱包里有钱”是假的，但“钱包里没钱”这个事实却是百分之百真的！

setup_books()
insert_data(books_data)
search_book()


cursor.close()
conn.close()
````

# 🐍 Python 学习笔记｜第 37.5 关：字符串相关知识点
1. 基础声明语法
单引号、双引号或三引号,它们在基础功能上是完全等价的。

    ```python
    #单引号
    str1 = 'Hello World'
    #双引号(如果字符串内部包含单引号，用双引号会很方便)
    str2 = "I'm fine"
    #三引号(支持多行文本和包含换行符)
    str3 = """这是一个
    多行字符串"""
    ```

2. 转义字符 `\`
`\n`    换行符           'Hello\nWorld'
`\t`    制表符（Tab）     'Name:\tJack'
`\\`    反斜杠本身        'C:\\Users\\Name'

3. 原始字符串
如果你不希望反斜杠 \ 被当作转义字符处理（例如在写正则表达式或Windows文件路径时），可以在字符串前面加上 r 或 R。

        ```python
        # 普通字符串，\n 会导致换行
        path1 = "C:\new_folder\test" 

        # 原始字符串，\n 保持原样
        path2 = r"C:\new_folder\test"
        ```

4. 索引与切片 (Indexing & Slicing)
    ① 索引 (Indexing): 正向索引从 0 开始，反向索引从 -1 开始
    
        ```python
        s = "Python"
        print(s[0])   # 'P'
        print(s[-1])  # 'n' (倒数第一个字符)
        ```
    ② 切片 (Slicing)：语法为：string[start:end:step]（包含 start，不包含 end）。start：开始索引 end: 结束索引 step: 步长. 
    补充：
    两个::连着写不是独立新符号，只是省略了中间的参数
    第一个冒号前没写：代表 start 采用默认值（从头开始，即索引 0）。
    两个冒号中间没写：代表 end 采用默认值（一直切到字符串的末尾）。
    :: 的默认起点和终点，并不是固定不变的，而是会根据 step（步长）的正负号自动“变脸”
        如果 step 是正数（比如 1 或 2）
            Python 会认为你想从左往右读。
            默认起点 = 字符串的最左端（索引 0）
            默认终点 = 字符串的最右端
        如果 step 是负数（比如 -1）
            Python 看到负号，立刻明白你想从右往左读。
            默认起点 = 字符串的最右端（倒数第一个字符，即索引 -1）
            默认终点 = 字符串的最左端 所以，当写下 s[::-1] 时，它的起点自动变成了最右边的 'n'，然后每次 -1 往左倒着走，直到把整个字符串“倒着”拿完。

        ```python
        s = "Hello World"
        print(s[0:5])    # 'Hello' (截取索引 0 到 4)
        print(s[6:])     # 'World' (从索引 6 开始一直到结束)
        print(s[:5])     # 'Hello' (从头开始截取到索引 4)
        print(s[::2])    # 'HloWrd' (每隔2个字符取一个，步长为2)
        print(s[::-1])   # 'dlroW olleH' (巧妙利用负步长实现字符串反转)
        ```

5. 字符串常用操作符
`+` 拼接
`*` 重复
`in/not in` 成员运算符，检查某个子串是否存在于字符串中

    ```python
    print("py" in "Python") #打印True
    ```

6. 常用字符串内置方法
Python 提供了丰富的内置方法来处理文本，注意这些方法不会改变原字符串，而是返回一个新的字符串。
`.upper()`
`text.strip()`
`.replace(替换字符串, 被替换字符串)`
`.split()`
`连接符字符串.join(要连接的序列)` 只能连接字符串

```python
text = "  python code  "

print(text.upper())      # '  PYTHON CODE  ' (全大写)
print(text.strip())      # 'python code' (去除两端空格)
print(text.replace("code", "program")) # '  python program  ' (替换)
print(text.split())      # ['python', 'code'] (按空格分割成列表)

words = ['Hello', 'World']
print(" ".join(words))   # 'Hello World' (用空格将列表拼接成字符串)
```

7. 布尔值的返回
    * 直接返回True 或 False
        ```python
        def is_even(number):
        if number % 2 == 0:
            return True
        else:
            return False
        ```
    * 直接返回表达式的结果（更优雅、更推荐）
        比较运算符（如 ==, >, <, != 等）本身就会计算出一个布尔值。因此，你不需要写 if-else，直接返回表达式即可。
        ```python
        def is_even(number):
        # 如果 number % 2 == 0 成立，会自动返回 True，否则返回 False
        return number % 2 == 0
        ```
    * 使用 bool() 函数进行显式转换
        如果你想把其他类型的值（如数字、字符串、列表）转换成布尔值并返回，可以使用内置的 bool() 函数。
            Python 中的“隐式布尔值”（Truthy & Falsy）：
            会被转为 False 的值： 0、None、空字符串 ""、空列表 []、空字典 {} 等。
            正、负非零数字、非空对象 都会被转为 True。
        ```python
        def has_content(text):
        # 如果 text 是空字符串 ""，则返回 False；如果有内容，返回 True
        return bool(text)
        ```
    * 结合逻辑运算符（and, or, not）
        也可以返回多个条件组合后的布尔值。
        ```python
        def is_valid_age(age):
        # 只有在 18 到 65 之间时才会返回 True
        return age >= 18 and age <= 65
        ```

# 🐍 Python 学习笔记｜第 38 关：单元测试基础：使用 unittest 编写测试用例
## 一、单元测试
1. 基础概念：单元测试是对程序中最小的可测试单元（如一个函数、一个类的方法）进行检查和验证的工作。就像制造零件时，每生产一个零件就立刻检验它是否合格，而不是等整台机器装好才发现问题。在 Python 中，标准库 unittest 提供了编写和运行单元测试的完整框架。
2. 核心思想：给定输入，断言输出。测试框架会自动运行所有测试用例，并报告哪些通过、哪些失败。

## 二、 unittest库

### 核心概念一: TestCase（测试用例）
1. Test Case（测试用例）：最小的测试单元。它检查特定输入组合下的特定响应。unittest 提供了 TestCase 基类来创建新的测试用例。

2. 测试用例基础语法：
    `test*.py` - 测试文件名必须以test开头.py结尾。否则unittest的discover机制找不到。
    `import unittest` - 导入unittest库
    `class TestMathUtils(unittest.TestCase):` - 创建测试类(类名最好能体现被测对象)并继承unittest.TestCase类，如果你的类不继承 TestCase，即使方法名以 test_ 开头，测试运行器也会直接视而不见，把它当成一个普通的开发类。
        1.让测试运行器（Runner）能够“识别”这个文件（只有文件中的类继承了unittest.TestCase并且文件命名符合上述规则时，Runner才能识别这个文件）
        2.获得了丰富的“断言（Assertions）”工具箱
    `from math_utils import add, divide` - 从被测试文件math_utils.py中导入被测试方法用于测试
    `test_` - 测试方法名必须以 test_ 开头。unittest 运行器是通过这个前缀来识别哪些方法是测试用例的。不是以 test_ 开头的方法不会被当作测试自动执行。
    `self.assert*` - 使用断言验证结果 - 断言（Assertion）：测试的核心。我们不用 assert 关键字，而是使用 TestCase 提供的各种self.assert* 方法，这样当测试失败时能提供更详细的错误信息。常用断言方法（Assertions）：
        `self.assertEqual(a, b)`	a == b
        `self.assertNotEqual(a, b) `	a != b
        `self.assertTrue(x)	bool(x)` is True
        `self.assertFalse(x)`	bool(x) is False
        `self.assertIs(a, b)`	a is b
        `self.assertIsNone(x)`	x is None
        `self.assertIn(a, b)`	a in b（适用于字符串、列表、字典等）
        `self.assertRaises(Error)`	验证是否抛出了指定的异常（通常配合 with 使用）

    `unittest.main()` - 是一个极其方便的入口函数（Entry Point），它的主要作用是自动查找并运行当前脚本中的所有测试用例。

    ```python 被测试代码
    #math_utils.py
    def add(a, b):
        return a + b

    def divide(a, b):
        if b == 0:
            raise ValueError("Cannot divide by zero")
        return a / b
    ```

    ```python 编写测试代码
    #test_math.py                               #测试文件/模块
    import unittest
    from math_utils import add, divide

    class TestMathUtils(unittest.TestCase):     #测试类
        
        # 基础测试
        def test_add_positive_numbers(self):   #测试用例一
            self.assertEqual(add(1, 2), 3)
            
        def test_add_negative_numbers(self):   #测试用例二
            self.assertEqual(add(-1, -1), -2)

        # 异常测试：确保触发特定异常
        def test_divide_by_zero(self):         #测试用例三
            with self.assertRaises(ValueError):
                divide(10, 0)

    if __name__ == '__main__':
        unittest.main()
    ```
    
3. 问答
问：为什么测试用例能找到被测试文件并将其导入测试用例？
答：Python 寻找 math_utils 时，它会从上到下依次去这些路径里找，只要在某个路径下找到了 math_utils.py，就会立刻停下并加载它。Python 默认的查找顺序通常是这样的：
[当前运行目录] ➔ [PYTHONPATH 环境变量] ➔ [标准库目录] ➔ [第三方库目录 (site-packages)]

### 核心概念二：TestSuite（测试套件）& 核心概念三：TestRunner（测试运行器）
1. Test Suite（测试套件）：多个测试用例的集合。它可以把多个测试用例组合在一起打包执行。
2. Test Runner（测试运行器）：负责执行测试并向用户提供结果的组件（可以是图形界面、文本界面或返回特定状态码）。
3. 结构：
    my_project/                                             项目文件夹
    │
    ├── tests/                                              test文件夹
    │   ├── __init__.py                                     __init__.py 是 Python 中的一个特殊标记文件。它的核心作用是：告诉 Python 解释器，“这个文件夹不是一个普通的文件夹，而是一个可以被导入的‘代码包’（Package）”。
    │   ├── test_login.py  （里面有 TestLogin 类）           测试文件/模块：可能包含多个测试用例
    │   └── test_cart.py   （里面有 TestCart 类）            测试文件/模块：可能包含多个测试用例
    │
    └── run_tests.py       （你的测试套件入口文件）           测试入口，负责组装和运行测试套件。
3. 基础语法：
    `suite = unittest.TestSuite()`
    `suite.addTest(测试类名('方法名'))`向套件中添加具体的测试用例
        批量添加整个类的用例：
        `cases = unittest.makeSuite(测试类名)` 将类里的所有测试用例打包成一个用例包
        `suite.addTest(cases)` 向套件中添加整个类下的所有测试用例
    `runner = unittest.TextTestRunner(verbosity=)` 实例化一个runner
        verbosity=0 (静音模式)： 极其简略，只在所有测试完成后给你一个总结果。
        verbosity=1 (默认模式)： 每个用例成功打印一个 .，失败打印一个 F，报错打印一个 E。
        verbosity=2 (详细模式 ⭐强烈推荐)： 打印每个测试类的名字、方法名以及各自的执行结果（如 ok 或 FAIL），排查问题时最方便。
    `runner.run(suite)`

    ```python
    import unittest

    # 假设这是你的两个测试用例文件（实际开发中通常在不同文件中）

    # TestLogin.py
    import unittest
    class TestLogin(unittest.TestCase):
        def test_login_success(self):
            self.assertTrue(True)

    # TestCart.py
    import unittest 
    class TestCart(unittest.TestCase):
        def test_add_item(self):
            self.assertEqual(1, 1)
    
    # run_tests.py
    import unittest
    # 1. 从不同的文件中导入测试类
    from tests.test_login import TestLogin
    from tests.test_cart import TestCart

    if __name__ == '__main__':
        # 2. 初始化测试套件
        suite = unittest.TestSuite()
        suite.addTest(TestLogin('test_login_success'))
        # 或者把整个类的用例都加进去
        suite.addTest(unittest.makeSuite(TestCart))
        # 3. 运行
        runner = unittest.TextTestRunner(verbosity=2) # verbosity=2 可以看到更详细的执行结果
        runner.run(suite)
    ```

4. 补充: 当你的项目越来越大，有几十个文件、几百个测试用例时，手动一个个去 import 和 addTest 会让人崩溃。unittest 提供了一个强大的工具叫 TestLoader（测试加载器），其中的 discover() 方法可以自动扫描指定目录，把所有符合命名规则的测试文件自动打包成一个 TestSuite，根本不需要你手动写 import。
    `loader = unittest.TestLoader()`    引入加载器
        **A 按“测试类”加载   - 需要提前import 测试类**
        `cases = loader.loadTestsFromTestCase(测试类名)` 将类里的所有测试用例打包成一个用例包
        `suite.addTests(cases)` 向套件中添加整个类下的所有测试用例 - 注意addTest**s**
        **B 按“文件/模块”加载  - 需要提前import 测试文件/模块**
        `cases = loader.loadTestsFromModule(模块名)` 将文件/模块里的所有测试用例打包成一个用例包
        `suite.addTests(cases)` 向套件中添加整个类下的所有测试用例 - 注意addTest**s**
       **C 终极自动化结构：discover 盲盒自动扫描（框架绝配 🚀） - 不需要手动 import 任何测试类**
        `unittest.defaultTestLoader.discover(start_dir, pattern)` - discover() 会在后台自动完成：扫描文件夹 -> 识别 __init__.py -> 导入匹配的文件 -> 提取测试类 -> 组装成 TestSuite 的全套动作。 所以不需要手动将用例塞入suite
            `- start_dir: 目标文件夹路径`
            `- pattern: 匹配规则（通常为 'test_*.py'）`


### 核心概念四： Test Fixture（测试脚手架/固件）：
1. Test Fixture（测试固件）：Test Fixture 就是指“在测试用例执行之前，为你准备好测试环境；并在测试结束之后，帮你清理干净现场”的机制。对应的函数是 setUp() 和 tearDown()。
            前置（测试前）： 浏览器都没打开、用户都没登录、数据库都没连接，你怎么发帖？
            后置（测试后）： 测完之后，测试生成的垃圾帖子如果不删掉，数据库迟早会爆掉；浏览器不关闭，内存迟早会卡死。

2. Python 的 unittest 框架中，提供了四个不同粒度的 Fixture 钩子函数（Hook），它们根据执行的频次和范围划分为两个层级：
    * 级别一：方法级（每条用例都跑一遍）
        `setUp()：`在每一条测试用例执行之前都会被自动调用。
        `tearDown()`：在每一条测试用例执行之后都会被自动调用。
        适用场景： 比如测登录，每测一种账号密码组合，都要重新刷新一下页面。

    * 级别二：类级（整个文件/类只跑一遍）
        `setUpClass()`：在整个测试类开始之前执行一次（必须用 @classmethod 装饰）。
        `tearDownClass()`：在整个测试类所有用例测完之后执行一次（必须用 @classmethod 装饰）。
        适用场景： 比如连接数据库、打开浏览器。我们不需要每测一个功能都重新启一次浏览器（太慢了），只需要开始时打开，全部测完再关闭。

    ```python
    import unittest
    class TestECommerce(unittest.TestCase):
        @classmethod
        def setUpClass(cls):
            print("\n=== [类级前置] 启动浏览器，连接测试数据库（只执行一次） ===")
        @classmethod
        def tearDownClass(cls):
            print("=== [类级后置] 关闭浏览器，断开数据库连接（只执行一次） ===")
        def setUp(self):
            print("--- [方法前置] 打开网店首页，清理浏览器 Cookie ---")
        def tearDown(self):
            print("--- [方法后置] 退出当前登录账号，回到首页 ---")
        # 具体的测试用例 1
        def test_buy_apple(self):
            print("    👉 正在测试：购买苹果功能...")
            self.assertTrue(True)
        # 具体的测试用例 2
        def test_buy_banana(self):
            print("    👉 正在测试：购买香蕉功能...")
            self.assertTrue(True)
    if __name__ == '__main__':
        unittest.main()
    ```


## 三、 跳过测试与预期失败
1. 跳过测试（Skipping Tests）：跳过测试的核心逻辑是：“我知道这个测试现在不该跑，所以请直接无视它，不要报错。”
基础语法：unittest 提供了三种主要的装饰器来实现跳过：
    `@unittest.skip(reason)`: 无条件跳过,当你发现某个测试用例有严重问题，或者对应的功能正在大规模重构时，可以先把它无条件跳过。 - reason（字符串，必须写明跳过的原因，方便日后排查）。 
        ```python
        @unittest.skip("由于后端接口重构，该测试暂时屏蔽，预计下周恢复")
        def test_legacy_api(self):
            response = my_api_call()
            self.assertEqual(response.status_code, 200)
        ```    
    `@unittest.skipIf(condition, reason)`: 当条件为真（True）时跳过。这在处理跨平台兼容性或环境依赖时非常有用。
        ```python
        import sys
        # 如果当前是 Windows 系统，就跳过这个测试（可能该功能只在 Linux 上支持）
        @unittest.skipIf(sys.platform.startswith("win"), "Linux 独占功能，Windows 下无法运行")
        def test_linux_permissions(self):
            # 执行只有 Linux 支持的路径或权限操作
            pass
        ```
    `@unittest.skipUnless(condition, reason)`:与 skipIf 相反，只有条件为真（True）时才运行，条件为假时跳过。通常用于检查前置依赖（如数据库、环境变量）。
        ```python
        import os
        # 只有当环境变量中配置了数据库 URL 时，才运行这个测试
        @unittest.skipUnless(os.environ.get("DATABASE_URL"), "未检测到 DATABASE_URL 环境变量，跳过数据库测试")
        def test_db_connection(self):
            # 连接数据库的测试
            pass
        ```
2. 预期失败（Expected Failure）：预期失败的核心逻辑是：“我知道这个测试一定会失败，但我现在不想修复它（或者它是一个已知的 Bug），请帮我标记出来，只要它失败了，就不算我测试不通过。”
基础语法：unittest提供了一个装饰器
    `@unittest.expectedFailure`
    使用场景：
    A 测试驱动开发（TDD）：你先根据需求写好了测试用例，但实际业务代码还没开始写。此时把测试标记为“预期失败”，等代码写完后再把装饰器去掉。
    B 已知的 Bug 记录：线上有一个不紧急的 Bug，已经提交到了 Issue 系统。你可以写一个测试用例复现这个 Bug，并标记为预期失败，防止 CI/CD 流程因为这个已知 Bug 而中断挂起。

        ```python
        class TestBugTracking(unittest.TestCase):

        @unittest.expectedFailure
        def test_coupon_overflow_bug(self):
            # 这是一个已知的 Bug：当优惠券金额大于订单金额时，系统会变成负数（预期应该报错或截断为0）
            # 此时开发还没修复这个 Bug，所以运行下面这行一定会断言失败
            self.assertEqual(calculate_total(price=10, coupon=20), 0)
        ```
    运行结果表现:
    XFAIL (Expected Failure)：测试真的失败了。
        结果：符合预期！系统认为你是对的，测试报告中会标记为 x，整个测试流程依然算作通过（PASSED）。
    XPASS (Unexpected Success)：测试竟然成功通过了。
        结果：不符合预期！因为你以为它会失败，结果它过了，说明可能有人偷偷把 Bug 修复了，或者你的断言写错了。测试报告中会标记为 u。在某些严格的测试框架配置下，XPASS 会被视作一种警告或隐性错误，提醒你该清理这个装饰器了。    


## 四、 编写单元测试的黄金法则（BEST 准则）
B - Bite-sized（粒度小）：一个测试用例只测一个核心逻辑。如果一个测试里写了 10 个断言，前面失败了后面就不会执行，很难排查。
E - Independent（独立性）：测试用例之间绝对不能有依赖关系！test_b 不能依赖 test_a 写入的数据。测试应该可以以任意顺序独立运行。
S - Repeatable（可重复）：无论运行多少次，在谁的电脑上运行，结果都应该是一样的。避免在测试中使用依赖当前时间的动态数据（除非使用 Mock 技术）。
T - Thorough（周密性）：不仅要测试“正常的输入”（开心路径），更要测试边界值（如空值、极大值、极小值）和异常输入（如类型错误、除以零）。

## 补充：
依赖注入（Dependency Injection）： 在实际的项目开发中，这是一种非常经典且推荐的设计模式
utils.py工具箱：只定义函数，不建立连接
test_utils.py测试员：连接内存数据库，测试工具箱
main.py（正式员工：连接真实数据库文件，真正运行业务）

## 五、 练习
1. 为字符串工具函数编写测试
    创建 string_utils.py，包含两个函数：
    * to_upper(s)：将字符串转为大写并返回。
    * is_palindrome(s)：检查字符串是否为回文（忽略大小写），返回布尔值。
    创建 test_string_utils.py，使用 unittest 编写以下测试用例：
    * 测试 to_upper 传入英文小写字符串，断言返回全大写。
    * 测试 to_upper 传入已含大写的字符串，断言保持大写。
    * 测试 is_palindrome 传入一个回文串（如 "Racecar"），断言为 True。
    * 测试 is_palindrome 传入非回文串，断言为 False。
    * 使用 setUp 存储常用测试字符串，避免在每个方法内重复定义。
2. 测试数据库操作
    新建 db_utils.py，包含以下函数（使用 sqlite3）：
    * create_table(db_name)：创建一个 scores 表（id, name, score）。
    * insert_score(db_name, name, score)：插入一条分数记录。
    * get_average_score(db_name)：查询所有分数的平均值并返回。
    创建 test_db_utils.py，编写测试用例：
    * 在 setUp 中使用内存数据库（:memory:）保证测试隔离，并创建表。
    * 测试 insert_score 插入一条记录后，查询表行数是否为 1（提示：执行 SELECT COUNT(*) FROM scores 进行断言）。
    * 测试 get_average_score：插入三条分数（如 80, 90, 85），断言平均值近似于 85.0。
    * 在 tearDown 中关闭数据库连接。
提示：使用内存数据库 sqlite3.connect(':memory:') 不会产生实际文件，非常适合单元测试，且速度更快。记得在每个测试方法前重新建表，确保数据独立。



````python
#string_utils.py
def to_upper(text):
    new_text = text.upper()
    return new_text

def is_palindrome(text):
    clean_text = text.lower
    return clean_text == clean_text[::-1]

#test_string.py

import unittest
from string_utils import to_upper, is_palindrome

class test_string_utils(unittest.TestCase):

    def setUp(self):
        self.upper_text = "upgrade"
        self.upper_result = "UPGRADE"
        self.mixedtext = "PyThon"
        self.mixedresult = "PYTHON"
        self.palindrome_text = "Noon"
        self.non_palindrome_text = "hello"

    def test_to_upper_lowercase(self):
        self.assertEqual(to_upper(self.upper_text),self.upper_result)

    def test_to_upper_mixedtext(self):
        self.assertEqual(to_upper(self.mixedtext), self.mixedresult)

    def test_is_palindrome_True(self):
        self.assertTrue(is_palindrome(self.palindrome_text))
    
    def test_is_palindrome_False(self):
        self.assertFalse(is_palindrome(self.non_palindrome_text))


if __name__ == "__main__":
    unittest.main()
````

````python
#db_utils.py

def create_table(cursor):
    # 修正了 EXISTS 的拼写
    sql_sentence = "CREATE TABLE IF NOT EXISTS scores(id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT NOT NULL, score INTEGER)"
    cursor.execute(sql_sentence)

def insert_score(cursor, conn, name, score):
    # 使用占位符 ? 防止 SQL 注入
    sql_sentence = "INSERT INTO scores (name, score) VALUES (?, ?)"
    cursor.execute(sql_sentence, (name, score))
    conn.commit()

def give_average_score(cursor):
    sql_sentence = "SELECT AVG(score) FROM scores"
    cursor.execute(sql_sentence)
    result = cursor.fetchone()[0]
    return result if result is not None else 0.0  # 如果没数据，返回 0.0

#test_db_utils.py
import unittest
import sqlite3
from db_utils import create_table, insert_score, give_average_score


class TestDBUtils(unittest.TestCase):

    def setUp(self):    
        self.conn = sqlite3.connect(':memory:')
        self.cursor = self.conn.cursor()
        create_table(self.cursor)
        
    def tearDown(self):  
        self.cursor.close()
        self.conn.close()
    def test_insert_score(self):
        insert_score(self.cursor, self.conn, '张三',98)
        result = self.cursor.execute('''SELECT COUNT(*) FROM scores''')
        self.assertEqual(result.fetchone()[0],1)

    def test_give_average_score(self):
        insert_score(self.cursor, self.conn, '张三',80)
        insert_score(self.cursor, self.conn, '李四',90)
        insert_score(self.cursor, self.conn, '王五',85)
        self.assertEqual(give_average_score(self.cursor), 85)

if __name__ == "__main__":
    unittest.main()
````

# 🐍 Python 学习笔记｜第 38.5 关：终端命令基础
## 一、核心概念
1. 终端 (Terminal)： 接收输入并显示输出的窗口程序（如 macOS 的 Terminal、Windows 的 PowerShell/Windows Terminal、Linux 的 Terminator）。
2. 外壳程序 (Shell)： 终端内部运行的命令解析器。它负责把你的文本命令翻译成操作系统能听懂的机器语言。常见的 Shell 有 Bash（大多数 Linux 的默认设置）和 Zsh（macOS 的默认设置）。
3. 工作目录（Working Directory）：终端总是停留在某一个具体的文件夹里，你输入的所有命令默认都会对这个文件夹里的文件生效。这个文件夹就叫“当前工作目录”。
4. 绝对路径： 从根目录 / 开始的完整路径，例如 /Users/username/Downloads。
   相对路径： 相对于你当前所在位置的路径。`.` 代表当前目录，`..` 代表上一级目录。例如 cd ./documents。
5. 命令语法结构： 绝大多数终端命令都遵循以下标准格式：`Command [-Options] [Arguments]`
    * Command（命令）： 核心动作（如 ls 列表）。
    * Options（选项/参数）： 修正命令的行为，通常以 -（短选项）或 --（长选项）开头（如 -l 表示长格式）。
        | 特性 | 短选项 (Short Options) | 长选项 (Long Options) |
        | :--- | :--- | :--- |
        | **开头符号** | 单横杠 `-` | 双横杠 `--` |
        | **长度** | 通常只有 1 个字符 | 完整的英文单词或短语 |
        | **能否组合** | 能（如 `-abc`） | 不能（必须独立写 `--foo --bar`） |
        | **参数传递** | 空格或紧跟（如 `-p8080` 或 `-p 8080`） | 通常用等号或空格（如 `--port=8080`） |
        | **主要优点** | 打字快，省时 | 易读性强，不容易忘 |
        | **适用场景** | 日常在终端中手动敲命令 | 编写自动化脚本、写教程文档 |

        有参数值和没有参数值
            没有参数值的（单字母）：它是一个“开关” ，在编程里，这相当于一个 布尔值（True / False）。
            有参数值的：它是一个“填空题” ，这种参数通常是一个配置项，终端只知道你要开启这个功能，但它猜不到你想怎么配置它，所以你必须在后面补上一个具体的值。

    * Arguments（参数/对象）： 命令操作的对象，比如文件名或文件夹名。

```Bash
python -m cProfile -s tottime my_script.py
#python：调用 Python 解释器。
#-m cProfile：
    #-m 参数表示“运行一个内置模块（module）”。
    #cProfile 是 Python 标准库中用 C 语言编写的性能分析模块（因为是用 C 写的，所以它自身的运行开销极小，测得准）。
#-s tottime：
    #-s 代表 Sort（排序）。
    #tottime 代表 Total Time（总耗时）。意思是：让分析结果按照“哪个函数自身消耗的总时间最多”从大到小排序。
# my_script.py：这是你的操作对象，即你想要分析的 Python 脚本文件名。
```



## 二、基础核心命令分类
1. 目录切换与路径导航：在终端中，你随时处于某一个具体的文件夹内，这个文件夹被称为当前工作目录。
    `pwd` (Print Working Directory)： 显示当前所在的绝对路径。
    `ls` (List)： 列出当前目录下的文件和文件夹。
    `cd` (Change Directory)： 切换目录。
        `cd /` ：进入根目录。
        `cd ~ `或只打 `cd`：回到当前用户的主目录（Home）。
        `cd ..`：返回上一级目录（父目录）。
        `cd -` ：回到上一次所在的目录。
2. 文件与目录操作（增、删、改、查）
    增：
    `mkdir [目录名]` (Make Directory)： 创建新文件夹。
        `mkdir -p a/b/c` ：进阶，可以一次性创建多层嵌套的目录。
    `touch [文件名]`： 创建一个空文件（如果文件已存在，则更新其修改时间）。
    改：
    `cp [源路径] [目标路径]`：(copy) 复制文件或目录。
        `cp -r [源路径] [目标路径]`：复制文件夹时 必须加上 -r 选项（表示递归复制）
    `mv [源路径] [目标路径]` ：(Move) 移动文件或重命名。
        移动：`mv file.txt /path/to/destination/`
        重命名：`mv old.txt new.txtrm [文件名]`
    删：
    `rm [文件名]`：(Remove) 删除文件。警告： 终端删除文件不经过回收站，是永久删除！
        `rm -rf [文件名]`：删除文件夹需要加 -rf（-r 递归，-f 强制不提示）：。使用时请务必极度小心，确认路径无误。
    查：
    `ls [路径]`：(List) 查看目录下的内容。
        `ls -l [路径]`：以列表形式显示详细信息（权限、大小、修改时间）。
        `ls -a [路径]`：显示所有文件，包括以 . 开头的隐藏文件。
    `cat [文件名]`：(Concatenate) 一次性在终端查看文件的全部内容（适合短文件）。
    `less [文件名]`： 分页查看大文件。进入后，使用 Space 翻页，↑ / ↓ 箭头滚动，输入 /关键词 进行搜索，按 q 退出。
    `head -n [行数] [文件名] \ tail -n [行数] [文件名]`：查看文件的前几行或后几行（常用于看日志）。
        `tail -f access.log`：高频用法，可以实时动态滚动查看日志文件的新增内容，按 Ctrl + C 退出。
    `find [路径] -name "[文件名]"`： 查找文件。例如在当前目录找 txt 文件：find . -name "*.txt"。

3. 系统状态与管理
    `clear（或快捷键 Ctrl + L）`： 清屏，让终端界面恢复干净。
    `top 或 htop`： 实时查看系统的进程、CPU、内存占用情况（相当于任务管理器）。按 q 退出。
    `ps -ef 或 ps aux`： 查看当前系统运行的所有进程静态快照。
    `kill [PID]`： 结束某个进程。PID 是进程的 ID 数字，可以通过 ps 或 top 查到。
        `kill -9 [PID]`：如果无法关闭，强制杀死进程。

## 三、 终端操作的“终极效率技巧”
1. 掌握以下几招，可以让你在键盘上飞起来：
    `Tab 键（自动补全）`： 最重要、最常用的键！ 输入命令或文件名的前几个字母，按下 Tab 键会自动补全。如果有多个匹配项，连按两下 Tab 会列出所有可能。
    `方向键 ↑ / ↓（历史记录）`： 快速翻阅你之前输入过的命令。
    `Ctrl + C（强行终止）`： 如果某个命令卡住了、或者在持续运行（如 tail -f），按下 Ctrl + C 可以直接中断该进程，退回到等待输入的状态。
    `history 命令`： 列出你过去输入过的成百上千条命令历史。你可以用 !编号 快速重复执行历史中的某条命令。
    `man 或 --help`：万能帮助手册，不知道一个命令怎么用？输入 man ls（Manual）即可查看该命令的官方详细说明书。大部分命令也支持 ls --help 或 ls -h 来查看简要的参数说明。

## 四、 进阶管道符与重定向（初学者必知）
1. 终端真正强大的地方在于，它可以把多个简单的命令组合在一起，完成复杂的任务。
    `|`：管道符,把前一个命令的输出，作为后一个命令的输入。
        示例： ls -l /bin | grep "sh" （列出 /bin 目录下的所有文件，并从中筛选出包含 "sh" 字符的行）。grep 是极其强大的文本搜索工具。
    `> 与 >>`：重定向, 把原本要输出到屏幕上的内容，写入到文件里。
        `>`（覆盖）：echo "hello" > test.txt （如果文件有内容，会被覆盖）。
        `>>`（追加）：echo "world" >> test.txt （在文件末尾另起一行追加内容）。

## 一、程序调试
1. 定义：程序调试就是查找并修复代码中的错误。Python 提供了内置工具 pdb（Python 调试器）帮助我们单步追踪代码运行。
2. pdb（Python Debugger调试器）
    传统
    `breakpoint()` :当代码运行到 breakpoint() 时，程序会暂停，**终端**会出现 (Pdb) 交互式命令行，允许你实时查看和修改变量。
        进入 pdb 环境后，你可以使用以下快捷字母来控制代码执行：
        | 命令 | 全称 | 作用 |
        | :--- | :--- | :--- |
        | **n** | next | 单步执行，运行下一行代码（不进入调用的函数内部）。 |
        | **s** | step | 单步进入，如果下一行有函数调用，会跳进该函数内部。 |
        | **c** | continue | 继续运行，直到遇到下一个断点或程序结束。 |
        | **p** | print | 打印变量，例如 `p total` 打印变量值。 |
        | **pp** | pretty print | 美化打印，适合查看复杂的字典或列表。 |
        | **l** | list | 列出源码，显示当前断点周围的 11 行代码。 |
        | **w** | where | 查看调用栈，显示当前代码是如何被一层层调用进来的。 |
        | **q** | quit | 退出调试，直接终止程序。 |

        ```python
        def calculate_average(numbers):
            total = sum(numbers)
            # 在这里插入断点
            breakpoint() 
            count = len(numbers)
            return total / count

        calculate_average([10, 20, 0, 40])
        ```
    现代
    现代 IDE（IDE 是 集成开发环境（Integrated Development Environment））（如 PyCharm, VS Code）提供了非常好用的图形化调试界面，其底层逻辑与 pdb 是一致的。
    第一步：点亮红点（打断点）
    第二步：开启调试模式（关键！）  .py文件中使用F5 .ipynb文件中使用代码块左上运行旁边的下拉箭头选择“调试单元格” (Debug Cell)
    第三步：掌控全局（调试控制台）
        代码行变黄：当前暂停的那一行代码会被黄色高亮显示。
        顶部出现调试控制面板：界面上方会浮现出一个有 6 个按钮的小工具栏。
            ▷ 继续 (Continue / F5)：直接放行，让代码继续往下跑，直到遇到下一个红点，或者程序结束。
            ↷ 单步跳过 (Step Over / F10)：最常用。让代码往下走一行。如果这一行是个函数，它不会跳进函数内部，而是直接拿到结果走到下一行。
            ↓ 单步调试 (Step Into / F11)：如果当前行是一个你自己写的函数，点击它会钻进这个函数内部第一行。
            ↑ 单步跳出 (Step Out / Shift+F11)：如果你钻进了一个函数，觉得看够了，点这个可以立刻执行完该函数剩下的代码，并跳回到外面。
            ↻ 重启 (Restart / Ctrl+Shift+F5)：重新从头开始调试。
            ⬜ 停止 (Stop / Shift+F5)：彻底退出调试。

    隐藏福利：查看变量
    悬停鼠标：把鼠标指针放在代码中的任何变量上，会直接弹出一个小气泡显示它当前的值。
    左侧变量面板：在 VS Code 左侧的 VARIABLES（变量）窗口里，能看到当前作用域下的所有局部变量和全局变量，甚至包括复杂的列表和字典，点开就能看结构。

## 二、性能优化
1. 定义：在代码能正确运行的基础上，提升执行速度和资源使用效率。

2. Python 提供了两个内置的剖析模块：profile 和 cProfile。
    * `profile`：纯 Python 实现，开销较大，会显著拖慢程序运行。
    * `cProfile`：基于 C 语言扩展实现，开销非常小，适合绝大多数生产和开发环境。
        cProfile输出结果：
        | 列名 | 全称 | 含义 |
        | :--- | :--- | :--- |
        | **ncalls** | Number of calls | 函数被调用的次数。如果出现 `3/1`，表示有递归调用，总共调用 3 次，其中 1 次是原生（顶级）调用。 |
        | **tottime** | Total time | 在该函数中消耗的总时间（**不包括**它调用的子函数的执行时间）。 |
        | **percall** | Per call | `tottime / ncalls`，即单次调用该函数平均消耗的时间（不含子函数）。 |
        | **cumtime** | Cumulative time | 该函数及其所有子函数消耗的累计总时间。**这是寻找性能瓶颈最常看的指标。** |
        | **percall** | Per call | `cumtime / 顶级ncalls`，单次调用该函数及其子函数平均消耗的时间。 |
        | **filename:lineno(function)** | Location | 函数所在的文件名、行号和函数名。 |

3. Python的内置模块pstats：经常配合性能分析工具 cProfile（或 profile）使用
    * `pstats`: pstats是 Python 标准库中的一个模块，全称为 Profile Statistics。cProfile 负责在程序运行时收集性能数据并将其保存为二进制文件，而 pstats 则负责解析、过滤、排序和打印这些数据，让你能直观地看到代码中哪一部分最耗时。
    使用 pstats 的典型工作流分为两步：
            采样/分析： 使用 cProfile 运行代码并导出数据。
            解读/统计： 使用 pstats 加载并分析数据。

4. 原则： 高德纳（Donald Knuth）的名言："过早的优化是万恶之源（Premature optimization is the root of all evil）。"
    性能优化的正确三部曲是：
        先让它跑起来（Make it work）：保证功能正确。
        让它变正确（Make it right）：补充测试，修复 Bug，优化架构。
        让它跑得快（Make it fast）：通过数据定位瓶颈，而非凭空猜测。

5. 调用方式 

    A 命令行直接运行（最常用、无侵入）
        基础调用命令：
        `python -m cProfile your_script.py` 这会在终端直接打印出所有的性能统计数据。不过，如果项目稍微大一点，输出会刷屏，很难阅读。
            * -s (Sort)：指定排序规则。 最常用的是按累计时间（cumtime）或函数自身运行时间（tottime）排序。（calls）
                `python -m cProfile -s cumtime your_script.py` 按累计时间排序（看看哪个函数及其子函数耗时最长）
                `python -m cProfile -s tottime your_script.py` 按函数自身耗时排序（看看哪个函数自身内部的代码最慢）
            * -o (Output)：将分析结果保存到文件中。 配合可视化工具使用时必选。
                `python -m cProfile -o program.prof your_script.py` 
                保存完文件以后可以配合`pstats模块`进行结果筛选与排序:
                `python -m pstats program.prof`
                进入后，你会看到提示符 program.prof%，此时你可以输入各种命令：
                    `sort cumulative`：按累计时间排序（找哪条调用链路最慢）。
                    `sort tottime`：按函数自身耗时排序（找哪个具体函数最榨干 CPU）。
                    `sort ncalls`：按调用次数排序（看有没有频繁调用的不合理情况）。
                    `stats [数字]`：打印前 N 行数据。例如输入 stats 10 显示前10行。
                    `callees [函数名]`：查看该函数调用了哪些子函数。
                    `callers [函数名]`：查看该函数被哪些父函数调用了（追溯来源）。

    B 在代码中局部嵌入（精准定位）：如果你只想分析代码中的某一个特定片段（例如某个复杂的函数或循环），而不是整个脚本，可以在代码中显式导入并使用 cProfile。
        方法1（基础）：
            `import cProfile`                           导入cProfile模块
            `def function()`                            定义函数
            `profiler = cProfile.Profile()`             创建性能分析器对象（profiler对象其实已经将数据写到了内存里）
            `profiler.enable()`                         启动分析
            `function()`                                执行你要测试的代码
            `profiler.disable()`                        停止分析
            `profiler.print_stats(sort='cumulative')`   打印报告
        方法2（傻瓜式）：
            `import cProfile`
            `def function()`
            `cProfile.run("function()")` cProfile.run() 默认的行为是直接在控制台（终端）打印出来。不过，你可以通过向它传递参数，让它把分析结果保存到文件中。
        方法3（进阶）：使用 with 上下文管理器（推荐）
            `import cProfile`
            `with cProfile.Profile() as profiler:`
            `function()` 
            `profiler.print_stats(sort='cumulative')`
        高级方法：保存数据后配合pstats模块进行结果筛选与排序
            `profiler.dump_stats(filename)`  使用cProfile将数据保存到文件中（如果不保存那只是在profiler对象里，暂时存入了内存）
            `import pstats`
            构造函数
            `stats = pstats.Stats(*filenames, stream=sys.stdout)`加载二进制性能数据文件
                `stats = pstats.Stats(profiler)`如果没有存文件可以直接将profiler中的数据copy到stats对象身上
                `filenames`: 可以传入一个或多个文件名，pstats 会自动将它们的数据合并（Merge）在一起。
                `stream`: 指定输出流，默认是控制台。你可以传入一个打开的文件对象，将结果直接写入文件。
            排序
            `stats.sort_stats(*keys)`改变数据的排列顺序
                (`cumtime`/`tottime`/ `ncalls`/`nfl` / `name` / `file` / `line`: 分别按“累计时间/函数自身耗时/被调用次数/姓名/文件/行号”排序、按函数名、按文件名、按行号排序)。sort_stats 支持多级排序。例如 stats.sort_stats('time', 'calls') 表示先按自身耗时倒序，如果耗时相同，再按调用次数倒序。
            数据清洗
            `stats.strip_dirs()`数据清洗。
                默认情况下，性能报告会显示文件的绝对路径（如 /usr/local/lib/python3.10/site-packages/.../foo.py），这会导致控制台表格极长、极难阅读。
                调用 strip_dirs() 后，会移除所有文件名前面的路径，只保留文件名（如 foo.py），让输出变得清爽。
            控制台数据输出 (Output)
            `stats.print_stats(*restrictions)`打印主要的性能统计表格（即包含 ncalls, tottime, cumtime 的表格）
                `restrictions` (限制参数): 可以传入整数或字符串，用于过滤输出。
                    传入整数：限制只打印前 $n$ 行。例如 print_stats(10)。
                    传入浮点数 (0.0 ~ 1.0)：限制只打印前百分之几的数据。例如 print_stats(0.1) 打印前 10%。
                    传入字符串/正则表达式：只打印函数名或文件名包含该特征的行。例如 print_stats('my_project')。
            `stats.print_callers(*restrictions)`打印调用者（父函数）列表。
            `print_callees(*restrictions)`打印被调用者（子函数）列表。

        提效技巧：pstats的链式调用：pstats.Stats 的方法（如 strip_dirs(), sort_stats(), print_stats()）全都支持链式调用。

6. 补充知识点：
    **三元运算符（Conditional Expression）**，或者叫一行流的 if-else
    口诀：结果1 if 条件 else 结果2

    传统
    ```python
    if cc == nc:
        ncalls = f"{nc}"
    else:
        ncalls = f"{nc}/{cc}"
    ```
    语法糖
    ```python
    ncalls = f"{nc}" if cc == nc else f"{nc}/{cc}"
    ```

2. 性能分析对比
    创建 profile_demo.py，实现两种生成“斐波那契数列前N项列表”的方法：
    ```python
    def fib_list(n):
        fibs = []
        a, b = 0, 1
        for _ in range(n):
            fibs.append(a)
            a, b = b, a + b
        return fibs

    def fib_recursive(n, fibs=None):
        if fibs is None:
            fibs = []
        if n == 0:
            return fibs
        if len(fibs) < 2:
            fibs.append(len(fibs))
        else:
            fibs.append(fibs[-1] + fibs[-2])
        return fib_recursive(n-1, fibs)
    ```
    * 使用 cProfile.run() 分别分析两个函数在 n=1000 时的性能，打印出 ncalls 和 tottime 的差异。结合分析结果，在代码注释中写下你认为哪种实现更好以及原因。

提示：递归方法会产生大量函数调用开销，在结果中会明显体现。你还可以用 timeit 比较两者的执行速度，验证你的判断。

````python
#profile_demo.py
import cProfile

def fib_list(n):
        fibs = []
        a, b = 0, 1
        for _ in range(n):
            fibs.append(a)
            a, b = b, a + b
        return fibs

def fib_recursive(n, fibs=None):
    if fibs is None:
            fibs = []
    if n == 0:
        return fibs
    if len(fibs) < 2:
        fibs.append(len(fibs))
    else:
        fibs.append(fibs[-1] + fibs[-2])
    return fib_recursive(n-1, fibs)

cProfile.run("fib_list(1000)")
#调用1004次函数耗时0.001秒
cProfile.run("fib_recursive(1000, fibs=None)")
#调用3064次函数用时0.022秒

#综上所述，第一种写法效率更高

````

# 🐍 Python 学习笔记｜第 40 关：工程规范与依赖管理（pip, venv, Poetry 与虚拟环境最佳实践）
## 一、总揽
1. 当Python项目越来越复杂时，我们会依赖大量第三方库。不同的项目可能需要同一个库的不同版本，如果都装在系统全局环境里，就会产生冲突。就像不同餐厅的厨师不能共用同一个调料架——每个厨房应该有自己独立的一套调料。虚拟环境就是为每个项目创建独立的Python运行环境，配合依赖管理工具可以精确记录和安装所需的库，保证项目在任何机器上都能复现。
    虚拟环境：隔离的项目空间，拥有独立的Python解释器和库。
    依赖管理：记录项目用了哪些库及其版本，常用requirements.txt或pyproject.toml。
    包管理器：pip 是Python默认的包安装工具，Poetry 等现代工具将依赖管理与虚拟环境一体化。

## 二、基础概念
1. 全局环境 vs. 虚拟环境
    * **全局环境（大厨房）**：电脑安装 Python 后的主厨房。所有项目共用，容易把“调料”（第三方库）搞混，导致环境崩溃。
    * **虚拟环境 (venv)（私密小厨位）**：
        * **比喻**：用纸板在大厨房里隔出来的专属空间。
        * **意义**：项目 A 用老抽（旧版库），项目 B 用生抽（新版库），互不干扰。

2. 库、包、pip
    * **库 (Library)**：功能的集合。例如“盖城堡”或“绘图”的代码逻辑。
    * **包 (Package)**：代码的物理形态。在电脑里表现为一个文件夹。
    * **pip（快递员/钱包管理器）**：
        * **本质**：运行在命令行里的工具软件。
        * **任务**：你下单 `pip install`，它负责从 **PyPI**（全球中央大仓库）把你需要的“盒子（包）”搬回你的小厨位。

3. 命令行生存：地图与导航 (CLI & Path)
    环境变量 PATH
    * **比喻**：操作系统的“快速搜索清单”。
    * **原理**：当你输入一个指令（如 `python` 或 `adb`）时：
        1. 电脑先看脚下（当前所在的文件夹）。
        2. 如果没找到，再翻清单（**PATH**）。
        3. 清单里有地址，它就直接去那儿执行；清单里没有，就会报错“不是内部或外部命令”。
    * **建议**：把常用工具（如 adb.exe）所在的路径加入 PATH，从此告别手动“指路”。

    路径符号（指路牌）
    * `.` (点)：“就在脚下”，代表当前所在的文件夹。
        * `.\adb`：告诉电脑“别去清单找了，就在这儿跑这个文件”。
    * `..` (双点)：“退回门口”。代表上一层文件夹。


## 二、pip基础操作
1. pip 是Python自带的包安装工具（通常随Python一起安装），连接PyPl仓库下载包。

2. pip基础操作
```bash
#安装包
pip install requests

#安装指定版本
pip install numpy==1.26.0

#卸载包
pip uninstall requests

#列出当前环境已安装的包
pip list

#导出依赖到文件
pip freeze > requirements.txt

#根据文件安装依赖
pip install -r requirements.txt

```

3. requirements.txt内容示例
```text 
requests==2.31.0
numpy>=1.24.0,<1.27.0
pandas==2.1.0
```

## 三、venv创建虚拟环境
1. Python内置的venv模块可以创建轻量级虚拟环境，无需额外安装。

2. 创建、激活和退出虚拟环境：
```bash  
#创建名为.venv的虚拟环境（推荐放在项目根目录）
python -m venv .venv

#windows激活
.venv\Scrips\activate

#macOS激活
source .venv/bin/activate

#退出虚拟环境
deacticate
```
激活后，终端提示符前会出现（.venv）标识。此时用pip安装的包只会存在于该环境中

3. 每个项目都应该创建自己的虚拟环境，不要把虚拟环境目录提交到Git（可将其路径如.venv加入.gitignore）
    * 在项目的根目录下，创建一个名为 .gitignore 的文件（注意前面有一个点 .，且没有任何后缀名）。
    * 用文本编辑器打开它，在里面添加你虚拟环境的文件夹名称。

        .gitignore
        ```Plaintext 
        #忽略Python虚拟环境目录
        #路径最后的斜杠 / 表示忽略整个文件夹。只要这一行存在，Git 就会对该目录视而不见。
        .venv/
        venv/
        env/
        ENV/
        ```
4. 在项目根目录保留 requirements.txt 或 pyproject.toml 记录依赖。

## 四、Poetry：现代化依赖管理
1. Poetry是集依赖管理、打包、发布于一体的工具，用pyproject.toml统一配置项目，并自动管理虚拟环境。

2. 安装Poetry（推荐用官方脚本）
```bash
#Windows(Powershell)
(Invoke-Webrequest -Uri http://install.python-poetry.org -UseBasicParsing).Content | python -
    #利用了“管道符（|）”，把“下载”和“执行”两个步骤串联在了一起。
    #( 蓝色部分：下载脚本 ) | ( 橙色部分：运行安装 )
        #前半部分：下载安装脚本
            #Invoke-WebRequest：这是 PowerShell 中用来发送 HTTP/HTTPS 请求的内置命令，类似于 Linux 中的 curl 或 wget。它的作用是访问指定的网址。
            #-Uri https://install.python-poetry.org：-Uri 参数指定了网络地址。这个 URL 是 Poetry 官方提供的安装脚本托管地址，里面全都是用来配置和安装 Poetry 的 Python 代码。
            #-UseBasicParsing：这是一个兼容性参数。它告诉 PowerShell ：“用基础的方式去解析这个网页就好，不要调用 Internet Explorer（IE）的引擎。” 这样即使在没有配置过 IE 的服务器或精简版 Windows 系统上，命令也能顺利运行。
            #( ... ).Content：外面的括号表示先执行括号内的下载命令，最后的 .Content 则是关键——它表示只提取下载下来的“文本内容”（即具体的 Python 代码本身），而不是一个复杂的HtmlWebResponseObject（网页响应对象）里面打包了各种网络传输的元数据。
        
        #中间纽带：管道符
            # |（管道符）：它的核心作用是“接力”。它把前半部分下载得到的脚本内容（纯文本代码），像流水线一样直接传送给后半部分的程序去处理。整个过程在内存中完成，不会在你的电脑上留下一个临时安装包。

        #后半部分：执行安装
            # python：调用你电脑里已经安装好的 Python 解释器。
            # -（连字符/减号）：这是这行命令中最精妙的地方。 在命令行中，python -（注意后面有一个空格和减号）意思是：“Python 啊，你不要去硬盘上找 .py 文件运行了，直接从管道（标准输入）里接收别人传给你的代码并立即执行吧！”
            #python install.py（找文件运行）：Python 接收到路径后，会亲自去硬盘里打开（Open）这个文件，读取（Read）里面的代码，然后执行。
            #... | python -（从管道接收）：Python 根本不去碰硬盘，它张开嘴（标准输入），等着前面的命令（PowerShell）把代码通过内存直接倒进它的嘴里，它一边接一边执行。
            # python -c "你的代码"：在终端里直接把一行或几行 Python 代码当场塞给 Python 运行。（c 代表 Command，即命令行代码）。

#macOS/Linux
curl -sSL https://install.python-poetry.org | python3 -
    #( 蓝色：Linux的下载特工 )  | ( 橙色：执行 )
        #前半部分：用最安静、最稳妥的方式，把官网的安装脚本代码抓取到内存中（同样不存盘，只出纯文本代码）。
            #curl 是 Linux 和 macOS 中最强悍的命令行下载工具（相当于 PowerShell 里的 Invoke-WebRequest）。后面那一串大写的 -sSL 是它的“特工隐身与追踪参数”：
                # -s (silent)：静音模式。让 curl 在下载时不要显示进度条、下载速度等杂乱的提示信息，保持终端干净。
                # -S (show-error)：报错提示。虽然是静音模式，但如果网络彻底断了或者网址错了，它还是会把错误信息打印出来，方便你排查。
                # -L (location)：自动重定向。如果 Poetry 官方的网址在服务器后台换了地方（比如从 A 域名跳转到 B 域名），curl 会聪明地跟着跳过去，直到把真正的脚本代码抓下来。
    #后半部分：python3 -
        #python3：在 Mac 和 Linux 系统中，系统往往同时存在 Python 2 和 Python 3。为了防止冲突，这些系统通常用 python3 来明确指定调用新版的 Python 解释器。

```

3. 常用命令
```bash
#创建新项目（自动生成pyproject.toml和虚拟环境）
poetry new my_project（你的项目名称）

#在已有项目中初始化
cd my_project
poetry init

#添加依赖
poetry add requests
poetry add --dev pytest      #开发依赖（如用来写测试的 pytest、用来检查代码格式的 flake8）

#移除依赖
poetry remove requests

#安装所有依赖（根据pyproject.toml 和 poetry.lock）
poetry install

poetry install --without dev #（部署非开发过程实际需要的依赖）
#在虚拟环境中运行脚本
poetry run python main.py

#导出reqirements.txt格式（用于兼容）
poetry export -f requirements.txt --output requirements.txt
```

4. pyproject.toml 示例
```toml
[tool.poetry]
name = "my-app"
version = "0.1.0"
description = ""
authors = ["Your Name <you@example.com>"]

[tool.poetry.dependencies]
python = "^3.9"
requests = "^2.31.0"

[tool.poetry.dev-dependencies]
pytest = "^7.4.0"

[build-system]
requires = ["poetry-core"]
build-backend = "poetry.core.masonry.api"
```
